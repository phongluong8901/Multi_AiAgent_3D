````python
import os
import time

def check_services():
    print("=== KIỂM TRA TRẠNG THÁI TIẾN TRÌNH ===")
    for process in ['Xvfb', 'xfce', 'x11vnc', 'websockify', 'cloudflared']:
        stream = os.popen(f'pgrep -af {process}')
        output = stream.read().strip()
        if output:
            print(f"[OK] Dịch vụ {process} đang chạy:\n{output}")
        else:
            print(f"[ERR] Dịch vụ {process} CHƯA hoạt động!")

check_services()

print("\n=== LOG CHI TIẾT CÁC DỊCH VỤ ===")
for log_file in ['/tmp/xvfb.log', '/tmp/xfce.log', '/tmp/vnc.log', '/tmp/novnc.log', '/tmp/cf.log']:
    if os.path.exists(log_file):
        print(f"\n--- {log_file} (3 dòng cuối) ---")
        with open(log_file, 'r') as f:
            lines = f.readlines()[-3:]
            print("".join(lines))
    else:
        print(f"{log_file} không tồn tại.")
````

2. Kiến trúc live_evolve_agent.py (Tự sinh tool + Tự sửa tool khi chạy thật)
Luồng chạy khi bạn đặt bất kỳ câu hỏi nào (thời tiết, lịch sử, chính trị, khoa học, cào web...):

Architect (qwq:32b): Quét kho skills/ hiện có, đối chiếu với câu hỏi để quyết định dùng tool cũ hay đặt hàng tạo tool mới.

Tool Forge song song (qwen2.5-coder:32b x3 nhánh + qwq:32b Reviewer): Tự viết pytest, sinh 3 bản code song song trên A100, qwq:32b kiểm duyệt (cấm dữ liệu giả, bắt dùng API mở không cần key), chạy sandbox pass mới lưu vào skills/.

Thực thi & Tự chữa lành (Self-Healing): qwen2.5-coder:32b viết script gọi các tool trong skills/ để lấy dữ liệu thật. Nếu chạy thật bị lỗi (403, sai format, lỗi logic), qwq:32b chẩn đoán xem lỗi do script hay lỗi do file tool -> nếu lỗi do tool thì tự vá thẳng file trong skills/ rồi chạy lại.

Tổng hợp (qwq:32b): Trả lời dựa trên dữ liệu thật kèm nguồn

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import asyncio
import subprocess
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str) -> str:
    m = re.search(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    return m.group(1).strip() if m else text.strip()


def extract_json(text: str) -> dict:
    s, e = text.find("{"), text.rfind("}") + 1
    if s == -1 or e <= s:
        raise ValueError("Khong tim thay JSON")
    return json.loads(text[s:e], strict=False)


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def scan_registry() -> str:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    lines = []
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        try:
            tree = ast.parse(open(f"{SKILLS_DIR}/{fn}", encoding="utf-8").read())
            sigs = []
            for n in tree.body:
                if isinstance(n, ast.ClassDef):
                    ms = [
                        f"{m.name}({', '.join(a.arg for a in m.args.args if a.arg != 'self')})"
                        for m in n.body
                        if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef))
                        and (not m.name.startswith("_") or m.name == "__init__")
                    ]
                    sigs.append(f"class {n.name}: [{', '.join(ms)}]")
                elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                    sigs.append(f"def {n.name}({', '.join(a.arg for a in n.args.args)})")
            lines.append(f"- `{mod}`: " + " | ".join(sigs))
        except Exception:
            lines.append(f"- `{mod}`")
    return "\n".join(lines) if lines else "(Kho dang trong)"


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        return p.returncode == 0, out
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)"


async def forge_tool(client: httpx.AsyncClient, spec: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    # 1. Coder viet pytest + 3 nhanh code song song (tan dung OLLAMA_NUM_PARALLEL=4)
    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
3-5 test case ro rang, mock mang internet khi test unit. Chi tra ve code Python."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
Yeu cau:
- Chi dung thu vien chuan Python, httpx, bs4, ddgs (neu search).
- Dung API mo KHONG CAN API KEY (vi du: Open-Meteo cho thoi tiet, MediaWiki API kem User-Agent cho Wikipedia, Google News RSS/ddgs cho tin tuc).
- Xu ly loi ro rang, tuyet doi khong tra ve du lieu gia (mock/hardcode).
Chi tra ve code Python."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0])
    candidates = [extract_code(c) for c in res[1:]]

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if ok:
                # 2. qwq:32b kiem duyet chong hardcode / dung API that
                rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
{FENCE}python
{curr}
{FENCE}
Kiem tra: Co goi nguon/API that khong hay hardcode du lieu gia? Co bam dung Stub khong?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
                rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
                if "STATUS: PASS" in rev.upper():
                    with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                        f.write(curr)
                    with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                        f.write(test_code)
                    print(f"  [FORGE OK] Da luu `{sname}.py` vao kho skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"  [REVIEW FAIL] Nhanh #{idx+1}: {rev.splitlines()[-1][:100]}")
                    break
            elif attempt < 2:
                fix_p = f"Sua `{sname}.py` de pass pytest.\nCode:\n{FENCE}python\n{curr}\n{FENCE}\nTest:\n{FENCE}python\n{test_code}\n{FENCE}\nLog:\n{log[-1500:]}"
                curr = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python.", fix_p, 0.2))

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str):
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) quet kho va quyet dinh tao tool moi neu thieu
        reg = scan_registry()
        plan_p = f"""CAU HOI NGUOI DUNG: {question}

KHO SKILL HIEN CO:
{reg}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
2. Neu THIEU tool (vi du: thoi tiet, Wikipedia API, tim kiem tin tuc, ArXiv...), hay thiet ke 1-2 tool moi can tao trong `missing_tools`.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT JSON:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra kho skills/...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        print(f"   -> Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        # BUOC 2: Tu che tao cac tool con thieu
        for spec in missing:
            await forge_tool(client, spec)

        # BUOC 3: Chay script thu thap du lieu that + Self-Healing (tu sua tool neu chay that bi loi)
        reg_updated = scan_registry()
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
Kho skill hien co (import truc tiep vi PYTHONPATH da co skills/):
{reg_updated}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau: In ket qua thu thap duoc (kem link nguon neu co) ra stdout, toi da ~4000 ky tu. Chi tra ve code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve code Python.", run_p, 0.2))

        collected_data = ""
        for attempt in range(1, 4):
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)
            if ok and len(out.strip()) > 30:
                collected_data = out[:6000]
                print(f"   -> [OK lan {attempt}] Thu thap thanh cong {len(collected_data)} ky tu du lieu that!")
                break

            print(f"   -> [FAIL lan {attempt}] {out.splitlines()[-1][:120] if out else 'Output rong'}")
            if attempt == 3:
                collected_data = out[-2000:]
                break

            # qwq:32b chan doan: loi do runner hay loi do file tool trong skills/?
            diag_p = f"""Script thu thap du lieu bi loi khi chay thuc te.
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Log loi:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong thu muc `skills/` (vi du: API doi, bi chan 403, parse sai): tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi chi do `live_runner.py`: tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": "Sua lai runner"}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath):
                    print(f"   -> [SELF-HEAL] Phat hien loi trong `{sname}.py` ({diag.get('reason')}) -> Dang tu va tool...")
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc loi chay thuc te: {diag.get('reason')}
Code cu:
{FENCE}python
{old_src}
{FENCE}
Log loi:
{out[-1500:]}
Giu nguyen ten class/ham public. Chi tra ve code Python."""
                    patched = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi tool Python.", patch_p, 0.2))
                    with open(spath, "w", encoding="utf-8") as f:
                        f.write(patched)
                    continue

            # Sua runner
            fix_run_p = f"Sua `live_runner.py` de chay thanh cong va in du lieu ra stdout.\nLy do: {diag.get('reason')}\nCode:\n{FENCE}python\n{runner_code}\n{FENCE}\nLog:\n{out[-1500:]}"
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python.", fix_run_p, 0.2))

        # BUOC 4: Tong hop cau tra loi
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau: Tra loi chinh xac, ro rang bang tieng Viet, chi dua tren du lieu tren, khong tu bia. Kem link nguon o cuoi."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {round(time.perf_counter() - t0, 2)}s\n")
        print("================ TRA LOI ================")
        print(final_ans)


if __name__ == "__main__":
    q = " ".join(sys.argv[1:]).strip()
    if not q:
        print('Dung: python3 live_evolve_agent.py "Cau hoi cua ban"')
        sys.exit(1)
    asyncio.run(solve_question(q))
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import asyncio
import subprocess
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str) -> str:
    m = re.search(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    return m.group(1).strip() if m else text.strip()


def extract_json(text: str) -> dict:
    s, e = text.find("{"), text.rfind("}") + 1
    if s == -1 or e <= s:
        raise ValueError("Khong tim thay JSON")
    return json.loads(text[s:e], strict=False)


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def scan_registry() -> str:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    lines = []
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        try:
            tree = ast.parse(open(f"{SKILLS_DIR}/{fn}", encoding="utf-8").read())
            sigs = []
            for n in tree.body:
                if isinstance(n, ast.ClassDef):
                    ms = [
                        f"{m.name}({', '.join(a.arg for a in m.args.args if a.arg != 'self')})"
                        for m in n.body
                        if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef))
                        and (not m.name.startswith("_") or m.name == "__init__")
                    ]
                    sigs.append(f"class {n.name}: [{', '.join(ms)}]")
                elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                    sigs.append(f"def {n.name}({', '.join(a.arg for a in n.args.args)})")
            lines.append(f"- `{mod}`: " + " | ".join(sigs))
        except Exception:
            lines.append(f"- `{mod}`")
    return "\n".join(lines) if lines else "(Kho dang trong)"


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:140] if errs else (lines[-1][:140] if lines else "Unknown")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def forge_tool(client: httpx.AsyncClient, spec: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    # Sinh 3 nhanh code + 1 bo test (chot cung httpx de test va code khong lech mock)
    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3-4 test case ro rang (happy path + loi).
- Code thuc thi dung thu vien `httpx`, nen dung `unittest.mock.patch("httpx.get")` hoac `patch("httpx.Client.get")` de mock mang, KHONG mock `requests`.
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve code Python trong `{FENCE}python ... {FENCE}`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- Dung `httpx.get(...)` (KHONG dung `requests`), kem `timeout=15.0` va `headers={{"User-Agent": "AutoEvolveBot/1.0"}}`.
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY:
  * Thoi tiet: Open-Meteo (`https://geocoding-api.open-meteo.com/v1/search` + `https://api.open-meteo.com/v1/forecast`).
  * Wikipedia: MediaWiki API (`https://vi.wikipedia.org/w/api.php` va `en.wikipedia.org/w/api.php`).
  * Tin tuc/Search: `ddgs` hoac Google News RSS.
- Tuyet doi khong hardcode du lieu gia.
Chi tra ve code Python trong `{FENCE}python ... {FENCE}`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0])
    candidates = [extract_code(c) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                        f.write(curr)
                    with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                        f.write(test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"Sua `{sname}.py` de pass pytest (van dung `httpx.get` va API that).\nCode:\n{FENCE}python\n{curr}\n{FENCE}\nTest:\n{FENCE}python\n{test_code}\n{FENCE}\nLog:\n{log[-1500:]}"
                    curr = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve code.", fix_p, 0.2))

    # Trong tai (qwq:32b) phan xu neu ca 3 nhanh deu fail (rat hay gap do Test mock sai)
    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, sai cau truc response, assert sai):
Tra ve `VERDICT: TEST_BUG` kem toan bo file test da sua trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                    f.write(last_code0)
                with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                    f.write(fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str):
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)

    async with httpx.AsyncClient() as client:
        reg = scan_registry()
        plan_p = f"""CAU HOI NGUOI DUNG: {question}

KHO SKILL HIEN CO:
{reg}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
2. Neu THIEU tool, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...).
   - Thoi tiet -> dung Open-Meteo (khong can key, khong nhan tham so `api_key`).
   - Kien thuc/Lich su/Khoa hoc -> dung MediaWiki / Wikipedia API (khong can key).
   - Tin tuc/Thoi su -> dung Google News RSS hoac `ddgs`.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT JSON:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra kho skills/...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        print(f"   -> Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        for spec in missing:
            await forge_tool(client, spec)

        reg_updated = scan_registry()
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
Kho skill hien co (import truc tiep vi PYTHONPATH da co skills/):
{reg_updated}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau: In ket qua thu thap duoc (kem link nguon neu co) ra stdout, toi da ~4000 ky tu. Chi tra ve code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve code Python.", run_p, 0.2))

        collected_data = ""
        for attempt in range(1, 4):
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)
            if ok and len(out.strip()) > 30:
                collected_data = out[:6000]
                print(f"   -> [OK lan {attempt}] Thu thap thanh cong {len(collected_data)} ky tu du lieu that!")
                break

            print(f"   -> [FAIL lan {attempt}] {summary}")
            if attempt == 3:
                collected_data = out[-2000:]
                break

            diag_p = f"""Script thu thap du lieu bi loi khi chay thuc te.
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Log loi:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi chi do `live_runner.py`: tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": "Sua lai runner"}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath):
                    print(f"   -> [SELF-HEAL] Phat hien loi trong `{sname}.py` ({diag.get('reason')}) -> Dang tu va tool...")
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc loi chay thuc te: {diag.get('reason')}
Code cu:
{FENCE}python
{old_src}
{FENCE}
Log loi:
{out[-1500:]}
Giu nguyen ten class/ham public. Chi tra ve code Python."""
                    patched = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi tool Python.", patch_p, 0.2))
                    with open(spath, "w", encoding="utf-8") as f:
                        f.write(patched)
                    continue

            fix_run_p = f"Sua `live_runner.py` de chay thanh cong va in du lieu ra stdout.\nLy do: {diag.get('reason')}\nCode:\n{FENCE}python\n{runner_code}\n{FENCE}\nLog:\n{out[-1500:]}"
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python.", fix_run_p, 0.2))

        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau: Tra loi chinh xac, ro rang bang tieng Viet, chi dua tren du lieu tren, khong tu bia. Kem link nguon o cuoi."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {round(time.perf_counter() - t0, 2)}s\n")
        print("================ TRA LOI ================")
        print(final_ans)


if __name__ == "__main__":
    q = " ".join(sys.argv[1:]).strip()
    if not q:
        print('Dung: python3 live_evolve_agent.py "Cau hoi cua ban"')
        sys.exit(1)
    asyncio.run(solve_question(q))
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import asyncio
import subprocess
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str, module_name: str | None = None, is_test: bool = False) -> str:
    blocks = re.findall(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    if not blocks and FENCE in text:
        part = text.split(FENCE, 1)[1]
        part = re.sub(r"^python\s*", "", part, flags=re.IGNORECASE)
        blocks = [part]
    if not blocks:
        blocks = [text]

    valid = []
    for b in blocks:
        code = b.strip()
        if not code:
            continue
        if module_name and not is_test:
            if re.search(rf"\b(from|import)\s+{re.escape(module_name)}\b", code):
                continue
            if "def test_" in code and "class " not in code:
                continue
        try:
            ast.parse(code)
            valid.append(code)
        except SyntaxError:
            continue

    if valid:
        return max(valid, key=len)
    return blocks[0].strip()


def extract_json(text: str) -> dict:
    s, e = text.find("{"), text.rfind("}") + 1
    if s == -1 or e <= s:
        raise ValueError("Khong tim thay JSON")
    return json.loads(text[s:e], strict=False)


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def scan_registry() -> str:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    lines = []
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        try:
            tree = ast.parse(open(f"{SKILLS_DIR}/{fn}", encoding="utf-8").read())
            sigs = []
            for n in tree.body:
                if isinstance(n, ast.ClassDef):
                    ms = [
                        f"{m.name}({', '.join(a.arg for a in m.args.args if a.arg != 'self')})"
                        for m in n.body
                        if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef))
                        and (not m.name.startswith("_") or m.name == "__init__")
                    ]
                    sigs.append(f"class {n.name}: [{', '.join(ms)}]")
                elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                    sigs.append(f"def {n.name}({', '.join(a.arg for a in n.args.args)})")
            lines.append(f"- `{mod}`: " + " | ".join(sigs))
        except Exception:
            lines.append(f"- `{mod}`")
    return "\n".join(lines) if lines else "(Kho dang trong)"


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:{BASE_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:160] if errs else (out.replace("\n", " | ")[:160] if out else "Output rong")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong (cho phep them tham so tuy chon co gia tri mac dinh)?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def verify_collected_data(client: httpx.AsyncClient, question: str, data: str) -> tuple[bool, str]:
    ver_p = f"""Cau hoi nguoi dung: {question}

Du lieu script vua thu thap duoc:
{data[:3000]}

Kiem tra: Du lieu tren da cung cap du thong tin thuc te de tra loi TAT CA cac ve cua cau hoi chua?
(Neu thieu bat ky ve nao, hoac bi loi/rong/N/A/khong tim thay -> FAIL).
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
MISSING: <neu FAIL, ghi ro phan nao bi thieu hoac rong; neu PASS ghi None>"""
    res = await ask_llm(client, BRAIN_THINK, "Ban la QA Data Verifier nghiem khac.", ver_p, 0.1)
    ok = "STATUS: PASS" in res.upper()
    m = re.search(r"MISSING:\s*(.+)", res)
    return ok, (m.group(1).strip() if m else res.splitlines()[-1][:120])


async def forge_tool(client: httpx.AsyncClient, spec: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3-4 test case ro rang (happy path + loi).
- Code thuc thi dung `httpx.get`, nen dung `unittest.mock.patch("httpx.get")` hoac `patch("httpx.Client.get")` de mock mang, KHONG mock `requests`.
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve duy nhat 1 block code Python cho `test_{sname}.py`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})` (KHONG dung `requests`).
- KHONG nuot loi mang im lang (neu HTTP status != 200, raise RuntimeError kem status_code va noi dung loi).
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY:
  * Thoi tiet: Open-Meteo (`https://geocoding-api.open-meteo.com/v1/search` + `https://api.open-meteo.com/v1/forecast`).
  * Wikipedia: MediaWiki API (`https://vi.wikipedia.org/w/api.php` va `en.wikipedia.org/w/api.php`), dung `action=query&list=search&format=json` tim tieu de chuan truoc, roi goi `action=query&prop=extracts&exintro=1&explaintext=1&redirects=1&format=json`.
  * Khoa hoc: ArXiv (`https://export.arxiv.org/api/query`) hoac OpenAlex (`https://api.openalex.org/works`).
  * Tin tuc/Search: `ddgs` hoac Google News RSS (`https://news.google.com/rss/search`).
  * Tai chinh/Ty gia: Frankfurter (`https://api.frankfurter.app/latest`) hoac Binance public API.
- Tuyet doi khong hardcode du lieu gia, khong viet code test trong file nay.
Chi tra ve duy nhat 1 block code Python cho `{sname}.py`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve 1 block code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0], module_name=sname, is_test=True)
    candidates = [extract_code(c, module_name=sname, is_test=False) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                        f.write(curr)
                    with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                        f.write(test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"""Sua file `{sname}.py` de pass pytest (van dung `httpx.get(..., follow_redirects=True)` va API that).
CHI tra ve code cua `{sname}.py`, KHONG tra ve file test.
Code:
{FENCE}python
{curr}
{FENCE}
Test:
{FENCE}python
{test_code}
{FENCE}
Log:
{log[-1500:]}"""
                    curr = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code cho file skill.", fix_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )

    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, sai cau truc response, assert sai):
Tra ve `VERDICT: TEST_BUG` kem toan bo file `test_{sname}.py` da sua trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj, module_name=sname, is_test=True)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                    f.write(last_code0)
                with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                    f.write(fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str):
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) phan tich nhu cau & quet kho skills/
        reg = scan_registry()
        plan_p = f"""CAU HOI NGUOI DUNG: {question}

KHO SKILL HIEN CO:
{reg}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
2. Neu THIEU tool, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...).
   - Thoi tiet -> dung Open-Meteo (khong can key, khong nhan tham so `api_key`).
   - Kien thuc/Lich su/Khoa hoc -> dung MediaWiki / Wikipedia API hoac ArXiv/OpenAlex.
   - Tin tuc/Thoi su -> dung Google News RSS hoac `ddgs`.
   - Tai chinh/Ty gia -> dung Frankfurter hoac Binance public API.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT JSON:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra kho skills/...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        print(f"   -> Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        # BUOC 2: Forge tu che tao tool con thieu
        for spec in missing:
            await forge_tool(client, spec)

        # BUOC 3: Coder chay thuc chien + Data Quality Gate + Self-Healing (toi da 5 luot)
        reg_updated = scan_registry()
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
Kho skill hien co (import truc tiep vi PYTHONPATH da co skills/):
{reg_updated}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau:
- Lay day du du lieu cho TAT CA cac ve cua cau hoi.
- KHONG nuot loi im lang thanh 'N/A': in thang ket qua tho (repr/str) hoac traceback loi ra stdout/stderr neu goi tool that bai.
- In ro ket qua tung phan (kem link nguon neu co) ra stdout, toi da ~4000 ky tu.
- Chi tra ve 1 block code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve 1 block code Python.", run_p, 0.2))

        collected_data = ""
        heal_counts: dict[str, int] = {}
        max_attempts = 5
        for attempt in range(1, max_attempts + 1):
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)

            fail_reason = summary
            if ok and len(out.strip()) > 80:
                gate_ok, missing_info = await verify_collected_data(client, question, out)
                if gate_ok:
                    collected_data = out[:6000]
                    print(f"   -> [OK lan {attempt}] Thu thap DU du lieu ({len(collected_data)} ky tu)!")
                    print(f"   -> [PREVIEW] {collected_data[:220].replace(chr(10), ' | ')}...")
                    break
                else:
                    fail_reason = f"DATA_INCOMPLETE: {missing_info}"
                    print(f"   -> [THIEU DU LIEU lan {attempt}] {missing_info}")
                    print(f"      [RAW OUT] {out[:200].replace(chr(10), ' | ')}")
            else:
                print(f"   -> [FAIL lan {attempt}] {summary}")

            if attempt == max_attempts:
                collected_data = out[-3000:]
                break

            diag_p = f"""Script thu thap du lieu bi loi hoac thieu du lieu khi chay thuc te.
Van de: {fail_reason}
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Output / Log:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi do `live_runner.py` (goi sai ham, doc sai key dict, nuot loi): tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": fail_reason}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath) and heal_counts.get(sname, 0) < 2:
                    heal_counts[sname] = heal_counts.get(sname, 0) + 1
                    print(f"   -> [SELF-HEAL #{heal_counts[sname]}] Va `{sname}.py` ({diag.get('reason')})...")
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc van de chay thuc te: {diag.get('reason')}

Code hien tai cua `{sname}.py`:
{FENCE}python
{old_src}
{FENCE}

Cach `live_runner.py` dang goi va doc ket qua tu tool nay:
{FENCE}python
{runner_code}
{FENCE}

Output / Log thuc te:
{out[-1500:]}

QUY TAC SUA BAT BUOC:
1. Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})`.
2. TUYET DOI KHONG nuot loi im lang tra ve rong/None; neu API loi phai raise RuntimeError kem status_code va response text.
3. Neu la Wikipedia API: dung `https://vi.wikipedia.org/w/api.php` (fallback `en.wikipedia.org/w/api.php`) voi 2 buoc:
   - Buoc A: `params={{"action": "query", "list": "search", "srsearch": query, "format": "json"}}` de lay `title` dau tien.
   - Buoc B: `params={{"action": "query", "prop": "extracts", "exintro": 1, "explaintext": 1, "redirects": 1, "titles": title, "format": "json"}}` de lay `extract` va link URL.
4. Tra ve dung dinh dang/key ma `live_runner.py` dang mong doi. Giu nguyen ten class/ham public.
Chi tra ve 1 block code Python cho `{sname}.py`."""
                    patched = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi tool Python. Chi tra ve 1 block code.", patch_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )
                    with open(spath, "w", encoding="utf-8") as f:
                        f.write(patched)
                    continue

            fix_run_p = f"""Sua `live_runner.py` de lay day du du lieu va in ra stdout (KHONG nuot loi thanh 'N/A', in ro gia tri tra ve cua tung tool).
Van de: {diag.get('reason')}
Code:
{FENCE}python
{runner_code}
{FENCE}
Output/Log:
{out[-1500:]}"""
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code.", fix_run_p, 0.2))

        # BUOC 4: Analyst (qwq:32b) tong hop bao cao
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau BAT BUOC:
- Tra loi chinh xac, ro rang bang tieng Viet chuan (TUYET DOI khong chen chu Han/tieng Trung).
- Chi dua tren du lieu thu thap o tren; phan nao khong co du lieu thi noi ro la chua thu thap duoc.
- KHONG bao gio tu bia link URL; chi trich dan cac link co mat trong du lieu thu thap."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {round(time.perf_counter() - t0, 2)}s\n")
        print("================ TRA LOI ================")
        print(final_ans)


if __name__ == "__main__":
    q = " ".join(sys.argv[1:]).strip()
    if not q:
        print('Dung: python3 live_evolve_agent.py "Cau hoi cua ban"')
        sys.exit(1)
    asyncio.run(solve_question(q))
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import asyncio
import subprocess
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str, module_name: str | None = None, is_test: bool = False) -> str:
    blocks = re.findall(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    if not blocks and FENCE in text:
        part = text.split(FENCE, 1)[1]
        part = re.sub(r"^python\s*", "", part, flags=re.IGNORECASE)
        blocks = [part]
    if not blocks:
        blocks = [text]

    valid = []
    for b in blocks:
        code = b.strip()
        if not code:
            continue
        if module_name and not is_test:
            if re.search(rf"\b(from|import)\s+{re.escape(module_name)}\b", code):
                continue
            if "def test_" in code and "class " not in code:
                continue
        try:
            ast.parse(code)
            valid.append(code)
        except SyntaxError:
            continue

    if valid:
        return max(valid, key=len)
    return blocks[0].strip()


def extract_json(text: str) -> dict:
    # Uu tien doc trong block ```json ... ``` truoc
    candidates = re.findall(FENCE + r"(?:json)?\s*(\{.*?\})\s*" + FENCE, text, re.DOTALL)
    candidates.append(text)
    decoder = json.JSONDecoder(strict=False)
    for chunk in candidates:
        pos = 0
        while True:
            idx = chunk.find("{", pos)
            if idx == -1:
                break
            try:
                obj, _ = decoder.raw_decode(chunk, idx)
                if isinstance(obj, dict):
                    return obj
            except json.JSONDecodeError:
                pass
            pos = idx + 1
    raise ValueError(f"Khong tim thay JSON hop le trong: {text[:200]}")


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def scan_registry() -> str:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    lines = []
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        try:
            tree = ast.parse(open(f"{SKILLS_DIR}/{fn}", encoding="utf-8").read())
            sigs = []
            for n in tree.body:
                if isinstance(n, ast.ClassDef):
                    ms = [
                        f"{m.name}({', '.join(a.arg for a in m.args.args if a.arg != 'self')})"
                        for m in n.body
                        if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef))
                        and (not m.name.startswith("_") or m.name == "__init__")
                    ]
                    sigs.append(f"class {n.name}: [{', '.join(ms)}]")
                elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                    sigs.append(f"def {n.name}({', '.join(a.arg for a in n.args.args)})")
            lines.append(f"- `{mod}`: " + " | ".join(sigs))
        except Exception:
            lines.append(f"- `{mod}`")
    return "\n".join(lines) if lines else "(Kho dang trong)"


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:{BASE_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:160] if errs else (out.replace("\n", " | ")[:160] if out else "Output rong")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong (cho phep them tham so tuy chon co gia tri mac dinh)?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def verify_collected_data(client: httpx.AsyncClient, question: str, data: str) -> tuple[bool, str]:
    ver_p = f"""Cau hoi nguoi dung: {question}

Du lieu script vua thu thap duoc:
{data[:3000]}

Kiem tra: Du lieu tren da cung cap du thong tin thuc te de tra loi TAT CA cac ve cua cau hoi chua?
(Neu thieu bat ky ve nao, hoac bi loi/rong/N/A/[example.com/404](https://example.com/404) -> FAIL).
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
MISSING: <neu FAIL, ghi ro phan nao bi thieu hoac rong; neu PASS ghi None>"""
    res = await ask_llm(client, BRAIN_THINK, "Ban la QA Data Verifier nghiem khac.", ver_p, 0.1)
    ok = "STATUS: PASS" in res.upper()
    m = re.search(r"MISSING:\s*(.+)", res)
    return ok, (m.group(1).strip() if m else res.splitlines()[-1][:120])


async def forge_tool(client: httpx.AsyncClient, spec: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3 test case ro rang (1 happy path + 1 loi HTTP status != 200 + 1 loi du lieu rong/mang).
- Code thuc thi dung `httpx.get(...)`, nen mock bang `unittest.mock.patch("httpx.get")` (KHONG mock `requests`).
- Khi test truong hop loi, LUON dung `with pytest.raises(Exception):` (KHONG ep loai exception cu the nhu ValueError/HTTPError de tranh lech voi RuntimeError).
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve duy nhat 1 block code Python cho `test_{sname}.py`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- LUON dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 ([https://example.org](https://example.org))"}})` (KHONG dung `requests` hay thu vien ngoai kho mock).
- KHONG nuot loi mang im lang (neu HTTP status != 200 hoac du lieu loi, raise `RuntimeError` kem status_code va noi dung loi).
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY:
  * Thoi tiet: Open-Meteo (`[https://geocoding-api.open-meteo.com/v1/search](https://geocoding-api.open-meteo.com/v1/search)` + `[https://api.open-meteo.com/v1/forecast](https://api.open-meteo.com/v1/forecast)`).
  * Wikipedia: MediaWiki API (`[https://vi.wikipedia.org/w/api.php](https://vi.wikipedia.org/w/api.php)` va `en.wikipedia.org/w/api.php`).
  * Khoa hoc: ArXiv (`[https://export.arxiv.org/api/query](https://export.arxiv.org/api/query)`, parse bang `xml.etree.ElementTree`) hoac OpenAlex (`[https://api.openalex.org/works](https://api.openalex.org/works)`).
  * Tin tuc/Thoi su: Google News RSS (`[https://news.google.com/rss/search?q=...&hl=vi&gl=VN&ceid=VN:vi](https://news.google.com/rss/search?q=...&hl=vi&gl=VN&ceid=VN:vi)`, parse XML bang `xml.etree.ElementTree`, doc `<item><title>`, `<link>`, `<pubDate>`).
  * Tai chinh/Ty gia: Frankfurter (`[https://api.frankfurter.app/latest](https://api.frankfurter.app/latest)`) hoac Binance public (`[https://api.binance.com/api/v3/ticker/price](https://api.binance.com/api/v3/ticker/price)`).
- Tuyet doi khong hardcode du lieu gia, khong viet code test trong file nay.
Chi tra ve duy nhat 1 block code Python cho `{sname}.py`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve 1 block code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0], module_name=sname, is_test=True)
    candidates = [extract_code(c, module_name=sname, is_test=False) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                        f.write(curr)
                    with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                        f.write(test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"""Sua file `{sname}.py` de pass pytest (van dung `httpx.get(..., follow_redirects=True)` va API that).
CHI tra ve code cua `{sname}.py`, KHONG tra ve file test.
Code:
{FENCE}python
{curr}
{FENCE}
Test:
{FENCE}python
{test_code}
{FENCE}
Log:
{log[-1500:]}"""
                    curr = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code cho file skill.", fix_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )

    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, ep sai loai exception, sai cau truc response):
Tra ve `VERDICT: TEST_BUG` kem toan bo file `test_{sname}.py` da sua (dung cu phap Python hop le) trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj, module_name=sname, is_test=True)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                    f.write(last_code0)
                with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                    f.write(fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str):
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)
    vn_now = datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) phan tich nhu cau & quet kho skills/
        reg = scan_registry()
        plan_p = f"""CAU HOI NGUOI DUNG: {question}
THOI GIAN HIEN TAI: {vn_now}

KHO SKILL HIEN CO:
{reg}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
2. Neu THIEU tool, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...).
   - Thoi tiet -> dung Open-Meteo.
   - Kien thuc/Lich su -> dung MediaWiki / Wikipedia API.
   - Khoa hoc -> dung ArXiv API (`[https://export.arxiv.org/api/query](https://export.arxiv.org/api/query)`) hoac OpenAlex (`[https://api.openalex.org/works](https://api.openalex.org/works)`).
   - Tin tuc/Thoi su -> dung Google News RSS (`[https://news.google.com/rss/search](https://news.google.com/rss/search)`, dung thuan `httpx.get` + `xml.etree.ElementTree`, KHONG dung `ddgs`).
   - Tai chinh/Ty gia -> dung Frankfurter hoac Binance public API.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT 1 JSON hop le, khong kem giai thich ben ngoai:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra kho skills/...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve 1 JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        print(f"   -> Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        # BUOC 2: Forge tu che tao tool con thieu
        failed_tools = []
        for spec in missing:
            ok_forge = await forge_tool(client, spec)
            if not ok_forge:
                failed_tools.append(spec["skill_name"])

        # BUOC 3: Coder chay thuc chien + Data Quality Gate + Self-Healing (toi da 5 luot)
        reg_updated = scan_registry()
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        failed_note = ""
        if failed_tools:
            failed_note = f"\nLUU Y QUAN TRONG: Cac tool sau CHUA DUOC TAO nen KHONG import: {failed_tools}. Hay goi truc tiep API mo (Google News RSS / ArXiv / Open-Meteo / Wikipedia...) bang `httpx.get` trong script nay!"

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
Kho skill hien co (chi import cac module CO MAT trong danh sach nay):
{reg_updated}
{failed_note}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau BAT BUOC:
- Lay day du du lieu cho TAT CA cac ve cua cau hoi.
- TUYET DOI KHONG dung URL gia nhu `example.com`; phai goi API/RSS that bang `httpx.get(..., follow_redirects=True)`.
- KHONG nuot loi im lang thanh 'N/A': in thang ket qua tho hoac traceback loi ra stdout/stderr.
- In ro ket qua tung phan (kem link nguon neu co) ra stdout, toi da ~4000 ky tu.
- Chi tra ve 1 block code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve 1 block code Python.", run_p, 0.2))

        collected_data = ""
        heal_counts: dict[str, int] = {}
        max_attempts = 5
        for attempt in range(1, max_attempts + 1):
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)

            fail_reason = summary
            if ok and len(out.strip()) > 80:
                gate_ok, missing_info = await verify_collected_data(client, question, out)
                if gate_ok:
                    collected_data = out[:6000]
                    print(f"   -> [OK lan {attempt}] Thu thap DU du lieu ({len(collected_data)} ky tu)!")
                    print(f"   -> [PREVIEW] {collected_data[:220].replace(chr(10), ' | ')}...")
                    break
                else:
                    fail_reason = f"DATA_INCOMPLETE: {missing_info}"
                    print(f"   -> [THIEU DU LIEU lan {attempt}] {missing_info}")
                    print(f"      [RAW OUT] {out[:200].replace(chr(10), ' | ')}")
            else:
                print(f"   -> [FAIL lan {attempt}] {summary}")

            if attempt == max_attempts:
                collected_data = out[-3000:]
                break

            diag_p = f"""Script thu thap du lieu bi loi hoac thieu du lieu khi chay thuc te.
Van de: {fail_reason}
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Output / Log:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi do `live_runner.py` (goi sai ham, import tool khong ton tai, dung URL gia): tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": fail_reason}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath) and heal_counts.get(sname, 0) < 2:
                    heal_counts[sname] = heal_counts.get(sname, 0) + 1
                    print(f"   -> [SELF-HEAL #{heal_counts[sname]}] Va `{sname}.py` ({diag.get('reason')})...")
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc van de chay thuc te: {diag.get('reason')}

Code hien tai cua `{sname}.py`:
{FENCE}python
{old_src}
{FENCE}

Cach `live_runner.py` dang goi va doc ket qua tu tool nay:
{FENCE}python
{runner_code}
{FENCE}

Output / Log thuc te:
{out[-1500:]}

QUY TAC SUA BAT BUOC:
1. Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 ([https://example.org](https://example.org))"}})`.
2. TUYET DOI KHONG nuot loi im lang tra ve rong/None; neu API loi phai raise RuntimeError kem status_code va response text.
3. Tra ve dung dinh dang/key ma `live_runner.py` dang mong doi. Giu nguyen ten class/ham public.
Chi tra ve 1 block code Python cho `{sname}.py`."""
                    patched = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi tool Python. Chi tra ve 1 block code.", patch_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )
                    with open(spath, "w", encoding="utf-8") as f:
                        f.write(patched)
                    continue

            fix_run_p = f"""Sua `live_runner.py` de lay day du du lieu that va in ra stdout (KHONG dung example.com, KHONG nuot loi thanh 'N/A').
Van de: {diag.get('reason')}
Kho skill hien co:
{reg_updated}
Code cu:
{FENCE}python
{runner_code}
{FENCE}
Output/Log:
{out[-1500:]}"""
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code.", fix_run_p, 0.2))

        # BUOC 4: Analyst (qwq:32b) tong hop bao cao
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""THOI GIAN HIEN TAI CUA HE THONG: {vn_now} (nam 2026 la thoi diem hien tai that, KHONG phai tuong lai hay gia lap).
Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau BAT BUOC:
- Tra loi chinh xac, ro rang bang tieng Viet chuan (TUYET DOI khong chen chu Han/tieng Trung).
- Chi dua tren du lieu thu thap o tren; phan nao khong co du lieu thi noi ro la chua thu thap duoc.
- KHONG bao gio tu bia link URL; chi trich dan cac link co mat trong du lieu thu thap."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {round(time.perf_counter() - t0, 2)}s\n")
        print("================ TRA LOI ================")
        print(final_ans)


if __name__ == "__main__":
    q = " ".join(sys.argv[1:]).strip()
    if not q:
        print('Dung: python3 live_evolve_agent.py "Cau hoi cua ban"')
        sys.exit(1)
    asyncio.run(solve_question(q))
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import asyncio
import subprocess
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str, module_name: str | None = None, is_test: bool = False) -> str:
    blocks = re.findall(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    if not blocks and FENCE in text:
        part = text.split(FENCE, 1)[1]
        part = re.sub(r"^python\s*", "", part, flags=re.IGNORECASE)
        blocks = [part]
    if not blocks:
        blocks = [text]

    valid = []
    for b in blocks:
        code = b.strip()
        if not code:
            continue
        if module_name and not is_test:
            if re.search(rf"\b(from|import)\s+{re.escape(module_name)}\b", code):
                continue
            if "def test_" in code and "class " not in code:
                continue
        try:
            ast.parse(code)
            valid.append(code)
        except SyntaxError:
            continue

    if valid:
        return max(valid, key=len)
    return blocks[0].strip()


def extract_json(text: str) -> dict:
    candidates = re.findall(FENCE + r"(?:json)?\s*(\{.*?\})\s*" + FENCE, text, re.DOTALL)
    candidates.append(text)
    decoder = json.JSONDecoder(strict=False)
    for chunk in candidates:
        pos = 0
        while True:
            idx = chunk.find("{", pos)
            if idx == -1:
                break
            try:
                obj, _ = decoder.raw_decode(chunk, idx)
                if isinstance(obj, dict):
                    return obj
            except json.JSONDecodeError:
                pass
            pos = idx + 1
    raise ValueError(f"Khong tim thay JSON hop le trong: {text[:200]}")


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def scan_registry() -> str:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    lines = []
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        try:
            tree = ast.parse(open(f"{SKILLS_DIR}/{fn}", encoding="utf-8").read())
            sigs = []
            for n in tree.body:
                if isinstance(n, ast.ClassDef):
                    ms = [
                        f"{m.name}({', '.join(a.arg for a in m.args.args if a.arg != 'self')})"
                        for m in n.body
                        if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef))
                        and (not m.name.startswith("_") or m.name == "__init__")
                    ]
                    sigs.append(f"class {n.name}: [{', '.join(ms)}]")
                elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                    sigs.append(f"def {n.name}({', '.join(a.arg for a in n.args.args)})")
            lines.append(f"- `{mod}`: " + " | ".join(sigs))
        except Exception:
            lines.append(f"- `{mod}`")
    return "\n".join(lines) if lines else "(Kho dang trong)"


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:{BASE_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:160] if errs else (out.replace("\n", " | ")[:160] if out else "Output rong")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong (cho phep them tham so tuy chon co gia tri mac dinh)?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def verify_collected_data(client: httpx.AsyncClient, question: str, data: str) -> tuple[bool, str]:
    ver_p = f"""Cau hoi nguoi dung: {question}

Du lieu script vua thu thap duoc:
{data[:8000]}

Kiem tra: Du lieu tren da cung cap du thong tin thuc te de tra loi TAT CA cac ve cua cau hoi chua?
(Neu thieu bat ky ve nao, hoac bi loi/rong/N/A/example.com/404 -> FAIL).
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
MISSING: <neu FAIL, ghi ro phan nao bi thieu hoac rong; neu PASS ghi None>"""
    res = await ask_llm(client, BRAIN_THINK, "Ban la QA Data Verifier nghiem khac.", ver_p, 0.1)
    ok = "STATUS: PASS" in res.upper()
    m = re.search(r"MISSING:\s*(.+)", res)
    return ok, (m.group(1).strip() if m else res.splitlines()[-1][:120])


async def forge_tool(client: httpx.AsyncClient, spec: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3 test case ro rang (1 happy path + 1 loi HTTP status != 200 + 1 loi du lieu rong/mang).
- Code thuc thi dung `httpx.get(...)`, nen mock bang `unittest.mock.patch("httpx.get")` (KHONG mock `requests`).
- Khi test truong hop loi, LUON dung `with pytest.raises(Exception):` (KHONG ep loai exception cu the nhu ValueError/HTTPError de tranh lech voi RuntimeError).
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve duy nhat 1 block code Python cho `test_{sname}.py`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- LUON dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})` (KHONG dung `requests` hay thu vien ngoai kho mock).
- KHONG nuot loi mang im lang (neu HTTP status != 200 hoac du lieu loi, raise `RuntimeError` kem status_code va noi dung loi).
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY:
  * Thoi tiet: Open-Meteo (`https://geocoding-api.open-meteo.com/v1/search` + `https://api.open-meteo.com/v1/forecast`).
  * Wikipedia: MediaWiki API (`https://vi.wikipedia.org/w/api.php` va `en.wikipedia.org/w/api.php`).
  * Khoa hoc: ArXiv (`https://export.arxiv.org/api/query`, parse bang `xml.etree.ElementTree`) hoac OpenAlex (`https://api.openalex.org/works`).
  * Tin tuc/Thoi su: Google News RSS (`https://news.google.com/rss/search?q=...&hl=vi&gl=VN&ceid=VN:vi`, parse XML bang `xml.etree.ElementTree`, doc `<item><title>`, `<link>`, `<pubDate>`).
  * Tai chinh/Ty gia: Frankfurter (`https://api.frankfurter.app/latest`) hoac Binance public (`https://api.binance.com/api/v3/ticker/price`).
- Tuyet doi khong hardcode du lieu gia, khong viet code test trong file nay.
Chi tra ve duy nhat 1 block code Python cho `{sname}.py`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve 1 block code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0], module_name=sname, is_test=True)
    candidates = [extract_code(c, module_name=sname, is_test=False) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                        f.write(curr)
                    with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                        f.write(test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"""Sua file `{sname}.py` de pass pytest (van dung `httpx.get(..., follow_redirects=True)` va API that).
CHI tra ve code cua `{sname}.py`, KHONG tra ve file test.
Code:
{FENCE}python
{curr}
{FENCE}
Test:
{FENCE}python
{test_code}
{FENCE}
Log:
{log[-1500:]}"""
                    curr = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code cho file skill.", fix_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )

    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, ep sai loai exception, sai cau truc response):
Tra ve `VERDICT: TEST_BUG` kem toan bo file `test_{sname}.py` da sua (dung cu phap Python hop le) trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj, module_name=sname, is_test=True)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                    f.write(last_code0)
                with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                    f.write(fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str):
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)
    vn_now = datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) phan tich nhu cau & quet kho skills/
        reg = scan_registry()
        plan_p = f"""CAU HOI NGUOI DUNG: {question}
THOI GIAN HIEN TAI: {vn_now}

KHO SKILL HIEN CO:
{reg}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
2. Neu THIEU tool, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...).
   - Thoi tiet -> dung Open-Meteo.
   - Kien thuc/Lich su -> dung MediaWiki / Wikipedia API.
   - Khoa hoc -> dung ArXiv API (`https://export.arxiv.org/api/query`) hoac OpenAlex (`https://api.openalex.org/works`).
   - Tin tuc/Thoi su -> dung Google News RSS (`https://news.google.com/rss/search`, dung thuan `httpx.get` + `xml.etree.ElementTree`, KHONG dung `ddgs`).
   - Tai chinh/Ty gia -> dung Frankfurter hoac Binance public API.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT 1 JSON hop le, khong kem giai thich ben ngoai:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra kho skills/...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve 1 JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        print(f"   -> Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        # BUOC 2: Forge tu che tao tool con thieu
        failed_tools = []
        for spec in missing:
            ok_forge = await forge_tool(client, spec)
            if not ok_forge:
                failed_tools.append(spec["skill_name"])

        # BUOC 3: Coder chay thuc chien + Data Quality Gate + Self-Healing (toi da 5 luot)
        reg_updated = scan_registry()
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        failed_note = ""
        if failed_tools:
            failed_note = f"\nLUU Y QUAN TRONG: Cac tool sau CHUA DUOC TAO nen KHONG import: {failed_tools}. Hay goi truc tiep API mo (Google News RSS / ArXiv / Open-Meteo / Wikipedia...) bang `httpx.get` trong script nay!"

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
Kho skill hien co (chi import cac module CO MAT trong danh sach nay):
{reg_updated}
{failed_note}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau BAT BUOC:
- Lay day du du lieu cho TAT CA cac ve cua cau hoi.
- TUYET DOI KHONG dung URL gia nhu `example.com`; phai goi API/RSS that bang `httpx.get(..., follow_redirects=True)`.
- KHONG nuot loi im lang thanh 'N/A': in thang ket qua tho hoac traceback loi ra stdout/stderr.
- Neu moi muc co doan tom tat (abstract/summary) qua dai, cat gon moi tom tat toi da ~500 ky tu de in day du TAT CA cac muc va link nguon trong ~6000 ky tu.
- Chi tra ve 1 block code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve 1 block code Python.", run_p, 0.2))

        collected_data = ""
        heal_counts: dict[str, int] = {}
        max_attempts = 5
        for attempt in range(1, max_attempts + 1):
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)

            fail_reason = summary
            if ok and len(out.strip()) > 80:
                gate_ok, missing_info = await verify_collected_data(client, question, out)
                if gate_ok:
                    collected_data = out[:8000]
                    print(f"   -> [OK lan {attempt}] Thu thap DU du lieu ({len(collected_data)} ky tu)!")
                    print(f"   -> [PREVIEW] {collected_data[:220].replace(chr(10), ' | ')}...")
                    break
                else:
                    fail_reason = f"DATA_INCOMPLETE: {missing_info}"
                    print(f"   -> [THIEU DU LIEU lan {attempt}] {missing_info}")
                    print(f"      [RAW OUT] {out[:200].replace(chr(10), ' | ')}")
            else:
                print(f"   -> [FAIL lan {attempt}] {summary}")

            if attempt == max_attempts:
                collected_data = out[-4000:]
                break

            diag_p = f"""Script thu thap du lieu bi loi hoac thieu du lieu khi chay thuc te.
Van de: {fail_reason}
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Output / Log:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi do `live_runner.py` (goi sai ham, import tool khong ton tai, dung URL gia): tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": fail_reason}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath) and heal_counts.get(sname, 0) < 2:
                    heal_counts[sname] = heal_counts.get(sname, 0) + 1
                    print(f"   -> [SELF-HEAL #{heal_counts[sname]}] Va `{sname}.py` ({diag.get('reason')})...")
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc van de chay thuc te: {diag.get('reason')}

Code hien tai cua `{sname}.py`:
{FENCE}python
{old_src}
{FENCE}

Cach `live_runner.py` dang goi va doc ket qua tu tool nay:
{FENCE}python
{runner_code}
{FENCE}

Output / Log thuc te:
{out[-1500:]}

QUY TAC SUA BAT BUOC:
1. Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})`.
2. TUYET DOI KHONG nuot loi im lang tra ve rong/None; neu API loi phai raise RuntimeError kem status_code va response text.
3. Tra ve dung dinh dang/key ma `live_runner.py` dang mong doi. Giu nguyen ten class/ham public.
Chi tra ve 1 block code Python cho `{sname}.py`."""
                    patched = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi tool Python. Chi tra ve 1 block code.", patch_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )
                    with open(spath, "w", encoding="utf-8") as f:
                        f.write(patched)
                    continue

            fix_run_p = f"""Sua `live_runner.py` de lay day du du lieu that va in ra stdout (KHONG dung example.com, KHONG nuot loi thanh 'N/A', cat gon moi tom tat toi da ~500 ky tu).
Van de: {diag.get('reason')}
Kho skill hien co:
{reg_updated}
Code cu:
{FENCE}python
{runner_code}
{FENCE}
Output/Log:
{out[-1500:]}"""
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code.", fix_run_p, 0.2))

        # BUOC 4: Analyst (qwq:32b) tong hop bao cao
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""THOI GIAN HIEN TAI CUA HE THONG: {vn_now} (nam 2026 la thoi diem hien tai that, KHONG phai tuong lai hay gia lap).
Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau BAT BUOC:
- Tra loi chinh xac, ro rang bang tieng Viet chuan (TUYET DOI khong chen chu Han/tieng Trung).
- Chi dua tren du lieu thu thap o tren; phan nao khong co du lieu thi noi ro la chua thu thap duoc.
- KHONG bao gio tu bia link URL; chi trich dan cac link co mat trong du lieu thu thap."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {round(time.perf_counter() - t0, 2)}s\n")
        print("================ TRA LOI ================")
        print(final_ans)


if __name__ == "__main__":
    q = " ".join(sys.argv[1:]).strip()
    if not q:
        print('Dung: python3 live_evolve_agent.py "Cau hoi cua ban"')
        sys.exit(1)
    asyncio.run(solve_question(q))
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import asyncio
import subprocess
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"
BACKUP_DIR = f"{BASE_DIR}/skill_backups"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str, module_name: str | None = None, is_test: bool = False) -> str:
    blocks = re.findall(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    if not blocks and FENCE in text:
        part = text.split(FENCE, 1)[1]
        part = re.sub(r"^python\s*", "", part, flags=re.IGNORECASE)
        blocks = [part]
    if not blocks:
        blocks = [text]

    valid = []
    for b in blocks:
        code = b.strip()
        if not code:
            continue
        if module_name and not is_test:
            if re.search(rf"\b(from|import)\s+{re.escape(module_name)}\b", code):
                continue
            if "def test_" in code and "class " not in code:
                continue
        try:
            ast.parse(code)
            valid.append(code)
        except SyntaxError:
            continue

    if valid:
        return max(valid, key=len)
    return blocks[0].strip()


def extract_json(text: str) -> dict:
    candidates = re.findall(FENCE + r"(?:json)?\s*(\{.*?\})\s*" + FENCE, text, re.DOTALL)
    candidates.append(text)
    decoder = json.JSONDecoder(strict=False)
    for chunk in candidates:
        pos = 0
        while True:
            idx = chunk.find("{", pos)
            if idx == -1:
                break
            try:
                obj, _ = decoder.raw_decode(chunk, idx)
                if isinstance(obj, dict):
                    return obj
            except json.JSONDecodeError:
                pass
            pos = idx + 1
    raise ValueError(f"Khong tim thay JSON hop le trong: {text[:200]}")


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def scan_registry() -> str:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    lines = []
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        try:
            tree = ast.parse(open(f"{SKILLS_DIR}/{fn}", encoding="utf-8").read())
            sigs = []
            for n in tree.body:
                if isinstance(n, ast.ClassDef):
                    ms = [
                        f"{m.name}({', '.join(a.arg for a in m.args.args if a.arg != 'self')})"
                        for m in n.body
                        if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef))
                        and (not m.name.startswith("_") or m.name == "__init__")
                    ]
                    sigs.append(f"class {n.name}: [{', '.join(ms)}]")
                elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                    sigs.append(f"def {n.name}({', '.join(a.arg for a in n.args.args)})")
            lines.append(f"- `{mod}`: " + " | ".join(sigs))
        except Exception:
            lines.append(f"- `{mod}`")
    return "\n".join(lines) if lines else "(Kho dang trong)"


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:{BASE_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:160] if errs else (out.replace("\n", " | ")[:160] if out else "Output rong")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


def backup_skill(sname: str) -> str | None:
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return None
    os.makedirs(BACKUP_DIR, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    bpath = f"{BACKUP_DIR}/{sname}_{ts}.py.bak"
    with open(spath, "r", encoding="utf-8") as src, open(bpath, "w", encoding="utf-8") as dst:
        dst.write(src.read())
    return bpath


def restore_skill(sname: str, bpath: str | None):
    spath = f"{SKILLS_DIR}/{sname}.py"
    if bpath and os.path.exists(bpath):
        with open(bpath, "r", encoding="utf-8") as src, open(spath, "w", encoding="utf-8") as dst:
            dst.write(src.read())
        print(f"     -> [ROLLBACK] Da khoi phuc ban goc cua `{sname}.py` do ban va chua dat.")


def verify_skill_pytest(sname: str) -> tuple[bool, str]:
    tpath = f"{SKILLS_DIR}/test_{sname}.py"
    if not os.path.exists(tpath):
        return True, "Khong co file test rieng"
    vdir = f"{SANDBOX_DIR}/heal_verify"
    os.makedirs(vdir, exist_ok=True)
    with open(f"{vdir}/test_{sname}.py", "w", encoding="utf-8") as src, open(f"{vdir}/test_{sname}.py", "w") as _:
        pass
    import shutil
    shutil.copy2(tpath, f"{vdir}/test_{sname}.py")
    shutil.copy2(f"{SKILLS_DIR}/{sname}.py", f"{vdir}/{sname}.py")
    return run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], vdir, 20)[0], "pytest"


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong (cho phep them tham so tuy chon co gia tri mac dinh)?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def verify_collected_data(client: httpx.AsyncClient, question: str, data: str) -> tuple[bool, str]:
    ver_p = f"""Cau hoi nguoi dung: {question}

Du lieu script vua thu thap duoc:
{data[:8000]}

Kiem tra: Du lieu tren da cung cap du thong tin thuc te de tra loi TAT CA cac ve cua cau hoi chua?
(Neu thieu bat ky ve nao, hoac bi loi/rong/N/A/example.com/404 -> FAIL).
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
MISSING: <neu FAIL, ghi ro phan nao bi thieu hoac rong; neu PASS ghi None>"""
    res = await ask_llm(client, BRAIN_THINK, "Ban la QA Data Verifier nghiem khac.", ver_p, 0.1)
    ok = "STATUS: PASS" in res.upper()
    m = re.search(r"MISSING:\s*(.+)", res)
    return ok, (m.group(1).strip() if m else res.splitlines()[-1][:120])


async def forge_tool(client: httpx.AsyncClient, spec: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3 test case ro rang (1 happy path + 1 loi HTTP status != 200 + 1 loi du lieu rong/mang).
- Code thuc thi dung `httpx.get(...)`, nen mock bang `unittest.mock.patch("httpx.get")` (KHONG mock `requests`).
- Khi test truong hop loi, LUON dung `with pytest.raises(Exception):` (KHONG ep loai exception cu the nhu ValueError/HTTPError de tranh lech voi RuntimeError).
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve duy nhat 1 block code Python cho `test_{sname}.py`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- LUON dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})` (KHONG dung `requests`).
- KHONG nuot loi mang im lang (neu HTTP status != 200 hoac du lieu loi, raise `RuntimeError` kem status_code va noi dung loi).
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY:
  * Thoi tiet: Open-Meteo (`https://geocoding-api.open-meteo.com/v1/search` + `https://api.open-meteo.com/v1/forecast`).
  * Wikipedia: MediaWiki API (`https://vi.wikipedia.org/w/api.php` va `en.wikipedia.org/w/api.php`).
  * Khoa hoc: ArXiv (`https://export.arxiv.org/api/query`, parse bang `xml.etree.ElementTree`) hoac OpenAlex (`https://api.openalex.org/works`).
  * Tin tuc/Tim kiem web: Google News RSS (`https://news.google.com/rss/search?q=...&hl=vi&gl=VN&ceid=VN:vi`, parse XML bang `xml.etree.ElementTree`) hoac `ddgs`.
  * Doc noi dung trang web: `httpx.get(..., follow_redirects=True)` + loc text sach (bo `<script>`, `<style>`, `<tag>`).
  * Tai chinh/Ty gia: Frankfurter (`https://api.frankfurter.app/latest`) hoac Binance public (`https://api.binance.com/api/v3/ticker/price`).
- Tuyet doi khong hardcode du lieu gia, khong viet code test trong file nay.
Chi tra ve duy nhat 1 block code Python cho `{sname}.py`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve 1 block code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0], module_name=sname, is_test=True)
    candidates = [extract_code(c, module_name=sname, is_test=False) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                        f.write(curr)
                    with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                        f.write(test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"""Sua file `{sname}.py` de pass pytest (van dung `httpx.get(..., follow_redirects=True)` va API that).
CHI tra ve code cua `{sname}.py`, KHONG tra ve file test.
Code:
{FENCE}python
{curr}
{FENCE}
Test:
{FENCE}python
{test_code}
{FENCE}
Log:
{log[-1500:]}"""
                    curr = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code cho file skill.", fix_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )

    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, ep sai loai exception, sai cau truc response):
Tra ve `VERDICT: TEST_BUG` kem toan bo file `test_{sname}.py` da sua (dung cu phap Python hop le) trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj, module_name=sname, is_test=True)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
                    f.write(last_code0)
                with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
                    f.write(fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str):
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)
    vn_now = datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) phan tich nhu cau & quet kho skills/
        reg = scan_registry()
        plan_p = f"""CAU HOI NGUOI DUNG: {question}
THOI GIAN HIEN TAI: {vn_now}

KHO SKILL HIEN CO:
{reg}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
2. Neu THIEU tool, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...).
   - Thoi tiet -> dung Open-Meteo.
   - Kien thuc/Lich su -> dung MediaWiki / Wikipedia API.
   - Khoa hoc -> dung ArXiv API (`https://export.arxiv.org/api/query`) hoac OpenAlex (`https://api.openalex.org/works`).
   - Tin tuc/Tim kiem web -> dung Google News RSS (`https://news.google.com/rss/search`) hoac `ddgs`.
   - Tai chinh/Ty gia -> dung Frankfurter hoac Binance public API.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT 1 JSON hop le, khong kem giai thich ben ngoai:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra kho skills/...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve 1 JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        print(f"   -> Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        # BUOC 2: Forge tu che tao tool con thieu
        failed_tools = []
        for spec in missing:
            ok_forge = await forge_tool(client, spec)
            if not ok_forge:
                failed_tools.append(spec["skill_name"])

        # BUOC 3: Coder chay thuc chien + Data Quality Gate + Self-Healing + Rollback (toi da 5 luot)
        reg_updated = scan_registry()
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        failed_note = ""
        if failed_tools:
            failed_note = f"\nLUU Y QUAN TRONG: Cac tool sau CHUA DUOC TAO nen KHONG import: {failed_tools}. Hay goi truc tiep API mo / RSS / web bang `httpx.get(..., follow_redirects=True)` trong script nay!"

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
Kho skill hien co (chi import cac module CO MAT trong danh sach nay):
{reg_updated}
{failed_note}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau BAT BUOC:
- Lay day du du lieu cho TAT CA cac ve cua cau hoi.
- TUYET DOI KHONG dung URL gia nhu `example.com`; phai goi API/RSS/trang web that bang `httpx.get(..., follow_redirects=True)`.
- KHONG nuot loi im lang thanh 'N/A': in thang ket qua tho hoac traceback loi ra stdout/stderr.
- Neu moi muc co doan tom tat qua dai, cat gon moi tom tat toi da ~500 ky tu de in day du TAT CA cac muc va link nguon trong ~6000 ky tu.
- Chi tra ve 1 block code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve 1 block code Python.", run_p, 0.2))

        collected_data = ""
        heal_counts: dict[str, int] = {}
        max_attempts = 5
        for attempt in range(1, max_attempts + 1):
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)

            fail_reason = summary
            if ok and len(out.strip()) > 80:
                gate_ok, missing_info = await verify_collected_data(client, question, out)
                if gate_ok:
                    collected_data = out[:8000]
                    print(f"   -> [OK lan {attempt}] Thu thap DU du lieu ({len(collected_data)} ky tu)!")
                    print(f"   -> [PREVIEW] {collected_data[:220].replace(chr(10), ' | ')}...")
                    break
                else:
                    fail_reason = f"DATA_INCOMPLETE: {missing_info}"
                    print(f"   -> [THIEU DU LIEU lan {attempt}] {missing_info}")
                    print(f"      [RAW OUT] {out[:200].replace(chr(10), ' | ')}")
            else:
                print(f"   -> [FAIL lan {attempt}] {summary}")

            if attempt == max_attempts:
                collected_data = out[-4000:]
                break

            diag_p = f"""Script thu thap du lieu bi loi hoac thieu du lieu khi chay thuc te.
Van de: {fail_reason}
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Output / Log:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi do `live_runner.py` (goi sai ham, import tool khong ton tai, dung URL gia): tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": fail_reason}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath) and heal_counts.get(sname, 0) < 2:
                    heal_counts[sname] = heal_counts.get(sname, 0) + 1
                    print(f"   -> [SELF-HEAL #{heal_counts[sname]}] Va `{sname}.py` ({diag.get('reason')})...")
                    bpath = backup_skill(sname)
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc van de chay thuc te: {diag.get('reason')}

Code hien tai cua `{sname}.py`:
{FENCE}python
{old_src}
{FENCE}

Cach `live_runner.py` dang goi va doc ket qua tu tool nay:
{FENCE}python
{runner_code}
{FENCE}

Output / Log thuc te:
{out[-1500:]}

QUY TAC SUA BAT BUOC:
1. Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})`.
2. TUYET DOI KHONG nuot loi im lang tra ve rong/None; neu API loi phai raise RuntimeError kem status_code va response text.
3. Tra ve dung dinh dang/key ma `live_runner.py` dang mong doi. Giu nguyen ten class/ham public.
Chi tra ve 1 block code Python cho `{sname}.py`."""
                    patched = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi tool Python. Chi tra ve 1 block code.", patch_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )
                    try:
                        ast.parse(patched)
                        with open(spath, "w", encoding="utf-8") as f:
                            f.write(patched)
                        pt_ok, _ = verify_skill_pytest(sname)
                        if not pt_ok:
                            print(f"     -> Ban va `{sname}.py` khong qua pytest.")
                            restore_skill(sname, bpath)
                    except SyntaxError:
                        print(f"     -> Ban va `{sname}.py` loi cu phap.")
                        restore_skill(sname, bpath)
                    continue

            fix_run_p = f"""Sua `live_runner.py` de lay day du du lieu that va in ra stdout (KHONG dung example.com, KHONG nuot loi thanh 'N/A', cat gon moi tom tat toi da ~500 ky tu).
Van de: {diag.get('reason')}
Kho skill hien co:
{reg_updated}
Code cu:
{FENCE}python
{runner_code}
{FENCE}
Output/Log:
{out[-1500:]}"""
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code.", fix_run_p, 0.2))

        # BUOC 4: Analyst (qwq:32b) tong hop bao cao
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""THOI GIAN HIEN TAI CUA HE THONG: {vn_now} (nam 2026 la thoi diem hien tai that, KHONG phai tuong lai hay gia lap).
Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau BAT BUOC:
- Tra loi chinh xac, ro rang bang tieng Viet chuan (TUYET DOI khong chen chu Han/tieng Trung).
- Chi dua tren du lieu thu thap o tren; phan nao khong co du lieu thi noi ro la chua thu thap duoc.
- KHONG bao gio tu bia link URL; chi trich dan cac link co mat trong du lieu thu thap."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {round(time.perf_counter() - t0, 2)}s\n")
        print("================ TRA LOI ================")
        print(final_ans)


if __name__ == "__main__":
    q = " ".join(sys.argv[1:]).strip()
    if not q:
        print('Dung: python3 live_evolve_agent.py "Cau hoi cua ban"')
        sys.exit(1)
    asyncio.run(solve_question(q))
````

Dưới đây là trọn bộ 2 phần nâng cấp: (1) Bộ 2 skill web nền tảng kèm unit test (skill_web_search + skill_web_reader) và (2) Bản nâng cấp live_evolve_agent.py có registry_meta.json, lọc skill cho Coder, tự dọn sandbox/ + xoay vòng tối đa 5 backup/skill, ghi log runs.jsonl, và chế độ --health-check.

````python
# Tạo skill_web_search.py và test_skill_web_search.py
#Dùng thuần httpx.get + thư viện chuẩn Python (không lệ thuộc gói ngoài): ưu tiên tìm web qua DuckDuckGo HTML, tự động fallback sang Google News RSS nếu bị chặn hoặc không có kết quả.

%%writefile /content/drive/MyDrive/AutoEvolve_AI/skills/skill_web_search.py
"""Zero-key web and news search client using DuckDuckGo HTML with Google News RSS fallback."""
import re
import html
import urllib.parse
import xml.etree.ElementTree as ET
import httpx


class WebSearchClient:
    """Search the web and news without an API key."""

    def __init__(self, timeout: float = 15.0):
        self.timeout = timeout
        self.headers = {
            "User-Agent": (
                "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
                "(KHTML, like Gecko) Chrome/128.0.0.0 Safari/537.36"
            )
        }

    def _clean_html(self, raw: str) -> str:
        text = re.sub(r"<[^>]+>", " ", raw or "")
        return re.sub(r"\s+", " ", html.unescape(text)).strip()

    def _extract_ddg_url(self, href: str) -> str:
        href = html.unescape(href)
        if "uddg=" in href:
            parsed = urllib.parse.urlparse(href)
            qs = urllib.parse.parse_qs(parsed.query)
            if "uddg" in qs and qs["uddg"]:
                return qs["uddg"][0]
        if href.startswith("//"):
            return "https:" + href
        return href

    def _search_ddg(self, query: str, max_results: int) -> list[dict]:
        url = "https://html.duckduckgo.com/html/"
        r = httpx.get(
            url,
            params={"q": query},
            headers=self.headers,
            timeout=self.timeout,
            follow_redirects=True,
        )
        if r.status_code != 200:
            raise RuntimeError(f"DuckDuckGo HTTP {r.status_code}: {r.text[:200]}")

        body = r.text
        links = re.findall(
            r'<a[^>]*class="[^"]*result__a[^"]*"[^>]*href="([^"]+)"[^>]*>(.*?)</a>',
            body,
            re.DOTALL | re.IGNORECASE,
        )
        snippets = re.findall(
            r'<(?:a|div)[^>]*class="[^"]*result__snippet[^"]*"[^>]*>(.*?)</(?:a|div)>',
            body,
            re.DOTALL | re.IGNORECASE,
        )

        results = []
        for idx, (href, raw_title) in enumerate(links[:max_results]):
            title = self._clean_html(raw_title)
            clean_url = self._extract_ddg_url(href)
            snippet = self._clean_html(snippets[idx]) if idx < len(snippets) else ""
            if title and clean_url.startswith("http"):
                results.append(
                    {
                        "title": title,
                        "url": clean_url,
                        "snippet": snippet,
                        "source": "duckduckgo",
                    }
                )
        return results

    def _search_google_news_rss(self, query: str, max_results: int, lang: str = "vi") -> list[dict]:
        hl = "vi" if lang == "vi" else "en-US"
        gl = "VN" if lang == "vi" else "US"
        ceid = "VN:vi" if lang == "vi" else "US:en"
        r = httpx.get(
            "https://news.google.com/rss/search",
            params={"q": query, "hl": hl, "gl": gl, "ceid": ceid},
            headers=self.headers,
            timeout=self.timeout,
            follow_redirects=True,
        )
        if r.status_code != 200:
            raise RuntimeError(f"Google News RSS HTTP {r.status_code}: {r.text[:200]}")

        root = ET.fromstring(r.text)
        results = []
        for item in root.findall(".//item")[:max_results]:
            title = (item.findtext("title") or "").strip()
            link = (item.findtext("link") or "").strip()
            pub_date = (item.findtext("pubDate") or "").strip()
            desc = self._clean_html(item.findtext("description") or "")
            if title and link:
                results.append(
                    {
                        "title": title,
                        "url": link,
                        "snippet": f"{pub_date} - {desc}".strip(" -"),
                        "source": "google_news_rss",
                    }
                )
        return results

    def search(self, query: str, max_results: int = 5, lang: str = "vi") -> list[dict]:
        if not query or not query.strip():
            raise ValueError("Query khong duoc rong")
        errors = []
        try:
            res = self._search_ddg(query, max_results)
            if res:
                return res
        except Exception as e:
            errors.append(f"DDG: {e}")

        try:
            res = self._search_google_news_rss(query, max_results, lang=lang)
            if res:
                return res
        except Exception as e:
            errors.append(f"RSS: {e}")

        if errors:
            raise RuntimeError(" | ".join(errors))
        raise RuntimeError(f"Khong tim thay ket qua nao cho: {query}")
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/skills/test_skill_web_search.py
from unittest.mock import MagicMock, patch
import pytest
from skill_web_search import WebSearchClient


def test_search_ddg_happy_path():
    fake_html = '''
    <div class="result">
      <a class="result__a" href="https://duckduckgo.com/l/?uddg=https%3A%2F%2Fexample.org%2Fai">AI News</a>
      <a class="result__snippet">Latest breakthrough in AI agents.</a>
    </div>
    '''
    mock_resp = MagicMock(status_code=200, text=fake_html)
    with patch("httpx.get", return_value=mock_resp):
        client = WebSearchClient()
        res = client.search("AI agents", max_results=3)
        assert len(res) == 1
        assert res[0]["title"] == "AI News"
        assert res[0]["url"] == "https://example.org/ai"
        assert "breakthrough" in res[0]["snippet"]


def test_search_fallback_to_rss():
    ddg_fail = MagicMock(status_code=403, text="Forbidden")
    rss_xml = '''<?xml version="1.0" encoding="UTF-8"?>
    <rss><channel><item>
      <title>RSS Headline</title>
      <link>https://news.example.org/1</link>
      <pubDate>Wed, 25 Nov 2026 03:00:00 GMT</pubDate>
      <description>Short summary</description>
    </item></channel></rss>'''
    rss_ok = MagicMock(status_code=200, text=rss_xml)
    with patch("httpx.get", side_effect=[ddg_fail, rss_ok]):
        client = WebSearchClient()
        res = client.search("AI", max_results=2)
        assert len(res) == 1
        assert res[0]["source"] == "google_news_rss"
        assert res[0]["title"] == "RSS Headline"


def test_search_all_backends_error():
    fail_resp = MagicMock(status_code=500, text="Server Error")
    with patch("httpx.get", return_value=fail_resp):
        client = WebSearchClient()
        with pytest.raises(Exception):
            client.search("AI")
````

````python
# Tạo skill_web_reader.py và test_skill_web_reader.py
# Tải bất kỳ URL nào bằng httpx.get(..., follow_redirects=True), tự loại bỏ <script>, <style>, <nav>, <header>, <footer>, <iframe> và trả về text sạch kèm tiêu đề.

%%writefile /content/drive/MyDrive/AutoEvolve_AI/skills/skill_web_reader.py
"""Zero-key webpage reader that fetches a URL and extracts clean readable text."""
import re
from html.parser import HTMLParser
import httpx


class _CleanTextExtractor(HTMLParser):
    SKIP_TAGS = {"script", "style", "noscript", "svg", "nav", "footer", "header", "iframe", "aside"}

    def __init__(self):
        super().__init__()
        self._skip_depth = 0
        self._in_title = False
        self.title_parts: list[str] = []
        self.text_parts: list[str] = []

    def handle_starttag(self, tag, attrs):
        tag = tag.lower()
        if tag in self.SKIP_TAGS:
            self._skip_depth += 1
        elif tag == "title":
            self._in_title = True

    def handle_endtag(self, tag):
        tag = tag.lower()
        if tag in self.SKIP_TAGS and self._skip_depth > 0:
            self._skip_depth -= 1
        elif tag == "title":
            self._in_title = False
        elif tag in {"p", "div", "li", "h1", "h2", "h3", "h4", "br", "tr"} and self._skip_depth == 0:
            self.text_parts.append("\n")

    def handle_data(self, data):
        if self._in_title:
            self.title_parts.append(data)
        if self._skip_depth == 0:
            cleaned = data.strip()
            if cleaned:
                self.text_parts.append(cleaned + " ")


class WebReaderClient:
    """Fetch any webpage and extract clean title and body text."""

    def __init__(self, timeout: float = 15.0):
        self.timeout = timeout
        self.headers = {
            "User-Agent": (
                "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
                "(KHTML, like Gecko) Chrome/128.0.0.0 Safari/537.36"
            )
        }

    def read_url(self, url: str, max_chars: int = 4000) -> dict:
        if not url or not url.startswith(("http://", "https://")):
            raise ValueError(f"URL khong hop le: {url}")

        r = httpx.get(
            url,
            headers=self.headers,
            timeout=self.timeout,
            follow_redirects=True,
        )
        if r.status_code != 200:
            raise RuntimeError(f"HTTP {r.status_code} khi tai {url}: {r.text[:200]}")

        parser = _CleanTextExtractor()
        parser.feed(r.text)
        title = re.sub(r"\s+", " ", "".join(parser.title_parts)).strip()
        raw_text = "".join(parser.text_parts)
        clean_text = re.sub(r"[ \t]+", " ", raw_text)
        clean_text = re.sub(r"\n{3,}", "\n\n", clean_text).strip()

        if not clean_text:
            raise RuntimeError(f"Khong trich xuat duoc noi dung van ban tu {url}")

        return {
            "url": url,
            "final_url": str(r.url),
            "title": title,
            "text": clean_text[:max_chars],
            "status_code": r.status_code,
        }
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import shutil
import asyncio
import argparse
import subprocess
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"
BACKUP_DIR = f"{BASE_DIR}/skill_backups"
META_PATH = f"{BASE_DIR}/registry_meta.json"
RUNS_LOG_PATH = f"{BASE_DIR}/runs.jsonl"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3
MAX_BACKUPS_PER_SKILL = 5
CORE_WEB_SKILLS = {"skill_web_search", "skill_web_reader"}


def now_vn_str() -> str:
    return datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str, module_name: str | None = None, is_test: bool = False) -> str:
    blocks = re.findall(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    if not blocks and FENCE in text:
        part = text.split(FENCE, 1)[1]
        part = re.sub(r"^python\s*", "", part, flags=re.IGNORECASE)
        blocks = [part]
    if not blocks:
        blocks = [text]

    valid = []
    for b in blocks:
        code = b.strip()
        if not code:
            continue
        if module_name and not is_test:
            if re.search(rf"\b(from|import)\s+{re.escape(module_name)}\b", code):
                continue
            if "def test_" in code and "class " not in code:
                continue
        try:
            ast.parse(code)
            valid.append(code)
        except SyntaxError:
            continue

    if valid:
        return max(valid, key=len)
    return blocks[0].strip()


def extract_json(text: str) -> dict:
    candidates = re.findall(FENCE + r"(?:json)?\s*(\{.*?\})\s*" + FENCE, text, re.DOTALL)
    candidates.append(text)
    decoder = json.JSONDecoder(strict=False)
    for chunk in candidates:
        pos = 0
        while True:
            idx = chunk.find("{", pos)
            if idx == -1:
                break
            try:
                obj, _ = decoder.raw_decode(chunk, idx)
                if isinstance(obj, dict):
                    return obj
            except json.JSONDecodeError:
                pass
            pos = idx + 1
    raise ValueError(f"Khong tim thay JSON hop le trong: {text[:200]}")


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def inspect_skill_file(filepath: str) -> tuple[str, str]:
    """Return (short_description, signature_summary) for a skill file."""
    try:
        src = open(filepath, encoding="utf-8").read()
        tree = ast.parse(src)
        doc = (ast.get_docstring(tree) or "").splitlines()
        desc = doc[0].strip()[:120] if doc else ""
        sigs = []
        for n in tree.body:
            if isinstance(n, ast.ClassDef):
                if not desc:
                    cdoc = (ast.get_docstring(n) or "").splitlines()
                    if cdoc:
                        desc = cdoc[0].strip()[:120]
                ms = [
                    f"{m.name}({', '.join(a.arg for a in m.args.args if a.arg != 'self')})"
                    for m in n.body
                    if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef))
                    and (not m.name.startswith("_") or m.name == "__init__")
                ]
                sigs.append(f"class {n.name}: [{', '.join(ms)}]")
            elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                sigs.append(f"def {n.name}({', '.join(a.arg for a in n.args.args)})")
        return desc or "Python skill module", " | ".join(sigs)
    except Exception as e:
        return f"Parse error: {e}", ""


def load_meta() -> dict:
    if os.path.exists(META_PATH):
        try:
            with open(META_PATH, "r", encoding="utf-8") as f:
                return json.load(f)
        except Exception:
            pass
    return {}


def save_meta(meta: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(META_PATH, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)


def sync_registry_meta() -> dict:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    meta = load_meta()
    existing = set()
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        existing.add(mod)
        desc, sigs = inspect_skill_file(f"{SKILLS_DIR}/{fn}")
        entry = meta.get(mod, {})
        entry["signatures"] = sigs
        if not entry.get("description") or entry["description"] == "Python skill module":
            entry["description"] = desc
        entry.setdefault("success_count", 0)
        entry.setdefault("fail_count", 0)
        entry.setdefault("health", "UNKNOWN")
        entry.setdefault("updated_at", now_vn_str())
        meta[mod] = entry

    for dead in list(meta.keys()):
        if dead not in existing:
            del meta[dead]

    save_meta(meta)
    return meta


def format_registry_for_architect(meta: dict) -> str:
    if not meta:
        return "(Kho dang trong)"
    lines = []
    for mod in sorted(meta.keys()):
        info = meta[mod]
        stats = f"ok={info.get('success_count', 0)}, fail={info.get('fail_count', 0)}, health={info.get('health', 'UNKNOWN')}"
        lines.append(f"- `{mod}` ({stats}): {info.get('description', '')} | {info.get('signatures', '')}")
    return "\n".join(lines)


def format_registry_for_coder(meta: dict, selected: list[str] | None = None) -> str:
    if not meta:
        return "(Kho dang trong)"
    wanted = set()
    if selected:
        for s in selected:
            name = s.replace(".py", "").strip()
            if name in meta:
                wanted.add(name)
    # Luon kem 2 skill web nen tang lam fallback neu co trong kho
    for core in CORE_WEB_SKILLS:
        if core in meta:
            wanted.add(core)
    if not wanted:
        wanted = set(meta.keys())
    lines = [f"- `{m}`: {meta[m].get('signatures', '')}" for m in sorted(wanted)]
    return "\n".join(lines)


def update_skill_stats(meta: dict, skill_names: list[str], success: bool):
    ts = now_vn_str()
    for s in skill_names:
        name = s.replace(".py", "").strip()
        if name in meta:
            key = "success_count" if success else "fail_count"
            meta[name][key] = meta[name].get(key, 0) + 1
            meta[name]["last_used"] = ts
    save_meta(meta)


def append_run_log(record: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(RUNS_LOG_PATH, "a", encoding="utf-8") as f:
        f.write(json.dumps(record, ensure_ascii=False) + "\n")


def cleanup_backups(max_per_skill: int = MAX_BACKUPS_PER_SKILL):
    if not os.path.exists(BACKUP_DIR):
        return
    groups: dict[str, list[str]] = {}
    for fn in os.listdir(BACKUP_DIR):
        if not fn.endswith(".py.bak"):
            continue
        m = re.match(r"^(skill_.+)_\d{8}_\d{6}\.py\.bak$", fn)
        key = m.group(1) if m else fn
        groups.setdefault(key, []).append(os.path.join(BACKUP_DIR, fn))
    for files in groups.values():
        files.sort(key=lambda p: os.path.getmtime(p), reverse=True)
        for old_path in files[max_per_skill:]:
            try:
                os.remove(old_path)
            except OSError:
                pass


def cleanup_sandbox():
    if not os.path.exists(SANDBOX_DIR):
        return
    for name in os.listdir(SANDBOX_DIR):
        if name.startswith(("forge_", "heal_verify", "health_")):
            full = os.path.join(SANDBOX_DIR, name)
            try:
                if os.path.isdir(full):
                    shutil.rmtree(full, ignore_errors=True)
                else:
                    os.remove(full)
            except OSError:
                pass


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:{BASE_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:160] if errs else (out.replace("\n", " | ")[:160] if out else "Output rong")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


def backup_skill(sname: str) -> str | None:
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return None
    os.makedirs(BACKUP_DIR, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    bpath = f"{BACKUP_DIR}/{sname}_{ts}.py.bak"
    shutil.copy2(spath, bpath)
    cleanup_backups()
    return bpath


def restore_skill(sname: str, bpath: str | None):
    spath = f"{SKILLS_DIR}/{sname}.py"
    if bpath and os.path.exists(bpath):
        shutil.copy2(bpath, spath)
        print(f"     -> [ROLLBACK] Da khoi phuc ban goc cua `{sname}.py` do ban va chua dat.")


def verify_skill_pytest(sname: str) -> tuple[bool, str]:
    tpath = f"{SKILLS_DIR}/test_{sname}.py"
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return False, "Khong tim thay file skill"
    if not os.path.exists(tpath):
        return True, "NO_TEST"
    vdir = f"{SANDBOX_DIR}/heal_verify_{sname}"
    os.makedirs(vdir, exist_ok=True)
    try:
        shutil.copy2(tpath, f"{vdir}/test_{sname}.py")
        shutil.copy2(spath, f"{vdir}/{sname}.py")
        ok, _, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], vdir, 20)
        return ok, ("PASS" if ok else summary)
    finally:
        shutil.rmtree(vdir, ignore_errors=True)


def run_health_check():
    print("=== HEALTH CHECK KHO SKILLS ===")
    meta = sync_registry_meta()
    if not meta:
        print("Kho skills/ dang trong.")
        return
    passed = failed = no_test = 0
    for mod in sorted(meta.keys()):
        spath = f"{SKILLS_DIR}/{mod}.py"
        try:
            ast.parse(open(spath, encoding="utf-8").read())
        except SyntaxError as e:
            meta[mod]["health"] = "SYNTAX_ERROR"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> SyntaxError: {e}")
            continue

        ok, msg = verify_skill_pytest(mod)
        if msg == "NO_TEST":
            meta[mod]["health"] = "NO_TEST"
            no_test += 1
            print(f"  [WARN] {mod:<35} -> Khong co file test")
        elif ok:
            meta[mod]["health"] = "PASS"
            passed += 1
            print(f"  [PASS] {mod:<35} -> pytest OK")
        else:
            meta[mod]["health"] = "FAIL"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> {msg}")
        meta[mod]["last_health_check"] = now_vn_str()

    save_meta(meta)
    cleanup_backups()
    cleanup_sandbox()
    print(f"\nTong ket: {passed} PASS | {failed} FAIL | {no_test} NO_TEST (da cap nhat registry_meta.json & don rac).")


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong (cho phep them tham so tuy chon co gia tri mac dinh)?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def verify_collected_data(client: httpx.AsyncClient, question: str, data: str) -> tuple[bool, str]:
    ver_p = f"""Cau hoi nguoi dung: {question}

Du lieu script vua thu thap duoc:
{data[:8000]}

Kiem tra: Du lieu tren da cung cap du thong tin thuc te de tra loi TAT CA cac ve cua cau hoi chua?
(Neu thieu bat ky ve nao, hoac bi loi/rong/N/A/example.com/404 -> FAIL).
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
MISSING: <neu FAIL, ghi ro phan nao bi thieu hoac rong; neu PASS ghi None>"""
    res = await ask_llm(client, BRAIN_THINK, "Ban la QA Data Verifier nghiem khac.", ver_p, 0.1)
    ok = "STATUS: PASS" in res.upper()
    m = re.search(r"MISSING:\s*(.+)", res)
    return ok, (m.group(1).strip() if m else res.splitlines()[-1][:120])


async def forge_tool(client: httpx.AsyncClient, spec: dict, meta: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3 test case ro rang (1 happy path + 1 loi HTTP status != 200 + 1 loi du lieu rong/mang).
- Code thuc thi dung `httpx.get(...)`, nen mock bang `unittest.mock.patch("httpx.get")` (KHONG mock `requests`).
- Khi test truong hop loi, LUON dung `with pytest.raises(Exception):`.
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve duy nhat 1 block code Python cho `test_{sname}.py`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- LUON dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})` (KHONG dung `requests`).
- KHONG nuot loi mang im lang (neu HTTP status != 200 hoac du lieu loi, raise `RuntimeError` kem status_code va noi dung loi).
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY (Open-Meteo, MediaWiki/Wikipedia, ArXiv, OpenAlex, Google News RSS, Frankfurter, Binance public).
- Tuyet doi khong hardcode du lieu gia, khong viet code test trong file nay.
Chi tra ve duy nhat 1 block code Python cho `{sname}.py`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve 1 block code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0], module_name=sname, is_test=True)
    candidates = [extract_code(c, module_name=sname, is_test=False) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    def save_forged(code_str: str, test_str: str):
        with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(code_str)
        with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_str)
        _, sigs = inspect_skill_file(f"{SKILLS_DIR}/{sname}.py")
        meta[sname] = {
            "description": goal[:120],
            "signatures": sigs,
            "success_count": 0,
            "fail_count": 0,
            "health": "PASS",
            "updated_at": now_vn_str(),
        }
        save_meta(meta)

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    save_forged(curr, test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"""Sua file `{sname}.py` de pass pytest (van dung `httpx.get(..., follow_redirects=True)` va API that).
CHI tra ve code cua `{sname}.py`, KHONG tra ve file test.
Code:
{FENCE}python
{curr}
{FENCE}
Test:
{FENCE}python
{test_code}
{FENCE}
Log:
{log[-1500:]}"""
                    curr = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code cho file skill.", fix_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )

    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, ep sai loai exception, sai cau truc response):
Tra ve `VERDICT: TEST_BUG` kem toan bo file `test_{sname}.py` da sua trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj, module_name=sname, is_test=True)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                save_forged(last_code0, fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str, dry_run: bool = False, no_forge: bool = False):
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)
    cleanup_sandbox()
    cleanup_backups()
    vn_now = now_vn_str()
    meta = sync_registry_meta()

    forged_ok: list[str] = []
    failed_tools: list[str] = []
    healed_skills: list[str] = []
    run_success = False

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) phan tich nhu cau & doc registry_meta.json
        reg_arch = format_registry_for_architect(meta)
        plan_p = f"""CAU HOI NGUOI DUNG: {question}
THOI GIAN HIEN TAI: {vn_now}

KHO SKILL HIEN CO (tu registry_meta.json):
{reg_arch}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
   (Luu y: da co `skill_web_search` va `skill_web_reader` de tim kiem & doc bat ky trang web nao; chi tao tool moi neu can goi 1 API chuyen biet chua co).
2. Neu THIEU tool chuyen biet, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...).
   - Thoi tiet -> Open-Meteo; Kien thuc -> MediaWiki/Wikipedia; Khoa hoc -> ArXiv/OpenAlex; Tai chinh -> Frankfurter/Binance public; Web/Tin tuc -> dung san `skill_web_search` + `skill_web_reader`.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT 1 JSON hop le, khong kem giai thich ben ngoai:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra registry_meta.json...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve 1 JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        use_skills = plan.get("use_skills", [])
        print(f"   -> Dung skills: {use_skills} | Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        if dry_run:
            print("\n[DRY-RUN] Ke hoach Architect:")
            print(json.dumps(plan, ensure_ascii=False, indent=2))
            return

        # BUOC 2: Forge tu che tao tool con thieu
        if not no_forge:
            for spec in missing:
                ok_forge = await forge_tool(client, spec, meta)
                if ok_forge:
                    sname = spec["skill_name"]
                    if not sname.startswith("skill_"):
                        sname = "skill_" + sname.lstrip("_")
                    forged_ok.append(sname)
                else:
                    failed_tools.append(spec["skill_name"])
        elif missing:
            print(f"   -> [--no-forge] Bo qua tao moi: {[m['skill_name'] for m in missing]}")
            failed_tools = [m["skill_name"] for m in missing]

        # BUOC 3: Coder chay thuc chien (chi nhan dung skills lien quan) + Data Quality Gate + Self-Healing + Rollback
        meta = sync_registry_meta()
        selected_for_coder = list(dict.fromkeys(use_skills + forged_ok))
        reg_coder = format_registry_for_coder(meta, selected_for_coder)
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        failed_note = ""
        if failed_tools:
            failed_note = f"\nLUU Y QUAN TRONG: Cac tool sau CHUA DUOC TAO nen KHONG import: {failed_tools}. Hay dung `skill_web_search` / `skill_web_reader` hoac goi truc tiep API mo bang `httpx.get(..., follow_redirects=True)`."

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
Cac skill duoc phep import (chi import cac module CO MAT trong danh sach nay):
{reg_coder}
{failed_note}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau BAT BUOC:
- Lay day du du lieu cho TAT CA cac ve cua cau hoi.
- TUYET DOI KHONG dung URL gia nhu `example.com`; phai dung skill hoac goi API/trang web that.
- KHONG nuot loi im lang thanh 'N/A': in thang ket qua tho hoac traceback loi ra stdout/stderr.
- Cat gon moi doan tom tat toi da ~500 ky tu de in day du TAT CA cac muc va link nguon trong ~6000 ky tu.
- Chi tra ve 1 block code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve 1 block code Python.", run_p, 0.2))

        collected_data = ""
        heal_counts: dict[str, int] = {}
        max_attempts = 5
        for attempt in range(1, max_attempts + 1):
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)

            fail_reason = summary
            if ok and len(out.strip()) > 80:
                gate_ok, missing_info = await verify_collected_data(client, question, out)
                if gate_ok:
                    collected_data = out[:8000]
                    run_success = True
                    print(f"   -> [OK lan {attempt}] Thu thap DU du lieu ({len(collected_data)} ky tu)!")
                    print(f"   -> [PREVIEW] {collected_data[:220].replace(chr(10), ' | ')}...")
                    break
                else:
                    fail_reason = f"DATA_INCOMPLETE: {missing_info}"
                    print(f"   -> [THIEU DU LIEU lan {attempt}] {missing_info}")
                    print(f"      [RAW OUT] {out[:200].replace(chr(10), ' | ')}")
            else:
                print(f"   -> [FAIL lan {attempt}] {summary}")

            if attempt == max_attempts:
                collected_data = out[-4000:]
                break

            diag_p = f"""Script thu thap du lieu bi loi hoac thieu du lieu khi chay thuc te.
Van de: {fail_reason}
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Output / Log:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi do `live_runner.py` (goi sai ham, import tool khong ton tai, dung URL gia): tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": fail_reason}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath) and heal_counts.get(sname, 0) < 2:
                    heal_counts[sname] = heal_counts.get(sname, 0) + 1
                    print(f"   -> [SELF-HEAL #{heal_counts[sname]}] Va `{sname}.py` ({diag.get('reason')})...")
                    bpath = backup_skill(sname)
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc van de chay thuc te: {diag.get('reason')}

Code hien tai cua `{sname}.py`:
{FENCE}python
{old_src}
{FENCE}

Cach `live_runner.py` dang goi va doc ket qua tu tool nay:
{FENCE}python
{runner_code}
{FENCE}

Output / Log thuc te:
{out[-1500:]}

QUY TAC SUA BAT BUOC:
1. Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})`.
2. TUYET DOI KHONG nuot loi im lang tra ve rong/None; neu API loi phai raise RuntimeError kem status_code va response text.
3. Tra ve dung dinh dang/key ma `live_runner.py` dang mong doi. Giu nguyen ten class/ham public.
Chi tra ve 1 block code Python cho `{sname}.py`."""
                    patched = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi tool Python. Chi tra ve 1 block code.", patch_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )
                    try:
                        ast.parse(patched)
                        with open(spath, "w", encoding="utf-8") as f:
                            f.write(patched)
                        pt_ok, pt_msg = verify_skill_pytest(sname)
                        if not pt_ok:
                            print(f"     -> Ban va `{sname}.py` khong qua pytest ({pt_msg}).")
                            restore_skill(sname, bpath)
                        else:
                            healed_skills.append(sname)
                            meta = sync_registry_meta()
                            meta[sname]["health"] = "PASS" if pt_msg == "PASS" else "NO_TEST"
                            save_meta(meta)
                    except SyntaxError:
                        print(f"     -> Ban va `{sname}.py` loi cu phap.")
                        restore_skill(sname, bpath)
                    continue

            fix_run_p = f"""Sua `live_runner.py` de lay day du du lieu that va in ra stdout (KHONG dung example.com, KHONG nuot loi thanh 'N/A', cat gon moi tom tat toi da ~500 ky tu).
Van de: {diag.get('reason')}
Cac skill duoc phep import:
{reg_coder}
Code cu:
{FENCE}python
{runner_code}
{FENCE}
Output/Log:
{out[-1500:]}"""
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code.", fix_run_p, 0.2))

        update_skill_stats(meta, selected_for_coder, run_success)

        # BUOC 4: Analyst (qwq:32b) tong hop bao cao
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""THOI GIAN HIEN TAI CUA HE THONG: {vn_now} (nam 2026 la thoi diem hien tai that, KHONG phai tuong lai hay gia lap).
Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau BAT BUOC:
- Tra loi chinh xac, ro rang bang tieng Viet chuan (TUYET DOI khong chen chu Han/tieng Trung).
- Chi dua tren du lieu thu thap o tren; phan nao khong co du lieu thi noi ro la chua thu thap duoc.
- KHONG bao gio tu bia link URL; chi trich dan cac link co mat trong du lieu thu thap."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        elapsed = round(time.perf_counter() - t0, 2)
        append_run_log(
            {
                "timestamp": vn_now,
                "question": question,
                "use_skills": selected_for_coder,
                "forged": forged_ok,
                "forged_failed": failed_tools,
                "healed": healed_skills,
                "success": run_success,
                "elapsed_sec": elapsed,
            }
        )
        cleanup_sandbox()
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {elapsed}s\n")
        print("================ TRA LOI ================")
        print(final_ans)


def main():
    parser = argparse.ArgumentParser(description="AutoEvolve AI - Demand-Driven Self-Evolving Agent")
    parser.add_argument("question", nargs="*", help="Cau hoi can tra cuu")
    parser.add_argument("--health-check", action="store_true", help="Chay pytest kiem tra toan bo kho skills/ va don rac")
    parser.add_argument("--dry-run", action="store_true", help="Chi len ke hoach Architect, khong tao tool hay chay code")
    parser.add_argument("--no-forge", action="store_true", help="Chi dung skill san co, khong tao tool moi")
    args = parser.parse_args()

    if args.health_check:
        run_health_check()
        return

    q = " ".join(args.question).strip()
    if not q:
        parser.print_help()
        sys.exit(1)
    asyncio.run(solve_question(q, dry_run=args.dry_run, no_forge=args.no_forge))


if __name__ == "__main__":
    main()
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/skills/skill_web_search.py
"""Zero-key web and news search client using DDGS, DuckDuckGo HTML/Lite, and Google News RSS fallback."""
import re
import html
import urllib.parse
import xml.etree.ElementTree as ET
import httpx


class WebSearchClient:
    """Search the web and news without an API key."""

    def __init__(self, timeout: float = 15.0):
        self.timeout = timeout
        self.headers = {
            "User-Agent": (
                "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
                "(KHTML, like Gecko) Chrome/128.0.0.0 Safari/537.36"
            )
        }

    def _clean_html(self, raw: str) -> str:
        text = re.sub(r"<[^>]+>", " ", raw or "")
        return re.sub(r"\s+", " ", html.unescape(text)).strip()

    def _extract_ddg_url(self, href: str) -> str:
        href = html.unescape(href)
        if "uddg=" in href:
            parsed = urllib.parse.urlparse(href)
            qs = urllib.parse.parse_qs(parsed.query)
            if "uddg" in qs and qs["uddg"]:
                return qs["uddg"][0]
        if href.startswith("//"):
            return "https:" + href
        return href

    def _search_ddg_html(self, query: str, max_results: int) -> list[dict]:
        r = httpx.get(
            "https://html.duckduckgo.com/html/",
            params={"q": query},
            headers=self.headers,
            timeout=self.timeout,
            follow_redirects=True,
        )
        if r.status_code != 200:
            raise RuntimeError(f"DuckDuckGo HTML HTTP {r.status_code}")

        body = r.text
        links = re.findall(
            r'<a[^>]*class="[^"]*result__a[^"]*"[^>]*href="([^"]+)"[^>]*>(.*?)</a>',
            body,
            re.DOTALL | re.IGNORECASE,
        )
        snippets = re.findall(
            r'<(?:a|div)[^>]*class="[^"]*result__snippet[^"]*"[^>]*>(.*?)</(?:a|div)>',
            body,
            re.DOTALL | re.IGNORECASE,
        )

        results = []
        for idx, (href, raw_title) in enumerate(links[:max_results]):
            title = self._clean_html(raw_title)
            clean_url = self._extract_ddg_url(href)
            snippet = self._clean_html(snippets[idx]) if idx < len(snippets) else ""
            if title and clean_url.startswith("http"):
                results.append(
                    {
                        "title": title,
                        "url": clean_url,
                        "snippet": snippet,
                        "source": "duckduckgo",
                        "readable": True,
                    }
                )
        return results

    def _search_ddgs_pkg(self, query: str, max_results: int) -> list[dict]:
        try:
            from ddgs import DDGS
        except ImportError:
            try:
                from duckduckgo_search import DDGS
            except ImportError:
                return []

        results = []
        with DDGS() as ddgs:
            for item in ddgs.text(query, max_results=max_results) or []:
                url = item.get("href") or item.get("url") or ""
                title = (item.get("title") or "").strip()
                snippet = (item.get("body") or item.get("snippet") or "").strip()
                if title and url.startswith("http"):
                    results.append(
                        {
                            "title": title,
                            "url": url,
                            "snippet": snippet,
                            "source": "ddgs",
                            "readable": True,
                        }
                    )
        return results

    def _search_google_news_rss(self, query: str, max_results: int, lang: str = "vi") -> list[dict]:
        hl = "vi" if lang == "vi" else "en-US"
        gl = "VN" if lang == "vi" else "US"
        ceid = "VN:vi" if lang == "vi" else "US:en"
        r = httpx.get(
            "https://news.google.com/rss/search",
            params={"q": query, "hl": hl, "gl": gl, "ceid": ceid},
            headers=self.headers,
            timeout=self.timeout,
            follow_redirects=True,
        )
        if r.status_code != 200:
            raise RuntimeError(f"Google News RSS HTTP {r.status_code}")

        root = ET.fromstring(r.text)
        results = []
        for item in root.findall(".//item")[:max_results]:
            title = (item.findtext("title") or "").strip()
            link = (item.findtext("link") or "").strip()
            pub_date = (item.findtext("pubDate") or "").strip()
            desc = self._clean_html(item.findtext("description") or "")
            if title and link:
                results.append(
                    {
                        "title": title,
                        "url": link,
                        "snippet": f"{pub_date} - {desc}".strip(" -"),
                        "source": "google_news_rss",
                        "readable": False,
                    }
                )
        return results

    def search(self, query: str, max_results: int = 5, lang: str = "vi") -> list[dict]:
        """Return list of dicts: {title, url, snippet, source, readable}."""
        if not query or not query.strip():
            raise ValueError("Query khong duoc rong")
        errors = []

        # 1. Thu DDG HTML (tuong thich mock httpx.get trong pytest)
        try:
            res = self._search_ddg_html(query, max_results)
            if res:
                return res
        except Exception as e:
            errors.append(f"DDG_HTML: {e}")

        # 2. Thu goi ddgs (vuot chan IP datacenter tot hon)
        try:
            res = self._search_ddgs_pkg(query, max_results)
            if res:
                return res
        except Exception as e:
            errors.append(f"DDGS: {e}")

        # 3. Fallback cuoi cung: Google News RSS (chi dung title/snippet, khong nen scrape link RSS)
        try:
            res = self._search_google_news_rss(query, max_results, lang=lang)
            if res:
                return res
        except Exception as e:
            errors.append(f"RSS: {e}")

        if errors:
            raise RuntimeError(" | ".join(errors))
        raise RuntimeError(f"Khong tim thay ket qua nao cho: {query}")
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import shutil
import asyncio
import argparse
import subprocess
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"
BACKUP_DIR = f"{BASE_DIR}/skill_backups"
META_PATH = f"{BASE_DIR}/registry_meta.json"
RUNS_LOG_PATH = f"{BASE_DIR}/runs.jsonl"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3
MAX_BACKUPS_PER_SKILL = 5
CORE_WEB_SKILLS = {"skill_web_search", "skill_web_reader"}


def now_vn_str() -> str:
    return datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str, module_name: str | None = None, is_test: bool = False) -> str:
    blocks = re.findall(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    if not blocks and FENCE in text:
        part = text.split(FENCE, 1)[1]
        part = re.sub(r"^python\s*", "", part, flags=re.IGNORECASE)
        blocks = [part]
    if not blocks:
        blocks = [text]

    valid = []
    for b in blocks:
        code = b.strip()
        if not code:
            continue
        if module_name and not is_test:
            if re.search(rf"\b(from|import)\s+{re.escape(module_name)}\b", code):
                continue
            if "def test_" in code and "class " not in code:
                continue
        try:
            ast.parse(code)
            valid.append(code)
        except SyntaxError:
            continue

    if valid:
        return max(valid, key=len)
    return blocks[0].strip()


def extract_json(text: str) -> dict:
    candidates = re.findall(FENCE + r"(?:json)?\s*(\{.*?\})\s*" + FENCE, text, re.DOTALL)
    candidates.append(text)
    decoder = json.JSONDecoder(strict=False)
    for chunk in candidates:
        pos = 0
        while True:
            idx = chunk.find("{", pos)
            if idx == -1:
                break
            try:
                obj, _ = decoder.raw_decode(chunk, idx)
                if isinstance(obj, dict):
                    return obj
            except json.JSONDecodeError:
                pass
            pos = idx + 1
    raise ValueError(f"Khong tim thay JSON hop le trong: {text[:200]}")


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def inspect_skill_file(filepath: str) -> tuple[str, str]:
    """Return (short_description, signature_summary) for a skill file."""
    try:
        src = open(filepath, encoding="utf-8").read()
        tree = ast.parse(src)
        doc = (ast.get_docstring(tree) or "").splitlines()
        desc = doc[0].strip()[:120] if doc else ""
        sigs = []
        for n in tree.body:
            if isinstance(n, ast.ClassDef):
                if not desc:
                    cdoc = (ast.get_docstring(n) or "").splitlines()
                    if cdoc:
                        desc = cdoc[0].strip()[:120]
                ms = []
                for m in n.body:
                    if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef)) and (
                        not m.name.startswith("_") or m.name == "__init__"
                    ):
                        args_s = ", ".join(a.arg for a in m.args.args if a.arg != "self")
                        ret_s = f" -> {ast.unparse(m.returns)}" if m.returns else ""
                        ms.append(f"{m.name}({args_s}){ret_s}")
                sigs.append(f"class {n.name}: [{', '.join(ms)}]")
            elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                args_s = ", ".join(a.arg for a in n.args.args)
                ret_s = f" -> {ast.unparse(n.returns)}" if n.returns else ""
                sigs.append(f"def {n.name}({args_s}){ret_s}")
        return desc or "Python skill module", " | ".join(sigs)
    except Exception as e:
        return f"Parse error: {e}", ""


def load_meta() -> dict:
    if os.path.exists(META_PATH):
        try:
            with open(META_PATH, "r", encoding="utf-8") as f:
                return json.load(f)
        except Exception:
            pass
    return {}


def save_meta(meta: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(META_PATH, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)


def sync_registry_meta() -> dict:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    meta = load_meta()
    existing = set()
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        existing.add(mod)
        desc, sigs = inspect_skill_file(f"{SKILLS_DIR}/{fn}")
        entry = meta.get(mod, {})
        entry["signatures"] = sigs
        if not entry.get("description") or entry["description"] == "Python skill module":
            entry["description"] = desc
        entry.setdefault("success_count", 0)
        entry.setdefault("fail_count", 0)
        entry.setdefault("health", "UNKNOWN")
        entry.setdefault("updated_at", now_vn_str())
        meta[mod] = entry

    for dead in list(meta.keys()):
        if dead not in existing:
            del meta[dead]

    save_meta(meta)
    return meta


def format_registry_for_architect(meta: dict) -> str:
    if not meta:
        return "(Kho dang trong)"
    lines = []
    for mod in sorted(meta.keys()):
        info = meta[mod]
        stats = f"ok={info.get('success_count', 0)}, fail={info.get('fail_count', 0)}, health={info.get('health', 'UNKNOWN')}"
        lines.append(f"- `{mod}` ({stats}): {info.get('description', '')} | {info.get('signatures', '')}")
    return "\n".join(lines)


def format_registry_for_coder(meta: dict, selected: list[str] | None = None) -> str:
    if not meta:
        return "(Kho dang trong)"
    wanted = set()
    if selected:
        for s in selected:
            name = s.replace(".py", "").strip()
            if name in meta:
                wanted.add(name)
    for core in CORE_WEB_SKILLS:
        if core in meta:
            wanted.add(core)
    if not wanted:
        wanted = set(meta.keys())
    lines = [f"- `{m}`: {meta[m].get('signatures', '')}" for m in sorted(wanted)]
    return "\n".join(lines)


def update_skill_stats(meta: dict, skill_names: list[str], success: bool):
    ts = now_vn_str()
    for s in skill_names:
        name = s.replace(".py", "").strip()
        if name in meta:
            key = "success_count" if success else "fail_count"
            meta[name][key] = meta[name].get(key, 0) + 1
            meta[name]["last_used"] = ts
    save_meta(meta)


def append_run_log(record: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(RUNS_LOG_PATH, "a", encoding="utf-8") as f:
        f.write(json.dumps(record, ensure_ascii=False) + "\n")


def cleanup_backups(max_per_skill: int = MAX_BACKUPS_PER_SKILL):
    if not os.path.exists(BACKUP_DIR):
        return
    groups: dict[str, list[str]] = {}
    for fn in os.listdir(BACKUP_DIR):
        if not fn.endswith(".py.bak"):
            continue
        m = re.match(r"^(skill_.+)_\d{8}_\d{6}\.py\.bak$", fn)
        key = m.group(1) if m else fn
        groups.setdefault(key, []).append(os.path.join(BACKUP_DIR, fn))
    for files in groups.values():
        files.sort(key=lambda p: os.path.getmtime(p), reverse=True)
        for old_path in files[max_per_skill:]:
            try:
                os.remove(old_path)
            except OSError:
                pass


def cleanup_sandbox():
    if not os.path.exists(SANDBOX_DIR):
        return
    for name in os.listdir(SANDBOX_DIR):
        if name.startswith(("forge_", "heal_verify", "health_")):
            full = os.path.join(SANDBOX_DIR, name)
            try:
                if os.path.isdir(full):
                    shutil.rmtree(full, ignore_errors=True)
                else:
                    os.remove(full)
            except OSError:
                pass


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:{BASE_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:160] if errs else (out.replace("\n", " | ")[:160] if out else "Output rong")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


def backup_skill(sname: str) -> str | None:
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return None
    os.makedirs(BACKUP_DIR, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    bpath = f"{BACKUP_DIR}/{sname}_{ts}.py.bak"
    shutil.copy2(spath, bpath)
    cleanup_backups()
    return bpath


def restore_skill(sname: str, bpath: str | None):
    spath = f"{SKILLS_DIR}/{sname}.py"
    if bpath and os.path.exists(bpath):
        shutil.copy2(bpath, spath)
        print(f"     -> [ROLLBACK] Da khoi phuc ban goc cua `{sname}.py` do ban va chua dat.")


def verify_skill_pytest(sname: str) -> tuple[bool, str]:
    tpath = f"{SKILLS_DIR}/test_{sname}.py"
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return False, "Khong tim thay file skill"
    if not os.path.exists(tpath):
        return True, "NO_TEST"
    vdir = f"{SANDBOX_DIR}/heal_verify_{sname}"
    os.makedirs(vdir, exist_ok=True)
    try:
        shutil.copy2(tpath, f"{vdir}/test_{sname}.py")
        shutil.copy2(spath, f"{vdir}/{sname}.py")
        ok, _, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], vdir, 20)
        return ok, ("PASS" if ok else summary)
    finally:
        shutil.rmtree(vdir, ignore_errors=True)


def run_health_check():
    print("=== HEALTH CHECK KHO SKILLS ===")
    meta = sync_registry_meta()
    if not meta:
        print("Kho skills/ dang trong.")
        return
    passed = failed = no_test = 0
    for mod in sorted(meta.keys()):
        spath = f"{SKILLS_DIR}/{mod}.py"
        try:
            ast.parse(open(spath, encoding="utf-8").read())
        except SyntaxError as e:
            meta[mod]["health"] = "SYNTAX_ERROR"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> SyntaxError: {e}")
            continue

        ok, msg = verify_skill_pytest(mod)
        if msg == "NO_TEST":
            meta[mod]["health"] = "NO_TEST"
            no_test += 1
            print(f"  [WARN] {mod:<35} -> Khong co file test")
        elif ok:
            meta[mod]["health"] = "PASS"
            passed += 1
            print(f"  [PASS] {mod:<35} -> pytest OK")
        else:
            meta[mod]["health"] = "FAIL"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> {msg}")
        meta[mod]["last_health_check"] = now_vn_str()

    save_meta(meta)
    cleanup_backups()
    cleanup_sandbox()
    print(f"\nTong ket: {passed} PASS | {failed} FAIL | {no_test} NO_TEST (da cap nhat registry_meta.json & don rac).")


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong (cho phep them tham so tuy chon co gia tri mac dinh)?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def verify_collected_data(client: httpx.AsyncClient, question: str, data: str) -> tuple[bool, str]:
    ver_p = f"""Cau hoi nguoi dung: {question}

Du lieu script vua thu thap duoc:
{data[:8000]}

Kiem tra: Du lieu tren da cung cap du thong tin thuc te de tra loi TAT CA cac ve cua cau hoi chua?
(Neu thieu bat ky ve nao, hoac bi loi/rong/N/A/example.com/404 -> FAIL).
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
MISSING: <neu FAIL, ghi ro phan nao bi thieu hoac rong; neu PASS ghi None>"""
    res = await ask_llm(client, BRAIN_THINK, "Ban la QA Data Verifier nghiem khac.", ver_p, 0.1)
    ok = "STATUS: PASS" in res.upper()
    m = re.search(r"MISSING:\s*(.+)", res)
    return ok, (m.group(1).strip() if m else res.splitlines()[-1][:120])


async def forge_tool(client: httpx.AsyncClient, spec: dict, meta: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3 test case ro rang (1 happy path + 1 loi HTTP status != 200 + 1 loi du lieu rong/mang).
- Code thuc thi dung `httpx.get(...)`, nen mock bang `unittest.mock.patch("httpx.get")` (KHONG mock `requests`).
- Khi test truong hop loi, LUON dung `with pytest.raises(Exception):`.
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve duy nhat 1 block code Python cho `test_{sname}.py`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- LUON dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})` (KHONG dung `requests`).
- KHONG nuot loi mang im lang (neu HTTP status != 200 hoac du lieu loi, raise `RuntimeError` kem status_code va noi dung loi).
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY (Open-Meteo, MediaWiki/Wikipedia, ArXiv, OpenAlex, Google News RSS, Frankfurter, Binance public).
- Tuyet doi khong hardcode du lieu gia, khong viet code test trong file nay.
Chi tra ve duy nhat 1 block code Python cho `{sname}.py`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve 1 block code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0], module_name=sname, is_test=True)
    candidates = [extract_code(c, module_name=sname, is_test=False) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    def save_forged(code_str: str, test_str: str):
        with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(code_str)
        with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_str)
        _, sigs = inspect_skill_file(f"{SKILLS_DIR}/{sname}.py")
        meta[sname] = {
            "description": goal[:120],
            "signatures": sigs,
            "success_count": 0,
            "fail_count": 0,
            "health": "PASS",
            "updated_at": now_vn_str(),
        }
        save_meta(meta)

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    save_forged(curr, test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"""Sua file `{sname}.py` de pass pytest (van dung `httpx.get(..., follow_redirects=True)` va API that).
CHI tra ve code cua `{sname}.py`, KHONG tra ve file test.
Code:
{FENCE}python
{curr}
{FENCE}
Test:
{FENCE}python
{test_code}
{FENCE}
Log:
{log[-1500:]}"""
                    curr = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code cho file skill.", fix_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )

    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, ep sai loai exception, sai cau truc response):
Tra ve `VERDICT: TEST_BUG` kem toan bo file `test_{sname}.py` da sua trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj, module_name=sname, is_test=True)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                save_forged(last_code0, fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str, dry_run: bool = False, no_forge: bool = False):
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)
    cleanup_sandbox()
    cleanup_backups()
    vn_now = now_vn_str()
    meta = sync_registry_meta()

    forged_ok: list[str] = []
    failed_tools: list[str] = []
    healed_skills: list[str] = []
    run_success = False

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) phan tich nhu cau & doc registry_meta.json
        reg_arch = format_registry_for_architect(meta)
        plan_p = f"""CAU HOI NGUOI DUNG: {question}
THOI GIAN HIEN TAI: {vn_now}

KHO SKILL HIEN CO (tu registry_meta.json):
{reg_arch}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
   (Luu y: da co `skill_web_search` va `skill_web_reader` de tim kiem & doc bat ky trang web nao; chi tao tool moi neu can goi 1 API chuyen biet chua co).
2. Neu THIEU tool chuyen biet, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...).
   - Thoi tiet -> Open-Meteo; Kien thuc -> MediaWiki/Wikipedia; Khoa hoc -> ArXiv/OpenAlex; Tai chinh -> Frankfurter/Binance public; Web/Tin tuc -> dung san `skill_web_search` + `skill_web_reader`.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT 1 JSON hop le, khong kem giai thich ben ngoai:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra registry_meta.json...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve 1 JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        use_skills = plan.get("use_skills", [])
        print(f"   -> Dung skills: {use_skills} | Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        if dry_run:
            print("\n[DRY-RUN] Ke hoach Architect:")
            print(json.dumps(plan, ensure_ascii=False, indent=2))
            return

        # BUOC 2: Forge tu che tao tool con thieu
        if not no_forge:
            for spec in missing:
                ok_forge = await forge_tool(client, spec, meta)
                if ok_forge:
                    sname = spec["skill_name"]
                    if not sname.startswith("skill_"):
                        sname = "skill_" + sname.lstrip("_")
                    forged_ok.append(sname)
                else:
                    failed_tools.append(spec["skill_name"])
        elif missing:
            print(f"   -> [--no-forge] Bo qua tao moi: {[m['skill_name'] for m in missing]}")
            failed_tools = [m["skill_name"] for m in missing]

        # BUOC 3: Coder chay thuc chien + Data Quality Gate + Self-Healing + Rollback
        meta = sync_registry_meta()
        selected_for_coder = list(dict.fromkeys(use_skills + forged_ok))
        reg_coder = format_registry_for_coder(meta, selected_for_coder)
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        failed_note = ""
        if failed_tools:
            failed_note = f"\nLUU Y QUAN TRONG: Cac tool sau CHUA DUOC TAO nen KHONG import: {failed_tools}. Hay dung `skill_web_search` / `skill_web_reader` hoac goi truc tiep API mo bang `httpx.get(..., follow_redirects=True)`."

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
Cac skill duoc phep import (chu y ky kieu tra ve `-> dict`, `-> list[dict]`):
{reg_coder}
{failed_note}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau BAT BUOC:
- Lay day du du lieu cho TAT CA cac ve cua cau hoi.
- TUYET DOI KHONG dung URL gia nhu `example.com`; neu `skill_web_search` tra ve link `news.google.com/rss/articles/` (readable=False) thi CHI dung title/snippet, KHONG truyen link do vao `WebReaderClient.read_url()`. Neu biet san trang chinh thuc (vd: `https://www.python.org/downloads/`), co the doc truc tiep bang `WebReaderClient.read_url()`.
- KHONG nuot loi im lang thanh 'N/A': in thang ket qua tho hoac traceback loi ra stdout/stderr.
- Cat gon moi doan tom tat toi da ~500 ky tu de in day du TAT CA cac muc va link nguon trong ~6000 ky tu.
- Chi tra ve 1 block code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve 1 block code Python.", run_p, 0.2))

        collected_data = ""
        heal_counts: dict[str, int] = {}
        max_attempts = 5
        for attempt in range(1, max_attempts + 1):
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)

            fail_reason = summary
            if ok and len(out.strip()) > 80:
                gate_ok, missing_info = await verify_collected_data(client, question, out)
                if gate_ok:
                    collected_data = out[:8000]
                    run_success = True
                    print(f"   -> [OK lan {attempt}] Thu thap DU du lieu ({len(collected_data)} ky tu)!")
                    print(f"   -> [PREVIEW] {collected_data[:220].replace(chr(10), ' | ')}...")
                    break
                else:
                    fail_reason = f"DATA_INCOMPLETE: {missing_info}"
                    print(f"   -> [THIEU DU LIEU lan {attempt}] {missing_info}")
                    print(f"      [RAW OUT] {out[:200].replace(chr(10), ' | ')}")
            else:
                print(f"   -> [FAIL lan {attempt}] {summary}")

            if attempt == max_attempts:
                collected_data = out[-4000:]
                break

            diag_p = f"""Script thu thap du lieu bi loi hoac thieu du lieu khi chay thuc te.
Van de: {fail_reason}
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Output / Log:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi do `live_runner.py` (goi sai ham, doc sai kieu dict/str, truyen link RSS vao WebReaderClient, dung URL gia): tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": fail_reason}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath) and heal_counts.get(sname, 0) < 2:
                    heal_counts[sname] = heal_counts.get(sname, 0) + 1
                    print(f"   -> [SELF-HEAL #{heal_counts[sname]}] Va `{sname}.py` ({diag.get('reason')})...")
                    bpath = backup_skill(sname)
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc van de chay thuc te: {diag.get('reason')}

Code hien tai cua `{sname}.py`:
{FENCE}python
{old_src}
{FENCE}

Cach `live_runner.py` dang goi va doc ket qua tu tool nay:
{FENCE}python
{runner_code}
{FENCE}

Output / Log thuc te:
{out[-1500:]}

QUY TAC SUA BAT BUOC:
1. Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})`.
2. TUYET DOI KHONG nuot loi im lang tra ve rong/None; neu API loi phai raise RuntimeError kem status_code va response text.
3. Giu nguyen ten class/ham public va kieu tra ve de pass pytest.
Chi tra ve 1 block code Python cho `{sname}.py`."""
                    patched = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi tool Python. Chi tra ve 1 block code.", patch_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )
                    try:
                        ast.parse(patched)
                        with open(spath, "w", encoding="utf-8") as f:
                            f.write(patched)
                        pt_ok, pt_msg = verify_skill_pytest(sname)
                        if not pt_ok:
                            print(f"     -> Ban va `{sname}.py` khong qua pytest ({pt_msg}).")
                            restore_skill(sname, bpath)
                            diag["reason"] = f"Skill `{sname}` giu nguyen interface (ban va bi rollback). Hay sua `live_runner.py` de dung dung kieu tra ve cua skill."
                        else:
                            healed_skills.append(sname)
                            meta = sync_registry_meta()
                            meta[sname]["health"] = "PASS" if pt_msg == "PASS" else "NO_TEST"
                            save_meta(meta)
                            continue
                    except SyntaxError:
                        print(f"     -> Ban va `{sname}.py` loi cu phap.")
                        restore_skill(sname, bpath)
                        diag["reason"] = f"Ban va `{sname}` loi cu phap va da rollback. Hay sua `live_runner.py`."

            fix_run_p = f"""Sua `live_runner.py` de lay day du du lieu that va in ra stdout (KHONG dung example.com, KHONG truyen link `news.google.com/rss/articles/` vao WebReaderClient, doc dung kieu dict/list tra ve cua skill).
Van de: {diag.get('reason')}
Cac skill duoc phep import:
{reg_coder}
Code cu:
{FENCE}python
{runner_code}
{FENCE}
Output/Log:
{out[-1500:]}"""
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code.", fix_run_p, 0.2))

        update_skill_stats(meta, selected_for_coder, run_success)

        # BUOC 4: Analyst (qwq:32b) tong hop bao cao
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""THOI GIAN HIEN TAI CUA HE THONG: {vn_now} (nam 2026 la thoi diem hien tai that, KHONG phai tuong lai hay gia lap).
Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau BAT BUOC:
- Tra loi chinh xac, ro rang bang tieng Viet chuan (TUYET DOI khong chen chu Han/tieng Trung).
- Chi dua tren du lieu thu thap o tren; phan nao khong co du lieu thi noi ro la chua thu thap duoc.
- KHONG bao gio tu bia link URL; chi trich dan cac link co mat trong du lieu thu thap."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        elapsed = round(time.perf_counter() - t0, 2)
        append_run_log(
            {
                "timestamp": vn_now,
                "question": question,
                "use_skills": selected_for_coder,
                "forged": forged_ok,
                "forged_failed": failed_tools,
                "healed": healed_skills,
                "success": run_success,
                "elapsed_sec": elapsed,
            }
        )
        cleanup_sandbox()
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {elapsed}s\n")
        print("================ TRA LOI ================")
        print(final_ans)


def main():
    parser = argparse.ArgumentParser(description="AutoEvolve AI - Demand-Driven Self-Evolving Agent")
    parser.add_argument("question", nargs="*", help="Cau hoi can tra cuu")
    parser.add_argument("--health-check", action="store_true", help="Chay pytest kiem tra toan bo kho skills/ va don rac")
    parser.add_argument("--dry-run", action="store_true", help="Chi len ke hoach Architect, khong tao tool hay chay code")
    parser.add_argument("--no-forge", action="store_true", help="Chi dung skill san co, khong tao tool moi")
    args = parser.parse_args()

    if args.health_check:
        run_health_check()
        return

    q = " ".join(args.question).strip()
    if not q:
        parser.print_help()
        sys.exit(1)
    asyncio.run(solve_question(q, dry_run=args.dry_run, no_forge=args.no_forge))


if __name__ == "__main__":
    main()
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/skills/test_skill_web_search.py
from unittest.mock import MagicMock, patch
import pytest
from skill_web_search import WebSearchClient


def test_search_ddg_happy_path():
    fake_html = '''
    <div class="result">
      <a class="result__a" href="https://duckduckgo.com/l/?uddg=https%3A%2F%2Fexample.org%2Fai">AI News</a>
      <a class="result__snippet">Latest breakthrough in AI agents.</a>
    </div>
    '''
    mock_resp = MagicMock(status_code=200, text=fake_html)
    with patch("httpx.get", return_value=mock_resp):
        client = WebSearchClient()
        res = client.search("AI agents", max_results=3)
        assert len(res) == 1
        assert res[0]["title"] == "AI News"
        assert res[0]["url"] == "https://example.org/ai"
        assert "breakthrough" in res[0]["snippet"]


def test_search_fallback_to_rss():
    ddg_fail = MagicMock(status_code=403, text="Forbidden")
    rss_xml = '''<?xml version="1.0" encoding="UTF-8"?>
    <rss><channel><item>
      <title>RSS Headline</title>
      <link>https://news.example.org/1</link>
      <pubDate>Wed, 25 Nov 2026 03:00:00 GMT</pubDate>
      <description>Short summary</description>
    </item></channel></rss>'''
    rss_ok = MagicMock(status_code=200, text=rss_xml)
    with patch("httpx.get", side_effect=[ddg_fail, rss_ok]), patch.object(
        WebSearchClient, "_search_ddgs_pkg", return_value=[]
    ):
        client = WebSearchClient()
        res = client.search("AI", max_results=2)
        assert len(res) == 1
        assert res[0]["source"] == "google_news_rss"
        assert res[0]["title"] == "RSS Headline"


def test_search_all_backends_error():
    fail_resp = MagicMock(status_code=500, text="Server Error")
    with patch("httpx.get", return_value=fail_resp), patch.object(
        WebSearchClient, "_search_ddgs_pkg", side_effect=RuntimeError("DDGS blocked")
    ):
        client = WebSearchClient()
        with pytest.raises(Exception):
            client.search("AI")
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import shutil
import asyncio
import argparse
import subprocess
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"
BACKUP_DIR = f"{BASE_DIR}/skill_backups"
META_PATH = f"{BASE_DIR}/registry_meta.json"
RUNS_LOG_PATH = f"{BASE_DIR}/runs.jsonl"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3
MAX_BACKUPS_PER_SKILL = 5
CORE_WEB_SKILLS = {"skill_web_search", "skill_web_reader"}


def now_vn_str() -> str:
    return datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str, module_name: str | None = None, is_test: bool = False) -> str:
    blocks = re.findall(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    if not blocks and FENCE in text:
        part = text.split(FENCE, 1)[1]
        part = re.sub(r"^python\s*", "", part, flags=re.IGNORECASE)
        blocks = [part]
    if not blocks:
        blocks = [text]

    valid = []
    for b in blocks:
        code = b.strip()
        if not code:
            continue
        if module_name and not is_test:
            if re.search(rf"\b(from|import)\s+{re.escape(module_name)}\b", code):
                continue
            if "def test_" in code and "class " not in code:
                continue
        try:
            ast.parse(code)
            valid.append(code)
        except SyntaxError:
            continue

    if valid:
        return max(valid, key=len)
    return blocks[0].strip()


def extract_json(text: str) -> dict:
    candidates = re.findall(FENCE + r"(?:json)?\s*(\{.*?\})\s*" + FENCE, text, re.DOTALL)
    candidates.append(text)
    decoder = json.JSONDecoder(strict=False)
    for chunk in candidates:
        pos = 0
        while True:
            idx = chunk.find("{", pos)
            if idx == -1:
                break
            try:
                obj, _ = decoder.raw_decode(chunk, idx)
                if isinstance(obj, dict):
                    return obj
            except json.JSONDecodeError:
                pass
            pos = idx + 1
    raise ValueError(f"Khong tim thay JSON hop le trong: {text[:200]}")


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def inspect_skill_file(filepath: str) -> tuple[str, str]:
    """Return (short_description, signature_summary) for a skill file."""
    try:
        src = open(filepath, encoding="utf-8").read()
        tree = ast.parse(src)
        doc = (ast.get_docstring(tree) or "").splitlines()
        desc = doc[0].strip()[:120] if doc else ""
        sigs = []
        for n in tree.body:
            if isinstance(n, ast.ClassDef):
                if not desc:
                    cdoc = (ast.get_docstring(n) or "").splitlines()
                    if cdoc:
                        desc = cdoc[0].strip()[:120]
                ms = []
                for m in n.body:
                    if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef)) and (
                        not m.name.startswith("_") or m.name == "__init__"
                    ):
                        args_s = ", ".join(a.arg for a in m.args.args if a.arg != "self")
                        ret_s = f" -> {ast.unparse(m.returns)}" if m.returns else ""
                        ms.append(f"{m.name}({args_s}){ret_s}")
                sigs.append(f"class {n.name}: [{', '.join(ms)}]")
            elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                args_s = ", ".join(a.arg for a in n.args.args)
                ret_s = f" -> {ast.unparse(n.returns)}" if n.returns else ""
                sigs.append(f"def {n.name}({args_s}){ret_s}")
        return desc or "Python skill module", " | ".join(sigs)
    except Exception as e:
        return f"Parse error: {e}", ""


def load_meta() -> dict:
    if os.path.exists(META_PATH):
        try:
            with open(META_PATH, "r", encoding="utf-8") as f:
                return json.load(f)
        except Exception:
            pass
    return {}


def save_meta(meta: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(META_PATH, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)


def sync_registry_meta() -> dict:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    meta = load_meta()
    existing = set()
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        existing.add(mod)
        desc, sigs = inspect_skill_file(f"{SKILLS_DIR}/{fn}")
        entry = meta.get(mod, {})
        entry["signatures"] = sigs
        if not entry.get("description") or entry["description"] == "Python skill module":
            entry["description"] = desc
        entry.setdefault("success_count", 0)
        entry.setdefault("fail_count", 0)
        entry.setdefault("health", "UNKNOWN")
        entry.setdefault("updated_at", now_vn_str())
        meta[mod] = entry

    for dead in list(meta.keys()):
        if dead not in existing:
            del meta[dead]

    save_meta(meta)
    return meta


def format_registry_for_architect(meta: dict) -> str:
    if not meta:
        return "(Kho dang trong)"
    lines = []
    for mod in sorted(meta.keys()):
        info = meta[mod]
        stats = f"ok={info.get('success_count', 0)}, fail={info.get('fail_count', 0)}, health={info.get('health', 'UNKNOWN')}"
        lines.append(f"- `{mod}` ({stats}): {info.get('description', '')} | {info.get('signatures', '')}")
    return "\n".join(lines)


def format_registry_for_coder(meta: dict, selected: list[str] | None = None) -> str:
    if not meta:
        return "(Kho dang trong)"
    wanted = set()
    if selected:
        for s in selected:
            name = s.replace(".py", "").strip()
            if name in meta:
                wanted.add(name)
    for core in CORE_WEB_SKILLS:
        if core in meta:
            wanted.add(core)
    if not wanted:
        wanted = set(meta.keys())
    lines = [f"- `{m}`: {meta[m].get('signatures', '')}" for m in sorted(wanted)]
    return "\n".join(lines)


def update_skill_stats(meta: dict, skill_names: list[str], success: bool):
    ts = now_vn_str()
    for s in skill_names:
        name = s.replace(".py", "").strip()
        if name in meta:
            key = "success_count" if success else "fail_count"
            meta[name][key] = meta[name].get(key, 0) + 1
            meta[name]["last_used"] = ts
    save_meta(meta)


def append_run_log(record: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(RUNS_LOG_PATH, "a", encoding="utf-8") as f:
        f.write(json.dumps(record, ensure_ascii=False) + "\n")


def cleanup_backups(max_per_skill: int = MAX_BACKUPS_PER_SKILL):
    if not os.path.exists(BACKUP_DIR):
        return
    groups: dict[str, list[str]] = {}
    for fn in os.listdir(BACKUP_DIR):
        if not fn.endswith(".py.bak"):
            continue
        m = re.match(r"^(skill_.+)_\d{8}_\d{6}\.py\.bak$", fn)
        key = m.group(1) if m else fn
        groups.setdefault(key, []).append(os.path.join(BACKUP_DIR, fn))
    for files in groups.values():
        files.sort(key=lambda p: os.path.getmtime(p), reverse=True)
        for old_path in files[max_per_skill:]:
            try:
                os.remove(old_path)
            except OSError:
                pass


def cleanup_sandbox():
    if not os.path.exists(SANDBOX_DIR):
        return
    for name in os.listdir(SANDBOX_DIR):
        if name.startswith(("forge_", "heal_verify", "health_")):
            full = os.path.join(SANDBOX_DIR, name)
            try:
                if os.path.isdir(full):
                    shutil.rmtree(full, ignore_errors=True)
                else:
                    os.remove(full)
            except OSError:
                pass


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:{BASE_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:160] if errs else (out.replace("\n", " | ")[:160] if out else "Output rong")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


def backup_skill(sname: str) -> str | None:
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return None
    os.makedirs(BACKUP_DIR, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    bpath = f"{BACKUP_DIR}/{sname}_{ts}.py.bak"
    shutil.copy2(spath, bpath)
    cleanup_backups()
    return bpath


def restore_skill(sname: str, bpath: str | None):
    spath = f"{SKILLS_DIR}/{sname}.py"
    if bpath and os.path.exists(bpath):
        shutil.copy2(bpath, spath)
        print(f"     -> [ROLLBACK] Da khoi phuc ban goc cua `{sname}.py` do ban va chua dat.")


def verify_skill_pytest(sname: str) -> tuple[bool, str]:
    tpath = f"{SKILLS_DIR}/test_{sname}.py"
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return False, "Khong tim thay file skill"
    if not os.path.exists(tpath):
        return True, "NO_TEST"
    vdir = f"{SANDBOX_DIR}/heal_verify_{sname}"
    os.makedirs(vdir, exist_ok=True)
    try:
        shutil.copy2(tpath, f"{vdir}/test_{sname}.py")
        shutil.copy2(spath, f"{vdir}/{sname}.py")
        ok, _, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], vdir, 20)
        return ok, ("PASS" if ok else summary)
    finally:
        shutil.rmtree(vdir, ignore_errors=True)


def run_health_check():
    print("=== HEALTH CHECK KHO SKILLS ===")
    meta = sync_registry_meta()
    if not meta:
        print("Kho skills/ dang trong.")
        return
    passed = failed = no_test = 0
    for mod in sorted(meta.keys()):
        spath = f"{SKILLS_DIR}/{mod}.py"
        try:
            ast.parse(open(spath, encoding="utf-8").read())
        except SyntaxError as e:
            meta[mod]["health"] = "SYNTAX_ERROR"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> SyntaxError: {e}")
            continue

        ok, msg = verify_skill_pytest(mod)
        if msg == "NO_TEST":
            meta[mod]["health"] = "NO_TEST"
            no_test += 1
            print(f"  [WARN] {mod:<35} -> Khong co file test")
        elif ok:
            meta[mod]["health"] = "PASS"
            passed += 1
            print(f"  [PASS] {mod:<35} -> pytest OK")
        else:
            meta[mod]["health"] = "FAIL"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> {msg}")
        meta[mod]["last_health_check"] = now_vn_str()

    save_meta(meta)
    cleanup_backups()
    cleanup_sandbox()
    print(f"\nTong ket: {passed} PASS | {failed} FAIL | {no_test} NO_TEST (da cap nhat registry_meta.json & don rac).")


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong (cho phep them tham so tuy chon co gia tri mac dinh)?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def verify_collected_data(client: httpx.AsyncClient, question: str, data: str) -> tuple[bool, str]:
    ver_p = f"""Cau hoi nguoi dung: {question}

Du lieu script vua thu thap duoc:
{data[:8000]}

Kiem tra:
- Du lieu tren da cung cap du thong tin thuc te de tra loi cau hoi chua?
- LUU Y: Neu cau hoi dung tu "hoac" / "or" (vi du "A hoac B"), chi can co du lieu tot cho A HOAC B la DAT (PASS), KHONG bat buoc phai co ca hai.
- Chi danh FAIL neu du lieu bi loi/Traceback/rong/N/A/example.com/404, hoac thieu mot ve bat buoc ("va").
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
MISSING: <neu FAIL, ghi ro phan nao bi thieu hoac rong; neu PASS ghi None>"""
    res = await ask_llm(client, BRAIN_THINK, "Ban la QA Data Verifier chinh xac.", ver_p, 0.1)
    ok = "STATUS: PASS" in res.upper()
    m = re.search(r"MISSING:\s*(.+)", res)
    return ok, (m.group(1).strip() if m else res.splitlines()[-1][:120])


async def forge_tool(client: httpx.AsyncClient, spec: dict, meta: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3 test case ro rang (1 happy path + 1 loi HTTP status != 200 + 1 loi du lieu rong/mang).
- Code thuc thi dung `httpx.get(...)`, nen mock bang `unittest.mock.patch("httpx.get")` (KHONG mock `requests`).
- Khi test truong hop loi, LUON dung `with pytest.raises(Exception):`.
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve duy nhat 1 block code Python cho `test_{sname}.py`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- LUON dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})` (KHONG dung `requests`).
- KHONG nuot loi mang im lang (neu HTTP status != 200 hoac du lieu loi, raise `RuntimeError` kem status_code va noi dung loi).
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY (Open-Meteo, MediaWiki/Wikipedia, ArXiv, OpenAlex, Google News RSS, Frankfurter, Binance public).
- Tuyet doi khong hardcode du lieu gia, khong viet code test trong file nay.
Chi tra ve duy nhat 1 block code Python cho `{sname}.py`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve 1 block code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0], module_name=sname, is_test=True)
    candidates = [extract_code(c, module_name=sname, is_test=False) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    def save_forged(code_str: str, test_str: str):
        with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(code_str)
        with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_str)
        _, sigs = inspect_skill_file(f"{SKILLS_DIR}/{sname}.py")
        meta[sname] = {
            "description": goal[:120],
            "signatures": sigs,
            "success_count": 0,
            "fail_count": 0,
            "health": "PASS",
            "updated_at": now_vn_str(),
        }
        save_meta(meta)

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    save_forged(curr, test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"""Sua file `{sname}.py` de pass pytest (van dung `httpx.get(..., follow_redirects=True)` va API that).
CHI tra ve code cua `{sname}.py`, KHONG tra ve file test.
Code:
{FENCE}python
{curr}
{FENCE}
Test:
{FENCE}python
{test_code}
{FENCE}
Log:
{log[-1500:]}"""
                    curr = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code cho file skill.", fix_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )

    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, ep sai loai exception, sai cau truc response):
Tra ve `VERDICT: TEST_BUG` kem toan bo file `test_{sname}.py` da sua trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj, module_name=sname, is_test=True)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                save_forged(last_code0, fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str, dry_run: bool = False, no_forge: bool = False):
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)
    cleanup_sandbox()
    cleanup_backups()
    vn_now = now_vn_str()
    meta = sync_registry_meta()

    forged_ok: list[str] = []
    failed_tools: list[str] = []
    healed_skills: list[str] = []
    run_success = False

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) phan tich nhu cau & doc registry_meta.json
        reg_arch = format_registry_for_architect(meta)
        plan_p = f"""CAU HOI NGUOI DUNG: {question}
THOI GIAN HIEN TAI: {vn_now}

KHO SKILL HIEN CO (tu registry_meta.json):
{reg_arch}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
   (Luu y: da co `skill_web_search` va `skill_web_reader` de tim kiem & doc bat ky trang web nao; chi tao tool moi neu can goi 1 API chuyen biet chua co).
2. Neu THIEU tool chuyen biet, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...).
   - Thoi tiet -> Open-Meteo; Kien thuc -> MediaWiki/Wikipedia; Khoa hoc -> ArXiv/OpenAlex; Tai chinh -> Frankfurter/Binance public; Web/Tin tuc -> dung san `skill_web_search` + `skill_web_reader`.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT 1 JSON hop le, khong kem giai thich ben ngoai:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra registry_meta.json...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve 1 JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        use_skills = plan.get("use_skills", [])
        print(f"   -> Dung skills: {use_skills} | Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        if dry_run:
            print("\n[DRY-RUN] Ke hoach Architect:")
            print(json.dumps(plan, ensure_ascii=False, indent=2))
            return

        # BUOC 2: Forge tu che tao tool con thieu
        if not no_forge:
            for spec in missing:
                ok_forge = await forge_tool(client, spec, meta)
                if ok_forge:
                    sname = spec["skill_name"]
                    if not sname.startswith("skill_"):
                        sname = "skill_" + sname.lstrip("_")
                    forged_ok.append(sname)
                else:
                    failed_tools.append(spec["skill_name"])
        elif missing:
            print(f"   -> [--no-forge] Bo qua tao moi: {[m['skill_name'] for m in missing]}")
            failed_tools = [m["skill_name"] for m in missing]

        # BUOC 3: Coder chay thuc chien + Data Quality Gate + Self-Healing + Rollback
        meta = sync_registry_meta()
        selected_for_coder = list(dict.fromkeys(use_skills + forged_ok))
        reg_coder = format_registry_for_coder(meta, selected_for_coder)
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        failed_note = ""
        if failed_tools:
            failed_note = f"\nLUU Y QUAN TRONG: Cac tool sau CHUA DUOC TAO nen KHONG import: {failed_tools}. Hay dung `skill_web_search` / `skill_web_reader` hoac goi truc tiep API mo bang `httpx.get(..., follow_redirects=True)`."

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
Cac skill duoc phep import (chu y ky kieu tra ve `-> dict`, `-> list[dict]`):
{reg_coder}
{failed_note}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau BAT BUOC:
- Lay day du du lieu cho TAT CA cac ve cua cau hoi.
- TUYET DOI KHONG dung URL gia nhu `example.com`; neu `skill_web_search` tra ve link `news.google.com/rss/articles/` (readable=False) thi CHI dung title/snippet, KHONG truyen link do vao `WebReaderClient.read_url()`. Neu biet san trang chinh thuc (vd: `https://www.python.org/downloads/`), co the doc truc tiep bang `WebReaderClient.read_url()`.
- KHONG nuot loi im lang thanh 'N/A': in thang ket qua tho hoac traceback loi ra stdout/stderr.
- Cat gon moi doan tom tat toi da ~500 ky tu de in day du TAT CA cac muc va link nguon trong ~6000 ky tu.
- Chi tra ve 1 block code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve 1 block code Python.", run_p, 0.2))

        collected_data = ""
        best_data = ""
        heal_counts: dict[str, int] = {}
        max_attempts = 5
        for attempt in range(1, max_attempts + 1):
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)

            if ok and len(out.strip()) > len(best_data) and "Traceback (most recent call last)" not in out:
                best_data = out[:8000]

            fail_reason = summary
            if ok and len(out.strip()) > 80:
                gate_ok, missing_info = await verify_collected_data(client, question, out)
                if gate_ok:
                    collected_data = out[:8000]
                    run_success = True
                    print(f"   -> [OK lan {attempt}] Thu thap DU du lieu ({len(collected_data)} ky tu)!")
                    print(f"   -> [PREVIEW] {collected_data[:220].replace(chr(10), ' | ')}...")
                    break
                else:
                    fail_reason = f"DATA_INCOMPLETE: {missing_info}"
                    print(f"   -> [THIEU DU LIEU lan {attempt}] {missing_info}")
                    print(f"      [RAW OUT] {out[:200].replace(chr(10), ' | ')}")
            else:
                print(f"   -> [FAIL lan {attempt}] {summary}")

            if attempt == max_attempts:
                collected_data = best_data if best_data else out[-4000:]
                break

            diag_p = f"""Script thu thap du lieu bi loi hoac thieu du lieu khi chay thuc te.
Van de: {fail_reason}
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Output / Log:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi do `live_runner.py` (goi sai ham, doc sai kieu dict/str, truyen link RSS vao WebReaderClient, dung URL gia): tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": fail_reason}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath) and heal_counts.get(sname, 0) < 2:
                    heal_counts[sname] = heal_counts.get(sname, 0) + 1
                    print(f"   -> [SELF-HEAL #{heal_counts[sname]}] Va `{sname}.py` ({diag.get('reason')})...")
                    bpath = backup_skill(sname)
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc van de chay thuc te: {diag.get('reason')}

Code hien tai cua `{sname}.py`:
{FENCE}python
{old_src}
{FENCE}

Cach `live_runner.py` dang goi va doc ket qua tu tool nay:
{FENCE}python
{runner_code}
{FENCE}

Output / Log thuc te:
{out[-1500:]}

QUY TAC SUA BAT BUOC:
1. Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})`.
2. TUYET DOI KHONG nuot loi im lang tra ve rong/None; neu API loi phai raise RuntimeError kem status_code va response text.
3. Giu nguyen ten class/ham public va kieu tra ve de pass pytest.
Chi tra ve 1 block code Python cho `{sname}.py`."""
                    patched = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi tool Python. Chi tra ve 1 block code.", patch_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )
                    try:
                        ast.parse(patched)
                        with open(spath, "w", encoding="utf-8") as f:
                            f.write(patched)
                        pt_ok, pt_msg = verify_skill_pytest(sname)
                        if not pt_ok:
                            print(f"     -> Ban va `{sname}.py` khong qua pytest ({pt_msg}).")
                            restore_skill(sname, bpath)
                            diag["reason"] = f"Skill `{sname}` giu nguyen interface (ban va bi rollback). Hay sua `live_runner.py` de dung dung kieu tra ve cua skill."
                        else:
                            healed_skills.append(sname)
                            meta = sync_registry_meta()
                            meta[sname]["health"] = "PASS" if pt_msg == "PASS" else "NO_TEST"
                            save_meta(meta)
                            continue
                    except SyntaxError:
                        print(f"     -> Ban va `{sname}.py` loi cu phap.")
                        restore_skill(sname, bpath)
                        diag["reason"] = f"Ban va `{sname}` loi cu phap va da rollback. Hay sua `live_runner.py`."

            fix_run_p = f"""Sua `live_runner.py` de lay day du du lieu that va in ra stdout (KHONG dung example.com, KHONG truyen link `news.google.com/rss/articles/` vao WebReaderClient, doc dung kieu dict/list tra ve cua skill).
Van de: {diag.get('reason')}
Cac skill duoc phep import:
{reg_coder}
Code cu:
{FENCE}python
{runner_code}
{FENCE}
Output/Log:
{out[-1500:]}"""
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code.", fix_run_p, 0.2))

        update_skill_stats(meta, selected_for_coder, run_success)

        # BUOC 4: Analyst (qwq:32b) tong hop bao cao
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""THOI GIAN HIEN TAI CUA HE THONG: {vn_now} (nam 2026 la thoi diem hien tai that, KHONG phai tuong lai hay gia lap).
Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau BAT BUOC:
- Tra loi chinh xac, ro rang bang tieng Viet chuan (TUYET DOI khong chen chu Han/tieng Trung).
- Chi dua tren du lieu thu thap o tren; phan nao khong co du lieu thi noi ro la chua thu thap duoc.
- KHONG bao gio tu bia link URL; chi trich dan cac link co mat trong du lieu thu thap."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        elapsed = round(time.perf_counter() - t0, 2)
        append_run_log(
            {
                "timestamp": vn_now,
                "question": question,
                "use_skills": selected_for_coder,
                "forged": forged_ok,
                "forged_failed": failed_tools,
                "healed": healed_skills,
                "success": run_success,
                "elapsed_sec": elapsed,
            }
        )
        cleanup_sandbox()
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {elapsed}s\n")
        print("================ TRA LOI ================")
        print(final_ans)


def main():
    parser = argparse.ArgumentParser(description="AutoEvolve AI - Demand-Driven Self-Evolving Agent")
    parser.add_argument("question", nargs="*", help="Cau hoi can tra cuu")
    parser.add_argument("--health-check", action="store_true", help="Chay pytest kiem tra toan bo kho skills/ va don rac")
    parser.add_argument("--dry-run", action="store_true", help="Chi len ke hoach Architect, khong tao tool hay chay code")
    parser.add_argument("--no-forge", action="store_true", help="Chi dung skill san co, khong tao tool moi")
    args = parser.parse_args()

    if args.health_check:
        run_health_check()
        return

    q = " ".join(args.question).strip()
    if not q:
        parser.print_help()
        sys.exit(1)
    asyncio.run(solve_question(q, dry_run=args.dry_run, no_forge=args.no_forge))


if __name__ == "__main__":
    main()
````

````python
# Cell 1: Tạo memory_store.py

%%writefile /content/drive/MyDrive/AutoEvolve_AI/memory_store.py
"""Long-term SQLite + vector memory store for AutoEvolve AI."""
import os
import re
import math
import json
import sqlite3
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
DB_PATH = f"{BASE_DIR}/memory.db"
OLLAMA_BASE = "http://localhost:11434"
EMBED_MODEL = "nomic-embed-text"


def _now_vn() -> str:
    return datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")


def _tokenize(text: str) -> list[str]:
    return re.findall(r"[a-zA-Z0-9_]+|[\u00C0-\u1EF9]+", (text or "").lower())


def _cosine(v1: list[float], v2: list[float]) -> float:
    if not v1 or not v2 or len(v1) != len(v2):
        return 0.0
    dot = sum(a * b for a, b in zip(v1, v2))
    n1 = math.sqrt(sum(a * a for a in v1))
    n2 = math.sqrt(sum(b * b for b in v2))
    return dot / (n1 * n2) if n1 > 0 and n2 > 0 else 0.0


def _token_similarity(q: str, doc: str) -> float:
    t1 = set(_tokenize(q))
    t2 = set(_tokenize(doc))
    if not t1 or not t2:
        return 0.0
    inter = len(t1 & t2)
    return inter / math.sqrt(len(t1) * len(t2))


class MemoryStore:
    def __init__(self, db_path: str = DB_PATH, embed_model: str = EMBED_MODEL):
        self.db_path = db_path
        self.embed_model = embed_model
        self._embed_available: bool | None = None
        os.makedirs(os.path.dirname(db_path), exist_ok=True)
        self._init_db()

    def _conn(self) -> sqlite3.Connection:
        conn = sqlite3.connect(self.db_path)
        conn.row_factory = sqlite3.Row
        return conn

    def _init_db(self):
        with self._conn() as conn:
            conn.executescript(
                """
                CREATE TABLE IF NOT EXISTS episodes (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    timestamp TEXT NOT NULL,
                    question TEXT NOT NULL,
                    skills_used TEXT NOT NULL,
                    outcome TEXT NOT NULL,
                    error_pattern TEXT DEFAULT '',
                    lesson TEXT NOT NULL,
                    embedding TEXT DEFAULT ''
                );

                CREATE TABLE IF NOT EXISTS blacklist (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    target TEXT UNIQUE NOT NULL,
                    reason TEXT NOT NULL,
                    fail_count INTEGER DEFAULT 1,
                    updated_at TEXT NOT NULL
                );

                CREATE TABLE IF NOT EXISTS explorations (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    topic TEXT UNIQUE NOT NULL,
                    status TEXT NOT NULL,
                    skill_created TEXT DEFAULT '',
                    notes TEXT DEFAULT '',
                    updated_at TEXT NOT NULL
                );

                CREATE TABLE IF NOT EXISTS skill_vectors (
                    skill_name TEXT PRIMARY KEY,
                    description TEXT NOT NULL,
                    embedding TEXT DEFAULT '',
                    updated_at TEXT NOT NULL
                );
                """
            )

    def _embed(self, text: str) -> list[float]:
        if self._embed_available is False or not text.strip():
            return []
        snippet = text.strip()[:1500]
        try:
            r = httpx.post(
                f"{OLLAMA_BASE}/api/embed",
                json={"model": self.embed_model, "input": snippet},
                timeout=15.0,
            )
            if r.status_code == 200:
                data = r.json()
                embs = data.get("embeddings")
                if embs and isinstance(embs[0], list):
                    self._embed_available = True
                    return embs[0]
            r2 = httpx.post(
                f"{OLLAMA_BASE}/api/embeddings",
                json={"model": self.embed_model, "prompt": snippet},
                timeout=15.0,
            )
            if r2.status_code == 200 and "embedding" in r2.json():
                self._embed_available = True
                return r2.json()["embedding"]
            self._embed_available = False
        except Exception:
            self._embed_available = False
        return []

    def add_episode(
        self,
        question: str,
        skills_used: list[str],
        outcome: str,
        error_pattern: str,
        lesson: str,
    ):
        text_for_vec = f"{question} | {' '.join(skills_used)} | {error_pattern} | {lesson}"
        vec = self._embed(text_for_vec)
        with self._conn() as conn:
            conn.execute(
                """
                INSERT INTO episodes (timestamp, question, skills_used, outcome, error_pattern, lesson, embedding)
                VALUES (?, ?, ?, ?, ?, ?, ?)
                """,
                (
                    _now_vn(),
                    question,
                    json.dumps(skills_used, ensure_ascii=False),
                    outcome,
                    (error_pattern or "")[:400],
                    (lesson or "").strip()[:800],
                    json.dumps(vec) if vec else "",
                ),
            )

    def search_lessons(self, query: str, top_k: int = 3) -> list[dict]:
        q_vec = self._embed(query)
        with self._conn() as conn:
            rows = conn.execute(
                "SELECT id, timestamp, question, skills_used, outcome, error_pattern, lesson, embedding FROM episodes ORDER BY id DESC LIMIT 200"
            ).fetchall()

        scored = []
        for row in rows:
            if not row["lesson"]:
                continue
            doc_text = f"{row['question']} {row['skills_used']} {row['error_pattern']} {row['lesson']}"
            r_vec = json.loads(row["embedding"]) if row["embedding"] else []
            if q_vec and r_vec and len(q_vec) == len(r_vec):
                score = _cosine(q_vec, r_vec)
            else:
                score = _token_similarity(query, doc_text)
            if score >= 0.15:
                scored.append(
                    {
                        "score": round(score, 3),
                        "outcome": row["outcome"],
                        "lesson": row["lesson"],
                        "error_pattern": row["error_pattern"],
                        "skills_used": json.loads(row["skills_used"]),
                    }
                )
        scored.sort(key=lambda x: x["score"], reverse=True)
        return scored[:top_k]

    def add_blacklist(self, target: str, reason: str):
        target = (target or "").strip().lower()[:200]
        if not target:
            return
        with self._conn() as conn:
            conn.execute(
                """
                INSERT INTO blacklist (target, reason, fail_count, updated_at)
                VALUES (?, ?, 1, ?)
                ON CONFLICT(target) DO UPDATE SET
                    reason = excluded.reason,
                    fail_count = blacklist.fail_count + 1,
                    updated_at = excluded.updated_at
                """,
                (target, (reason or "")[:300], _now_vn()),
            )

    def get_blacklist(self, limit: int = 15) -> list[dict]:
        with self._conn() as conn:
            rows = conn.execute(
                "SELECT target, reason, fail_count, updated_at FROM blacklist ORDER BY fail_count DESC, updated_at DESC LIMIT ?",
                (limit,),
            ).fetchall()
        return [dict(r) for r in rows]

    def record_exploration(self, topic: str, status: str, skill_created: str = "", notes: str = ""):
        topic = (topic or "").strip()[:200]
        if not topic:
            return
        with self._conn() as conn:
            conn.execute(
                """
                INSERT INTO explorations (topic, status, skill_created, notes, updated_at)
                VALUES (?, ?, ?, ?, ?)
                ON CONFLICT(topic) DO UPDATE SET
                    status = excluded.status,
                    skill_created = CASE WHEN excluded.skill_created != '' THEN excluded.skill_created ELSE explorations.skill_created END,
                    notes = excluded.notes,
                    updated_at = excluded.updated_at
                """,
                (topic, status, skill_created, (notes or "")[:400], _now_vn()),
            )

    def get_explorations(self, limit: int = 50) -> list[dict]:
        with self._conn() as conn:
            rows = conn.execute(
                "SELECT topic, status, skill_created, notes, updated_at FROM explorations ORDER BY updated_at DESC LIMIT ?",
                (limit,),
            ).fetchall()
        return [dict(r) for r in rows]

    def sync_skill_vectors(self, meta: dict):
        with self._conn() as conn:
            existing = {
                r["skill_name"]: r["description"]
                for r in conn.execute("SELECT skill_name, description FROM skill_vectors").fetchall()
            }
            for mod, info in meta.items():
                desc = f"{mod}: {info.get('description', '')} | {info.get('signatures', '')}"
                if existing.get(mod) != desc:
                    vec = self._embed(desc)
                    conn.execute(
                        """
                        INSERT INTO skill_vectors (skill_name, description, embedding, updated_at)
                        VALUES (?, ?, ?, ?)
                        ON CONFLICT(skill_name) DO UPDATE SET
                            description = excluded.description,
                            embedding = excluded.embedding,
                            updated_at = excluded.updated_at
                        """,
                        (mod, desc, json.dumps(vec) if vec else "", _now_vn()),
                    )
            for dead in set(existing.keys()) - set(meta.keys()):
                conn.execute("DELETE FROM skill_vectors WHERE skill_name = ?", (dead,))

    def search_skills(self, query: str, meta: dict, top_k: int = 12) -> list[str]:
        if len(meta) <= top_k:
            return sorted(meta.keys())
        q_vec = self._embed(query)
        with self._conn() as conn:
            rows = conn.execute("SELECT skill_name, description, embedding FROM skill_vectors").fetchall()
        scored = []
        for r in rows:
            if r["skill_name"] not in meta:
                continue
            r_vec = json.loads(r["embedding"]) if r["embedding"] else []
            if q_vec and r_vec and len(q_vec) == len(r_vec):
                score = _cosine(q_vec, r_vec)
            else:
                score = _token_similarity(query, r["description"])
            scored.append((score, r["skill_name"]))
        scored.sort(key=lambda x: x[0], reverse=True)
        return [name for _, name in scored[:top_k]]

    def format_context_for_prompt(self, query: str) -> str:
        lessons = self.search_lessons(query, top_k=3)
        bl = self.get_blacklist(limit=10)
        parts = []
        if lessons:
            parts.append("BAI HOC TU CAC PHIEN TRUOC (Bo nho dai han):")
            for idx, item in enumerate(lessons, 1):
                parts.append(f"  {idx}. [{item['outcome']}] {item['lesson']}")
        if bl:
            parts.append("NGUON/API TRONG DANH SACH DEN (KHONG DUNG):")
            for item in bl:
                parts.append(f"  - {item['target']} ({item['reason']})")
        return "\n".join(parts)

    def get_stats(self) -> dict:
        with self._conn() as conn:
            ep = conn.execute("SELECT COUNT(*) FROM episodes").fetchone()[0]
            bl = conn.execute("SELECT COUNT(*) FROM blacklist").fetchone()[0]
            ex = conn.execute("SELECT COUNT(*) FROM explorations").fetchone()[0]
            sv = conn.execute("SELECT COUNT(*) FROM skill_vectors").fetchone()[0]
            recent = [
                dict(r)
                for r in conn.execute(
                    "SELECT timestamp, outcome, lesson FROM episodes ORDER BY id DESC LIMIT 5"
                ).fetchall()
            ]
        return {
            "db_path": self.db_path,
            "embed_model": self.embed_model,
            "episodes": ep,
            "blacklist": bl,
            "explorations": ex,
            "skill_vectors": sv,
            "recent_lessons": recent,
        }
````

````python
# Cell 2: Cập nhật live_evolve_agent.py tích hợp MemoryStore
````

Tự nạp bài học liên quan + blacklist từ memory.db vào prompt của cả Architect và Coder.

Khi kho vượt quá 25 skill, tự động dùng mem.search_skills() lọc đúng top skill liên quan nhất cho Architect.

Sau khi chạy xong, nếu có lỗi hoặc phải qua Self-Heal, qwq:32b tự đúc kết 1–2 câu bài học ngắn gọn lưu vào memory.db (chạy 1 phát ăn ngay thì ghi log nhanh, không tốn thêm lượt gọi LLM).

Tự động phát hiện lỗi 403/429 lặp lại từ một domain cụ thể để đưa vào blacklist.

Thêm cờ --memory-stats để soi nhanh nội dung memory.db.

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import shutil
import asyncio
import argparse
import subprocess
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
if BASE_DIR not in sys.path:
    sys.path.insert(0, BASE_DIR)

from memory_store import MemoryStore

SKILLS_DIR = f"{BASE_DIR}/skills"
SANDBOX_DIR = f"{BASE_DIR}/sandbox"
BACKUP_DIR = f"{BASE_DIR}/skill_backups"
META_PATH = f"{BASE_DIR}/registry_meta.json"
RUNS_LOG_PATH = f"{BASE_DIR}/runs.jsonl"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3
MAX_BACKUPS_PER_SKILL = 5
CORE_WEB_SKILLS = {"skill_web_search", "skill_web_reader"}


def now_vn_str() -> str:
    return datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str, module_name: str | None = None, is_test: bool = False) -> str:
    blocks = re.findall(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    if not blocks and FENCE in text:
        part = text.split(FENCE, 1)[1]
        part = re.sub(r"^python\s*", "", part, flags=re.IGNORECASE)
        blocks = [part]
    if not blocks:
        blocks = [text]

    valid = []
    for b in blocks:
        code = b.strip()
        if not code:
            continue
        if module_name and not is_test:
            if re.search(rf"\b(from|import)\s+{re.escape(module_name)}\b", code):
                continue
            if "def test_" in code and "class " not in code:
                continue
        try:
            ast.parse(code)
            valid.append(code)
        except SyntaxError:
            continue

    if valid:
        return max(valid, key=len)
    return blocks[0].strip()


def extract_json(text: str) -> dict:
    candidates = re.findall(FENCE + r"(?:json)?\s*(\{.*?\})\s*" + FENCE, text, re.DOTALL)
    candidates.append(text)
    decoder = json.JSONDecoder(strict=False)
    for chunk in candidates:
        pos = 0
        while True:
            idx = chunk.find("{", pos)
            if idx == -1:
                break
            try:
                obj, _ = decoder.raw_decode(chunk, idx)
                if isinstance(obj, dict):
                    return obj
            except json.JSONDecodeError:
                pass
            pos = idx + 1
    raise ValueError(f"Khong tim thay JSON hop le trong: {text[:200]}")


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def inspect_skill_file(filepath: str) -> tuple[str, str]:
    try:
        src = open(filepath, encoding="utf-8").read()
        tree = ast.parse(src)
        doc = (ast.get_docstring(tree) or "").splitlines()
        desc = doc[0].strip()[:120] if doc else ""
        sigs = []
        for n in tree.body:
            if isinstance(n, ast.ClassDef):
                if not desc:
                    cdoc = (ast.get_docstring(n) or "").splitlines()
                    if cdoc:
                        desc = cdoc[0].strip()[:120]
                ms = []
                for m in n.body:
                    if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef)) and (
                        not m.name.startswith("_") or m.name == "__init__"
                    ):
                        args_s = ", ".join(a.arg for a in m.args.args if a.arg != "self")
                        ret_s = f" -> {ast.unparse(m.returns)}" if m.returns else ""
                        ms.append(f"{m.name}({args_s}){ret_s}")
                sigs.append(f"class {n.name}: [{', '.join(ms)}]")
            elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                args_s = ", ".join(a.arg for a in n.args.args)
                ret_s = f" -> {ast.unparse(n.returns)}" if n.returns else ""
                sigs.append(f"def {n.name}({args_s}){ret_s}")
        return desc or "Python skill module", " | ".join(sigs)
    except Exception as e:
        return f"Parse error: {e}", ""


def load_meta() -> dict:
    if os.path.exists(META_PATH):
        try:
            with open(META_PATH, "r", encoding="utf-8") as f:
                return json.load(f)
        except Exception:
            pass
    return {}


def save_meta(meta: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(META_PATH, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)


def sync_registry_meta() -> dict:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    meta = load_meta()
    existing = set()
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        existing.add(mod)
        desc, sigs = inspect_skill_file(f"{SKILLS_DIR}/{fn}")
        entry = meta.get(mod, {})
        entry["signatures"] = sigs
        if not entry.get("description") or entry["description"] == "Python skill module":
            entry["description"] = desc
        entry.setdefault("success_count", 0)
        entry.setdefault("fail_count", 0)
        entry.setdefault("health", "UNKNOWN")
        entry.setdefault("updated_at", now_vn_str())
        meta[mod] = entry

    for dead in list(meta.keys()):
        if dead not in existing:
            del meta[dead]

    save_meta(meta)
    return meta


def format_registry_for_architect(meta: dict, subset: list[str] | None = None) -> str:
    if not meta:
        return "(Kho dang trong)"
    keys = sorted(subset) if subset else sorted(meta.keys())
    lines = []
    for mod in keys:
        if mod not in meta:
            continue
        info = meta[mod]
        stats = f"ok={info.get('success_count', 0)}, fail={info.get('fail_count', 0)}, health={info.get('health', 'UNKNOWN')}"
        lines.append(f"- `{mod}` ({stats}): {info.get('description', '')} | {info.get('signatures', '')}")
    return "\n".join(lines)


def format_registry_for_coder(meta: dict, selected: list[str] | None = None) -> str:
    if not meta:
        return "(Kho dang trong)"
    wanted = set()
    if selected:
        for s in selected:
            name = s.replace(".py", "").strip()
            if name in meta:
                wanted.add(name)
    for core in CORE_WEB_SKILLS:
        if core in meta:
            wanted.add(core)
    if not wanted:
        wanted = set(meta.keys())
    lines = [f"- `{m}`: {meta[m].get('signatures', '')}" for m in sorted(wanted)]
    return "\n".join(lines)


def update_skill_stats(meta: dict, skill_names: list[str], success: bool):
    ts = now_vn_str()
    for s in skill_names:
        name = s.replace(".py", "").strip()
        if name in meta:
            key = "success_count" if success else "fail_count"
            meta[name][key] = meta[name].get(key, 0) + 1
            meta[name]["last_used"] = ts
    save_meta(meta)


def append_run_log(record: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(RUNS_LOG_PATH, "a", encoding="utf-8") as f:
        f.write(json.dumps(record, ensure_ascii=False) + "\n")


def cleanup_backups(max_per_skill: int = MAX_BACKUPS_PER_SKILL):
    if not os.path.exists(BACKUP_DIR):
        return
    groups: dict[str, list[str]] = {}
    for fn in os.listdir(BACKUP_DIR):
        if not fn.endswith(".py.bak"):
            continue
        m = re.match(r"^(skill_.+)_\d{8}_\d{6}\.py\.bak$", fn)
        key = m.group(1) if m else fn
        groups.setdefault(key, []).append(os.path.join(BACKUP_DIR, fn))
    for files in groups.values():
        files.sort(key=lambda p: os.path.getmtime(p), reverse=True)
        for old_path in files[max_per_skill:]:
            try:
                os.remove(old_path)
            except OSError:
                pass


def cleanup_sandbox():
    if not os.path.exists(SANDBOX_DIR):
        return
    for name in os.listdir(SANDBOX_DIR):
        if name.startswith(("forge_", "heal_verify", "health_")):
            full = os.path.join(SANDBOX_DIR, name)
            try:
                if os.path.isdir(full):
                    shutil.rmtree(full, ignore_errors=True)
                else:
                    os.remove(full)
            except OSError:
                pass


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:{BASE_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:160] if errs else (out.replace("\n", " | ")[:160] if out else "Output rong")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


def backup_skill(sname: str) -> str | None:
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return None
    os.makedirs(BACKUP_DIR, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    bpath = f"{BACKUP_DIR}/{sname}_{ts}.py.bak"
    shutil.copy2(spath, bpath)
    cleanup_backups()
    return bpath


def restore_skill(sname: str, bpath: str | None):
    spath = f"{SKILLS_DIR}/{sname}.py"
    if bpath and os.path.exists(bpath):
        shutil.copy2(bpath, spath)
        print(f"     -> [ROLLBACK] Da khoi phuc ban goc cua `{sname}.py` do ban va chua dat.")


def verify_skill_pytest(sname: str) -> tuple[bool, str]:
    tpath = f"{SKILLS_DIR}/test_{sname}.py"
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return False, "Khong tim thay file skill"
    if not os.path.exists(tpath):
        return True, "NO_TEST"
    vdir = f"{SANDBOX_DIR}/heal_verify_{sname}"
    os.makedirs(vdir, exist_ok=True)
    try:
        shutil.copy2(tpath, f"{vdir}/test_{sname}.py")
        shutil.copy2(spath, f"{vdir}/{sname}.py")
        ok, _, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], vdir, 20)
        return ok, ("PASS" if ok else summary)
    finally:
        shutil.rmtree(vdir, ignore_errors=True)


def run_health_check():
    print("=== HEALTH CHECK KHO SKILLS ===")
    meta = sync_registry_meta()
    mem = MemoryStore()
    mem.sync_skill_vectors(meta)
    if not meta:
        print("Kho skills/ dang trong.")
        return
    passed = failed = no_test = 0
    for mod in sorted(meta.keys()):
        spath = f"{SKILLS_DIR}/{mod}.py"
        try:
            ast.parse(open(spath, encoding="utf-8").read())
        except SyntaxError as e:
            meta[mod]["health"] = "SYNTAX_ERROR"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> SyntaxError: {e}")
            continue

        ok, msg = verify_skill_pytest(mod)
        if msg == "NO_TEST":
            meta[mod]["health"] = "NO_TEST"
            no_test += 1
            print(f"  [WARN] {mod:<35} -> Khong co file test")
        elif ok:
            meta[mod]["health"] = "PASS"
            passed += 1
            print(f"  [PASS] {mod:<35} -> pytest OK")
        else:
            meta[mod]["health"] = "FAIL"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> {msg}")
        meta[mod]["last_health_check"] = now_vn_str()

    save_meta(meta)
    cleanup_backups()
    cleanup_sandbox()
    print(f"\nTong ket: {passed} PASS | {failed} FAIL | {no_test} NO_TEST (da cap nhat registry_meta.json, memory.db & don rac).")


def show_memory_stats():
    mem = MemoryStore()
    stats = mem.get_stats()
    print("=== TRANG THAI BO NHO DAI HAN (memory.db) ===")
    print(f"  File DB        : {stats['db_path']}")
    print(f"  Embed Model    : {stats['embed_model']}")
    print(f"  Episodes       : {stats['episodes']} bai hoc")
    print(f"  Blacklist      : {stats['blacklist']} nguon bi chan")
    print(f"  Explorations   : {stats['explorations']} chu de da nghien cuu")
    print(f"  Skill Vectors  : {stats['skill_vectors']} skill da lap chi muc")
    if stats["recent_lessons"]:
        print("\n5 bai hoc gan nhat:")
        for r in stats["recent_lessons"]:
            print(f"  - [{r['timestamp']}] ({r['outcome']}) {r['lesson']}")
    bl = mem.get_blacklist(10)
    if bl:
        print("\nDanh sach den:")
        for b in bl:
            print(f"  - {b['target']} (fail={b['fail_count']}): {b['reason']}")


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong (cho phep them tham so tuy chon co gia tri mac dinh)?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def verify_collected_data(client: httpx.AsyncClient, question: str, data: str) -> tuple[bool, str]:
    ver_p = f"""Cau hoi nguoi dung: {question}

Du lieu script vua thu thap duoc:
{data[:8000]}

Kiem tra:
- Du lieu tren da cung cap du thong tin thuc te de tra loi cau hoi chua?
- LUU Y: Neu cau hoi dung tu "hoac" / "or" (vi du "A hoac B"), chi can co du lieu tot cho A HOAC B la DAT (PASS), KHONG bat buoc phai co ca hai.
- Chi danh FAIL neu du lieu bi loi/Traceback/rong/N/A/example.com/404, hoac thieu mot ve bat buoc ("va").
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
MISSING: <neu FAIL, ghi ro phan nao bi thieu hoac rong; neu PASS ghi None>"""
    res = await ask_llm(client, BRAIN_THINK, "Ban la QA Data Verifier chinh xac.", ver_p, 0.1)
    ok = "STATUS: PASS" in res.upper()
    m = re.search(r"MISSING:\s*(.+)", res)
    return ok, (m.group(1).strip() if m else res.splitlines()[-1][:120])


async def distill_lesson(
    client: httpx.AsyncClient,
    question: str,
    skills: list[str],
    errors: list[str],
    healed: list[str],
    success: bool,
) -> tuple[str, list[tuple[str, str]]]:
    """Return (concise_lesson, [(blocked_target, reason), ...])."""
    prompt = f"""Tom tat 1 bai hoc ky thuat ngan gon (1-2 cau) de cac phien sau tranh lap lai loi:
- Cau hoi: {question}
- Skills da dung: {skills}
- Ket qua: {'PASS' if success else 'FAIL'}
- Skills da Self-Heal: {healed}
- Cac loi gap phai: {errors[-3:]}

Tra ve DUY NHAT JSON:
{{
  "lesson": "1-2 cau bai hoc cu the (vi du: khong dung WebReaderClient cho link news.google.com/rss/articles vi bi 429)",
  "blacklist": [
    {{"target": "ten domain hoac API bi chan/doi key (chi them neu that su bi 403/429/API key)", "reason": "ly do ngan gon"}}
  ]
}}"""
    try:
        raw = await ask_llm(client, BRAIN_THINK, "Ban la ky su rut kinh nghiem he thong. Chi tra ve JSON.", prompt, 0.1)
        data = extract_json(raw)
        lesson = str(data.get("lesson", "")).strip()
        bl = [
            (str(b.get("target", "")).strip(), str(b.get("reason", "")).strip())
            for b in (data.get("blacklist") or [])
            if isinstance(b, dict) and b.get("target")
        ]
        return lesson, bl
    except Exception:
        fallback = f"Dung {skills} cho tac vu tuong tu; loi tung gap: {errors[-1][:140] if errors else 'khong'}"
        return fallback, []


async def forge_tool(client: httpx.AsyncClient, spec: dict, meta: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3 test case ro rang (1 happy path + 1 loi HTTP status != 200 + 1 loi du lieu rong/mang).
- Code thuc thi dung `httpx.get(...)`, nen mock bang `unittest.mock.patch("httpx.get")` (KHONG mock `requests`).
- Khi test truong hop loi, LUON dung `with pytest.raises(Exception):`.
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve duy nhat 1 block code Python cho `test_{sname}.py`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- LUON dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})` (KHONG dung `requests`).
- KHONG nuot loi mang im lang (neu HTTP status != 200 hoac du lieu loi, raise `RuntimeError` kem status_code va noi dung loi).
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY (Open-Meteo, MediaWiki/Wikipedia, ArXiv, OpenAlex, Google News RSS, Frankfurter, Binance public).
- Tuyet doi khong hardcode du lieu gia, khong viet code test trong file nay.
Chi tra ve duy nhat 1 block code Python cho `{sname}.py`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve 1 block code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0], module_name=sname, is_test=True)
    candidates = [extract_code(c, module_name=sname, is_test=False) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    def save_forged(code_str: str, test_str: str):
        with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(code_str)
        with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_str)
        _, sigs = inspect_skill_file(f"{SKILLS_DIR}/{sname}.py")
        meta[sname] = {
            "description": goal[:120],
            "signatures": sigs,
            "success_count": 0,
            "fail_count": 0,
            "health": "PASS",
            "updated_at": now_vn_str(),
        }
        save_meta(meta)

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    save_forged(curr, test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"""Sua file `{sname}.py` de pass pytest (van dung `httpx.get(..., follow_redirects=True)` va API that).
CHI tra ve code cua `{sname}.py`, KHONG tra ve file test.
Code:
{FENCE}python
{curr}
{FENCE}
Test:
{FENCE}python
{test_code}
{FENCE}
Log:
{log[-1500:]}"""
                    curr = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code cho file skill.", fix_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )

    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, ep sai loai exception, sai cau truc response):
Tra ve `VERDICT: TEST_BUG` kem toan bo file `test_{sname}.py` da sua trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj, module_name=sname, is_test=True)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                save_forged(last_code0, fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(question: str, dry_run: bool = False, no_forge: bool = False) -> dict:
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)
    cleanup_sandbox()
    cleanup_backups()
    vn_now = now_vn_str()
    meta = sync_registry_meta()
    mem = MemoryStore()
    mem.sync_skill_vectors(meta)

    memory_ctx = mem.format_context_for_prompt(question)
    forged_ok: list[str] = []
    failed_tools: list[str] = []
    healed_skills: list[str] = []
    session_errors: list[str] = []
    run_success = False
    attempts_used = 0

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) phan tich nhu cau + doc bo nho dai han
        if len(meta) > 25:
            top_skills = list(set(mem.search_skills(question, meta, top_k=15)) | (CORE_WEB_SKILLS & set(meta.keys())))
            reg_arch = format_registry_for_architect(meta, subset=top_skills)
        else:
            reg_arch = format_registry_for_architect(meta)

        mem_block = f"\n{memory_ctx}\n" if memory_ctx else ""
        plan_p = f"""CAU HOI NGUOI DUNG: {question}
THOI GIAN HIEN TAI: {vn_now}
{mem_block}
KHO SKILL HIEN CO (tu registry_meta.json):
{reg_arch}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
   (Luu y: da co `skill_web_search` va `skill_web_reader` de tim kiem & doc bat ky trang web nao; chi tao tool moi neu can goi 1 API chuyen biet chua co).
2. Neu THIEU tool chuyen biet, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...) hoac nguon trong DANH SACH DEN.
   - Thoi tiet -> Open-Meteo; Kien thuc -> MediaWiki/Wikipedia; Khoa hoc -> ArXiv/OpenAlex; Tai chinh -> Frankfurter/Binance public; Web/Tin tuc -> dung san `skill_web_search` + `skill_web_reader`.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT 1 JSON hop le, khong kem giai thich ben ngoai:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        if memory_ctx:
            print(f"[MEM] Da nap kinh nghiem & danh sach den tu memory.db")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra registry_meta.json...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve 1 JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        use_skills = plan.get("use_skills", [])
        print(f"   -> Dung skills: {use_skills} | Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        if dry_run:
            print("\n[DRY-RUN] Ke hoach Architect:")
            print(json.dumps(plan, ensure_ascii=False, indent=2))
            return {"success": True, "dry_run": True, "plan": plan}

        # BUOC 2: Forge tu che tao tool con thieu
        if not no_forge:
            for spec in missing:
                ok_forge = await forge_tool(client, spec, meta)
                if ok_forge:
                    sname = spec["skill_name"]
                    if not sname.startswith("skill_"):
                        sname = "skill_" + sname.lstrip("_")
                    forged_ok.append(sname)
                else:
                    failed_tools.append(spec["skill_name"])
        elif missing:
            print(f"   -> [--no-forge] Bo qua tao moi: {[m['skill_name'] for m in missing]}")
            failed_tools = [m["skill_name"] for m in missing]

        # BUOC 3: Coder chay thuc chien + Data Quality Gate + Self-Healing + Rollback
        meta = sync_registry_meta()
        mem.sync_skill_vectors(meta)
        selected_for_coder = list(dict.fromkeys(use_skills + forged_ok))
        reg_coder = format_registry_for_coder(meta, selected_for_coder)
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        failed_note = ""
        if failed_tools:
            failed_note = f"\nLUU Y QUAN TRONG: Cac tool sau CHUA DUOC TAO nen KHONG import: {failed_tools}. Hay dung `skill_web_search` / `skill_web_reader` hoac goi truc tiep API mo bang `httpx.get(..., follow_redirects=True)`."

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
{mem_block}
Cac skill duoc phep import (chu y ky kieu tra ve `-> dict`, `-> list[dict]`):
{reg_coder}
{failed_note}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau BAT BUOC:
- Lay day du du lieu cho TAT CA cac ve cua cau hoi.
- TUYET DOI KHONG dung URL gia nhu `example.com`; neu `skill_web_search` tra ve link `news.google.com/rss/articles/` (readable=False) thi CHI dung title/snippet, KHONG truyen link do vao `WebReaderClient.read_url()`. Neu biet san trang chinh thuc (vd: `https://www.python.org/downloads/`), co the doc truc tiep bang `WebReaderClient.read_url()`.
- KHONG nuot loi im lang thanh 'N/A': in thang ket qua tho hoac traceback loi ra stdout/stderr.
- Cat gon moi doan tom tat toi da ~500 ky tu de in day du TAT CA cac muc va link nguon trong ~6000 ky tu.
- Chi tra ve 1 block code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve 1 block code Python.", run_p, 0.2))

        collected_data = ""
        best_data = ""
        heal_counts: dict[str, int] = {}
        max_attempts = 5
        for attempt in range(1, max_attempts + 1):
            attempts_used = attempt
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)

            if ok and len(out.strip()) > len(best_data) and "Traceback (most recent call last)" not in out:
                best_data = out[:8000]

            fail_reason = summary
            if ok and len(out.strip()) > 80:
                gate_ok, missing_info = await verify_collected_data(client, question, out)
                if gate_ok:
                    collected_data = out[:8000]
                    run_success = True
                    print(f"   -> [OK lan {attempt}] Thu thap DU du lieu ({len(collected_data)} ky tu)!")
                    print(f"   -> [PREVIEW] {collected_data[:220].replace(chr(10), ' | ')}...")
                    break
                else:
                    fail_reason = f"DATA_INCOMPLETE: {missing_info}"
                    session_errors.append(fail_reason)
                    print(f"   -> [THIEU DU LIEU lan {attempt}] {missing_info}")
                    print(f"      [RAW OUT] {out[:200].replace(chr(10), ' | ')}")
            else:
                session_errors.append(summary)
                print(f"   -> [FAIL lan {attempt}] {summary}")

            if attempt == max_attempts:
                collected_data = best_data if best_data else out[-4000:]
                break

            diag_p = f"""Script thu thap du lieu bi loi hoac thieu du lieu khi chay thuc te.
Van de: {fail_reason}
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Output / Log:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi do `live_runner.py` (goi sai ham, doc sai kieu dict/str, truyen link RSS vao WebReaderClient, dung URL gia): tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": fail_reason}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath) and heal_counts.get(sname, 0) < 2:
                    heal_counts[sname] = heal_counts.get(sname, 0) + 1
                    print(f"   -> [SELF-HEAL #{heal_counts[sname]}] Va `{sname}.py` ({diag.get('reason')})...")
                    bpath = backup_skill(sname)
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc van de chay thuc te: {diag.get('reason')}

Code hien tai cua `{sname}.py`:
{FENCE}python
{old_src}
{FENCE}

Cach `live_runner.py` dang goi va doc ket qua tu tool nay:
{FENCE}python
{runner_code}
{FENCE}

Output / Log thuc te:
{out[-1500:]}

QUY TAC SUA BAT BUOC:
1. Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})`.
2. TUYET DOI KHONG nuot loi im lang tra ve rong/None; neu API loi phai raise RuntimeError kem status_code va response text.
3. Giu nguyen ten class/ham public va kieu tra ve de pass pytest.
Chi tra ve 1 block code Python cho `{sname}.py`."""
                    patched = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi tool Python. Chi tra ve 1 block code.", patch_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )
                    try:
                        ast.parse(patched)
                        with open(spath, "w", encoding="utf-8") as f:
                            f.write(patched)
                        pt_ok, pt_msg = verify_skill_pytest(sname)
                        if not pt_ok:
                            print(f"     -> Ban va `{sname}.py` khong qua pytest ({pt_msg}).")
                            restore_skill(sname, bpath)
                            diag["reason"] = f"Skill `{sname}` giu nguyen interface (ban va bi rollback). Hay sua `live_runner.py` de dung dung kieu tra ve cua skill."
                        else:
                            healed_skills.append(sname)
                            meta = sync_registry_meta()
                            meta[sname]["health"] = "PASS" if pt_msg == "PASS" else "NO_TEST"
                            save_meta(meta)
                            continue
                    except SyntaxError:
                        print(f"     -> Ban va `{sname}.py` loi cu phap.")
                        restore_skill(sname, bpath)
                        diag["reason"] = f"Ban va `{sname}` loi cu phap va da rollback. Hay sua `live_runner.py`."

            fix_run_p = f"""Sua `live_runner.py` de lay day du du lieu that va in ra stdout (KHONG dung example.com, KHONG truyen link `news.google.com/rss/articles/` vao WebReaderClient, doc dung kieu dict/list tra ve cua skill).
Van de: {diag.get('reason')}
Cac skill duoc phep import:
{reg_coder}
Code cu:
{FENCE}python
{runner_code}
{FENCE}
Output/Log:
{out[-1500:]}"""
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code.", fix_run_p, 0.2))

        update_skill_stats(meta, selected_for_coder, run_success)

        # Luu bai hoc vao bo nho dai han (chi goi LLM duc ket khi co loi / Self-Heal de khong ton thoi gian phien sach)
        if session_errors or healed_skills or not run_success:
            lesson, bl_items = await distill_lesson(
                client, question, selected_for_coder, session_errors, healed_skills, run_success
            )
            mem.add_episode(
                question=question,
                skills_used=selected_for_coder,
                outcome="PASS" if run_success else "FAIL",
                error_pattern=" | ".join(session_errors[-2:]),
                lesson=lesson,
            )
            for target, reason in bl_items:
                mem.add_blacklist(target, reason)
            print(f"   -> [MEM SAVE] Da luu bai hoc vao memory.db: {lesson[:140]}")
        else:
            mem.add_episode(
                question=question,
                skills_used=selected_for_coder,
                outcome="PASS",
                error_pattern="",
                lesson=f"Thanh cong ngay luot 1 bang {selected_for_coder} cho dang cau hoi tuong tu.",
            )

        # BUOC 4: Analyst (qwq:32b) tong hop bao cao
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""THOI GIAN HIEN TAI CUA HE THONG: {vn_now} (nam 2026 la thoi diem hien tai that, KHONG phai tuong lai hay gia lap).
Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau BAT BUOC:
- Tra loi chinh xac, ro rang bang tieng Viet chuan (TUYET DOI khong chen chu Han/tieng Trung).
- Chi dua tren du lieu thu thap o tren; phan nao khong co du lieu thi noi ro la chua thu thap duoc.
- KHONG bao gio tu bia link URL; chi trich dan cac link co mat trong du lieu thu thap."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        elapsed = round(time.perf_counter() - t0, 2)
        result_record = {
            "timestamp": vn_now,
            "question": question,
            "use_skills": selected_for_coder,
            "forged": forged_ok,
            "forged_failed": failed_tools,
            "healed": healed_skills,
            "attempts": attempts_used,
            "success": run_success,
            "elapsed_sec": elapsed,
        }
        append_run_log(result_record)
        cleanup_sandbox()
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {elapsed}s\n")
        print("================ TRA LOI ================")
        print(final_ans)
        return result_record


def main():
    parser = argparse.ArgumentParser(description="AutoEvolve AI - Demand-Driven Self-Evolving Agent")
    parser.add_argument("question", nargs="*", help="Cau hoi can tra cuu")
    parser.add_argument("--health-check", action="store_true", help="Chay pytest kiem tra toan bo kho skills/ va don rac")
    parser.add_argument("--memory-stats", action="store_true", help="Xem trang thai bo nho dai han memory.db")
    parser.add_argument("--dry-run", action="store_true", help="Chi len ke hoach Architect, khong tao tool hay chay code")
    parser.add_argument("--no-forge", action="store_true", help="Chi dung skill san co, khong tao tool moi")
    args = parser.parse_args()

    if args.health_check:
        run_health_check()
        return

    if args.memory_stats:
        show_memory_stats()
        return

    q = " ".join(args.question).strip()
    if not q:
        parser.print_help()
        sys.exit(1)
    asyncio.run(solve_question(q, dry_run=args.dry_run, no_forge=args.no_forge))


if __name__ == "__main__":
    main()
````

Bước tiếp theo là Bước 2: Tạo auto_researcher.py — vòng lặp tự vận hành có kiểm soát ngân sách với 3 nhiệm vụ theo thứ tự ưu tiên:

Tự bảo trì ([MAINTENANCE]): Tự quét registry_meta.json, thấy skill nào FAIL (hiện tại là skill_arxiv_api_client và skill_wikipedia_api_client) thì đọc code + log pytest, dùng qwq:32b phân định lỗi do file test cũ bị lệch hay do code, tự sửa kèm backup/rollback cho tới khi xanh 20/20 PASS.

Tự nghiên cứu có định hướng ([EXPLORE]): Dựa trên kho skill hiện có, danh sách đã làm (explorations), và blacklist trong memory.db, tự chọn 1 lĩnh vực Zero-Key còn thiếu (GitHub API, PyPI/Crates.io, Hacker News, Hugging Face Hub, World Bank, kiểm tra DNS/SSL...), tự đặt câu hỏi thực chiến rồi gọi thẳng solve_question() để tự rèn tool mới và nghiệm thu bằng dữ liệu thật.

Đề xuất nâng cấp lõi an toàn ([SELF-IMPROVE]): Cuối phiên đọc lại runs.jsonl và các lỗi lặp lại trong memory.db, xuất đề xuất cải tiến ra thư mục proposed_patches/ để bạn duyệt (tuyệt đối không tự ghi đè live_evolve_agent.py).

````python
# Cell 1: Tạo auto_researcher.py

%%writefile /content/drive/MyDrive/AutoEvolve_AI/auto_researcher.py
"""Autonomous Maintenance + Directed Self-Evolution Loop for AutoEvolve AI."""
import os
import sys
import json
import time
import shutil
import sqlite3
import asyncio
import argparse
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
if BASE_DIR not in sys.path:
    sys.path.insert(0, BASE_DIR)

from memory_store import MemoryStore, DB_PATH
import live_evolve_agent as agent

PATCHES_DIR = f"{BASE_DIR}/proposed_patches"

DEFAULT_CURRICULUM = [
    "GitHub Public REST API (tra cuu repo trending, release moi nhat, thong ke stars/issues khong can token)",
    "PyPI & Crates.io Public API (tra cuu phien ban moi nhat, thong tin goi Python va Rust)",
    "Hacker News Firebase API (lay top stories, bai thao luan cong nghe hot nhat)",
    "Hugging Face Public Hub API (tim kiem model/dataset moi, luot tai, thong so khong can token)",
    "World Bank Open Data API (tra cuu chi so kinh te vi mo: GDP, dan so, lam phat theo quoc gia)",
    "OpenStreetMap Nominatim / Overpass (tra cuu toa do dia ly, thong tin thanh pho khong can key)",
    "Network & SSL Inspector (kiem tra DNS, chung chi SSL/TLS, HTTP headers, thoi gian phan hoi cua 1 domain)",
    "YouTube / Reddit / RSS Feed Aggregator (doc feed RSS cong khai tu cac blog ky thuat, Rust/Python blog)",
]


def reindex_empty_vectors(meta: dict, mem: MemoryStore):
    """Re-embed skill_vectors rows that were saved before nomic-embed-text was pulled."""
    with sqlite3.connect(DB_PATH) as conn:
        conn.execute("DELETE FROM skill_vectors WHERE embedding = '' OR embedding IS NULL")
    mem.sync_skill_vectors(meta)


async def maintain_failed_skill(client: httpx.AsyncClient, sname: str, meta: dict, mem: MemoryStore) -> bool:
    """Repair a failing skill or its out-of-sync unit test with backup & rollback."""
    spath = f"{agent.SKILLS_DIR}/{sname}.py"
    tpath = f"{agent.SKILLS_DIR}/test_{sname}.py"
    if not os.path.exists(spath) or not os.path.exists(tpath):
        return False

    vdir = f"{agent.SANDBOX_DIR}/maint_{sname}"
    os.makedirs(vdir, exist_ok=True)
    shutil.copy2(spath, f"{vdir}/{sname}.py")
    shutil.copy2(tpath, f"{vdir}/test_{sname}.py")
    ok, log, summary = agent.run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], vdir, 20)
    shutil.rmtree(vdir, ignore_errors=True)
    if ok:
        meta[sname]["health"] = "PASS"
        agent.save_meta(meta)
        return True

    print(f"\n[MAINTENANCE] Dang sua `{sname}` (loi: {summary})")
    skill_src = open(spath, encoding="utf-8").read()
    test_src = open(tpath, encoding="utf-8").read()

    prompt = f"""Skill `{sname}.py` dang bi FAIL pytest trong dot health-check.
Luu y: Nhieu skill da duoc Self-Heal va chay thuc te tot, nhung file `test_{sname}.py` cu bi lech interface/mock.

--- `{sname}.py` ---
{agent.FENCE}python
{skill_src}
{agent.FENCE}

--- `test_{sname}.py` ---
{agent.FENCE}python
{test_src}
{agent.FENCE}

--- PYTEST LOG ---
{log[-2000:]}

Nhiem vu:
- Neu code `{sname}.py` da dung logic API that va chi do `test_{sname}.py` bi lech mock/assert -> tra ve `TARGET: TEST` kem toan bo file `test_{sname}.py` da sua trong `{agent.FENCE}python ... {agent.FENCE}`.
- Neu code `{sname}.py` bi loi logic that -> tra ve `TARGET: CODE` kem toan bo file `{sname}.py` da sua trong `{agent.FENCE}python ... {agent.FENCE}`.
- Khi mock mang trong test, dung `unittest.mock.patch("httpx.get")` va `with pytest.raises(Exception):` cho test loi."""

    raw = await agent.ask_llm(client, agent.BRAIN_THINK, "Ban la Staff Maintenance Engineer.", prompt, 0.15)
    is_test_fix = "TARGET: TEST" in raw.upper()
    fixed_code = agent.extract_code(raw, module_name=sname, is_test=is_test_fix)

    backup_skill_path = agent.backup_skill(sname)
    old_test_src = test_src

    try:
        if is_test_fix:
            with open(tpath, "w", encoding="utf-8") as f:
                f.write(fixed_code)
            print(f"   -> Cap nhat `test_{sname}.py` theo interface thuc te...")
        else:
            with open(spath, "w", encoding="utf-8") as f:
                f.write(fixed_code)
            print(f"   -> Cap nhat code `{sname}.py`...")

        pt_ok, pt_msg = agent.verify_skill_pytest(sname)
        if pt_ok:
            meta = agent.sync_registry_meta()
            meta[sname]["health"] = "PASS"
            meta[sname]["last_health_check"] = agent.now_vn_str()
            agent.save_meta(meta)
            mem.add_episode(
                question=f"[MAINTENANCE] Fix {sname}",
                skills_used=[sname],
                outcome="PASS",
                error_pattern=summary,
                lesson=f"Da dong bo {'test' if is_test_fix else 'code'} cho `{sname}` thanh cong va pass pytest.",
            )
            print(f"   -> [MAINTENANCE OK] `{sname}` da chuyen sang PASS!")
            return True
        else:
            print(f"   -> [MAINTENANCE FAIL] Van fail pytest ({pt_msg}) -> Rollback!")
            if is_test_fix:
                with open(tpath, "w", encoding="utf-8") as f:
                    f.write(old_test_src)
            else:
                agent.restore_skill(sname, backup_skill_path)
            return False
    except Exception as e:
        print(f"   -> [MAINTENANCE ERROR] {e} -> Rollback!")
        if is_test_fix:
            with open(tpath, "w", encoding="utf-8") as f:
                f.write(old_test_src)
        else:
            agent.restore_skill(sname, backup_skill_path)
        return False


async def propose_next_research(
    client: httpx.AsyncClient,
    meta: dict,
    mem: MemoryStore,
    curriculum: list[str],
) -> dict | None:
    """Pick 1 missing high-value zero-key capability and craft a real test question."""
    past = mem.get_explorations(limit=30)
    bl = mem.get_blacklist(limit=15)
    existing_skills = [f"- `{k}`: {v.get('description', '')}" for k, v in sorted(meta.items())]

    prompt = f"""Ban la AI Research Director cua he thong AutoEvolve AI.
Nhiem vu: Chon DUNG 1 nang luc/API con THIEU trong kho skill de tu nghien cuu va tao tool moi trong vong nay.

KHO SKILL HIEN CO:
{chr(10).join(existing_skills)}

DANH SACH LINH VUC UU TIEN (Zero-Key, khong can API key):
{chr(10).join(f'- {c}' for c in curriculum)}

CAC CHU DE DA NGHIEN CUU (KHONG chon lap lai):
{json.dumps([p['topic'] for p in past], ensure_ascii=False)}

DANH SACH DEN (KHONG dung):
{json.dumps([b['target'] for b in bl], ensure_ascii=False)}

Yeu cau:
- Chon 1 chu de cu the CHUA CO tool chuyen biet trong kho skill (va chua nam trong danh sach da nghien cuu).
- Dat 1 cau hoi kiem thu thuc te, ro rang, co the xac minh duoc bang du lieu that (vi du: lay top 3 repo GitHub, tra phien ban moi nhat cua `tokio` tren crates.io va `httpx` tren PyPI, lay top 5 bai tren Hacker News...).
- Cau hoi phai khuyen khich tao 1 tool chuyen biet moi (thay vi chi search Google chung chung).

Tra ve DUY NHAT 1 JSON:
{{
  "topic": "ten ngan gon chu de (vd: github_public_api)",
  "expected_skill": "ten skill ky vong (vd: skill_github_api_client)",
  "question": "cau hoi tieng Viet thuc te de giao cho live_evolve_agent chay va nghiem thu",
  "rationale": "ly do ngan gon tai sao tool nay huu ich"
}}"""
    raw = await agent.ask_llm(client, agent.BRAIN_THINK, "Ban la AI Research Director. Chi tra ve 1 JSON.", prompt, 0.5)
    try:
        return agent.extract_json(raw)
    except Exception as e:
        print(f"[EXPLORE] Khong parse duoc de xuat nghien cuu: {e}")
        return None


async def generate_core_improvement_proposal(client: httpx.AsyncClient, mem: MemoryStore):
    """Analyze recent failures and write a human-reviewable improvement report (never auto-edits core)."""
    runs = []
    if os.path.exists(agent.RUNS_LOG_PATH):
        with open(agent.RUNS_LOG_PATH, encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if line:
                    try:
                        runs.append(json.loads(line))
                    except Exception:
                        pass

    recent_runs = runs[-15:]
    failed_runs = [r for r in recent_runs if not r.get("success") or r.get("forged_failed") or r.get("healed")]
    if not failed_runs:
        print("\n[SELF-IMPROVE] Cac phien gan day deu chay sach, chua can de xuat sua file loi.")
        return

    stats = mem.get_stats()
    prompt = f"""Phan tich lich su chay gan day cua AutoEvolve AI va de xuat cai tien cho file loi `live_evolve_agent.py` (KHONG sua truc tiep, chi viet bao cao de xuat cho lap trinh vien duyet).

Cac phien co loi / Self-Heal / Forge fail gan day:
{json.dumps(failed_runs, ensure_ascii=False, indent=2)}

5 bai hoc gan nhat trong memory.db:
{json.dumps(stats['recent_lessons'], ensure_ascii=False, indent=2)}

Viet bao cao Markdown ngan gon (tieng Viet) gom:
1. Diem nghen lap lai nhieu nhat
2. Nguyen nhan goc re
3. De xuat sua cu the trong `live_evolve_agent.py` (kem doan code/prompt goi y neu can)"""

    report = await agent.ask_llm(client, agent.BRAIN_THINK, "Ban la Principal Systems Architect.", prompt, 0.2)
    os.makedirs(PATCHES_DIR, exist_ok=True)
    ts = time.strftime("%Y%m%d_%H%M%S")
    out_path = f"{PATCHES_DIR}/proposal_{ts}.md"
    with open(out_path, "w", encoding="utf-8") as f:
        f.write(f"# De xuat nang cap he thong ({agent.now_vn_str()})\n\n{report}\n")
    print(f"\n[SELF-IMPROVE] Da luu bao cao de xuat nang cap tai: {out_path}")


async def run_autonomous_loop(
    max_cycles: int = 3,
    budget_minutes: float = 20.0,
    maintenance_only: bool = False,
):
    t_start = time.perf_counter()
    budget_sec = budget_minutes * 60.0
    meta = agent.sync_registry_meta()
    mem = MemoryStore()
    reindex_empty_vectors(meta, mem)

    print("============================================================")
    print(f"AUTO-RESEARCHER KHOI DONG | max_cycles={max_cycles} | budget={budget_minutes} phut")
    print("============================================================")

    async with httpx.AsyncClient() as client:
        # GIAI DOAN 1: Uu tien bao tri cac skill dang FAIL trong kho
        failing = [k for k, v in sorted(meta.items()) if v.get("health") in ("FAIL", "SYNTAX_ERROR")]
        if failing:
            print(f"\n[PHASE 1] Phat hien {len(failing)} skill dang FAIL: {failing}")
            for sname in failing:
                if time.perf_counter() - t_start >= budget_sec:
                    print("[BUDGET] Het ngan sach thoi gian trong giai doan bao tri.")
                    break
                await maintain_failed_skill(client, sname, meta, mem)
            meta = agent.sync_registry_meta()
            mem.sync_skill_vectors(meta)
        else:
            print("\n[PHASE 1] Toan bo kho skills/ dang xanh, bo qua bao tri.")

        if maintenance_only:
            agent.run_health_check()
            return

        # GIAI DOAN 2: Tu nghien cuu co dinh huong & tao tool moi
        print(f"\n[PHASE 2] Bat dau vong tu nghien cuu ({max_cycles} chu ky)...")
        for cycle in range(1, max_cycles + 1):
            elapsed = time.perf_counter() - t_start
            if elapsed >= budget_sec:
                print(f"\n[BUDGET] Da dung {elapsed/60:.1f}/{budget_minutes} phut -> Dung an toan.")
                break

            print(f"\n---------------- CHU KY NGHIEN CUU #{cycle}/{max_cycles} ----------------")
            meta = agent.sync_registry_meta()
            idea = await propose_next_research(client, meta, mem, DEFAULT_CURRICULUM)
            if not idea or not idea.get("question"):
                print("   -> Khong tao duoc y tuong hop le, bo qua chu ky nay.")
                continue

            topic = idea.get("topic", f"topic_{cycle}")
            q = idea["question"]
            print(f"   -> Chu de : {topic} (ky vong: {idea.get('expected_skill', 'N/A')})")
            print(f"   -> Ly do  : {idea.get('rationale', '')}")
            print(f"   -> Cau hoi: {q}\n")

            try:
                res = await agent.solve_question(q)
                ok = bool(res and res.get("success"))
                forged = ", ".join(res.get("forged", [])) if isinstance(res, dict) else ""
                mem.record_exploration(
                    topic=topic,
                    status="DONE" if ok else "FAILED",
                    skill_created=forged,
                    notes=f"Q: {q[:120]} | forged={forged} | elapsed={res.get('elapsed_sec', 0)}s",
                )
            except Exception as e:
                print(f"   -> [ERROR] Chu ky #{cycle} gap loi: {e}")
                mem.record_exploration(topic=topic, status="ERROR", notes=str(e)[:200])

        # GIAI DOAN 3: Tong hop de xuat nang cap file loi (chi xuat file, khong tu ghi de)
        if time.perf_counter() - t_start < budget_sec + 120:
            await generate_core_improvement_proposal(client, mem)

    print("\n============================================================")
    print(f"HOAN THANH AUTO-RESEARCHER trong {(time.perf_counter() - t_start)/60:.1f} phut.")
    print("============================================================")
    agent.run_health_check()
    agent.show_memory_stats()


def main():
    parser = argparse.ArgumentParser(description="AutoEvolve AI - Autonomous Maintenance & Research Loop")
    parser.add_argument("--max-cycles", type=int, default=2, help="So chu ky tu nghien cuu toi da")
    parser.add_argument("--budget-minutes", type=float, default=15.0, help="Gioi han thoi gian chay (phut)")
    parser.add_argument("--maintenance-only", action="store_true", help="Chi bao tri sua cac skill FAIL, khong tao tool moi")
    args = parser.parse_args()
    asyncio.run(
        run_autonomous_loop(
            max_cycles=args.max_cycles,
            budget_minutes=args.budget_minutes,
            maintenance_only=args.maintenance_only,
        )
    )


if __name__ == "__main__":
    main()
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/auto_researcher.py
"""Autonomous Maintenance + Directed Self-Evolution Loop for AutoEvolve AI."""
import os
import sys
import json
import time
import shutil
import sqlite3
import asyncio
import argparse
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
if BASE_DIR not in sys.path:
    sys.path.insert(0, BASE_DIR)

from memory_store import MemoryStore, DB_PATH
import live_evolve_agent as agent

PATCHES_DIR = f"{BASE_DIR}/proposed_patches"

DEFAULT_CURRICULUM = [
    "GitHub Public REST API (dung https://api.github.com/search/repositories de tim repo nhieu sao/trending khong can token)",
    "PyPI & Crates.io JSON API (dung https://pypi.org/pypi/<pkg>/json va https://crates.io/api/v1/crates/<pkg>)",
    "Hacker News Firebase API (dung https://hacker-news.firebaseio.com/v0/topstories.json va /v0/item/<id>.json)",
    "Hugging Face Public Hub API (dung https://huggingface.co/api/models de tim model moi/luot tai)",
    "World Bank Open Data API (dung https://api.worldbank.org/v2/country/.../indicator/...?format=json)",
    "OpenStreetMap Nominatim API (dung https://nominatim.openstreetmap.org/search?format=json)",
    "Network & SSL Inspector (kiem tra DNS, chung chi SSL/TLS, HTTP headers, thoi gian phan hoi cua 1 domain)",
]


def reindex_empty_vectors(meta: dict, mem: MemoryStore):
    with sqlite3.connect(DB_PATH) as conn:
        conn.execute("DELETE FROM skill_vectors WHERE embedding = '' OR embedding IS NULL")
    mem.sync_skill_vectors(meta)


async def maintain_failed_skill(client: httpx.AsyncClient, sname: str, meta: dict, mem: MemoryStore) -> bool:
    """Repair a failing skill or its out-of-sync unit test (up to 2 attempts) with backup & rollback."""
    spath = f"{agent.SKILLS_DIR}/{sname}.py"
    tpath = f"{agent.SKILLS_DIR}/test_{sname}.py"
    if not os.path.exists(spath) or not os.path.exists(tpath):
        return False

    def run_local_pytest() -> tuple[bool, str, str]:
        vdir = f"{agent.SANDBOX_DIR}/maint_{sname}"
        os.makedirs(vdir, exist_ok=True)
        try:
            shutil.copy2(spath, f"{vdir}/{sname}.py")
            shutil.copy2(tpath, f"{vdir}/test_{sname}.py")
            return agent.run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], vdir, 20)
        finally:
            shutil.rmtree(vdir, ignore_errors=True)

    ok, log, summary = run_local_pytest()
    if ok:
        meta[sname]["health"] = "PASS"
        agent.save_meta(meta)
        return True

    print(f"\n[MAINTENANCE] Dang sua `{sname}` (loi: {summary})")
    backup_skill_path = agent.backup_skill(sname)
    orig_test_src = open(tpath, encoding="utf-8").read()

    for attempt in range(1, 3):
        skill_src = open(spath, encoding="utf-8").read()
        test_src = open(tpath, encoding="utf-8").read()
        prompt = f"""Skill `{sname}.py` dang bi FAIL pytest (lan sua #{attempt}/2).
Luu y: Nhieu skill da duoc Self-Heal va chay thuc te tot, nhung file `test_{sname}.py` cu bi lech mock/assert.

--- `{sname}.py` ---
{agent.FENCE}python
{skill_src}
{agent.FENCE}

--- `test_{sname}.py` ---
{agent.FENCE}python
{test_src}
{agent.FENCE}

--- PYTEST LOG ---
{log[-2200:]}

Nhiem vu:
- Neu code `{sname}.py` da dung logic API that (vd: da raise RuntimeError khi loi/rong) va chi do `test_{sname}.py` bi lech mock/assert -> tra ve `TARGET: TEST` kem toan bo file `test_{sname}.py` da sua trong `{agent.FENCE}python ... {agent.FENCE}`.
- Neu code `{sname}.py` bi loi logic that -> tra ve `TARGET: CODE` kem toan bo file `{sname}.py` da sua trong `{agent.FENCE}python ... {agent.FENCE}` (KHONG import chinh `{sname}` trong file code).
- Khi mock mang trong test, dung `unittest.mock.patch("httpx.get")` va `with pytest.raises(Exception):` cho test loi."""

        raw = await agent.ask_llm(client, agent.BRAIN_THINK, "Ban la Staff Maintenance Engineer.", prompt, 0.15)
        is_test_fix = "TARGET: TEST" in raw.upper()
        fixed_code = agent.extract_code(raw, module_name=sname, is_test=is_test_fix)

        try:
            if is_test_fix:
                with open(tpath, "w", encoding="utf-8") as f:
                    f.write(fixed_code)
                print(f"   -> [Lan {attempt}] Cap nhat `test_{sname}.py` theo interface thuc te...")
            else:
                with open(spath, "w", encoding="utf-8") as f:
                    f.write(fixed_code)
                print(f"   -> [Lan {attempt}] Cap nhat code `{sname}.py`...")

            ok, log, summary = run_local_pytest()
            if ok:
                meta = agent.sync_registry_meta()
                meta[sname]["health"] = "PASS"
                meta[sname]["last_health_check"] = agent.now_vn_str()
                agent.save_meta(meta)
                mem.add_episode(
                    question=f"[MAINTENANCE] Fix {sname}",
                    skills_used=[sname],
                    outcome="PASS",
                    error_pattern=summary,
                    lesson=f"Da dong bo {'test' if is_test_fix else 'code'} cho `{sname}` thanh cong va pass pytest.",
                )
                print(f"   -> [MAINTENANCE OK] `{sname}` da chuyen sang PASS!")
                return True
            else:
                print(f"   -> [Lan {attempt} FAIL] {summary}")
        except Exception as e:
            print(f"   -> [Lan {attempt} ERROR] {e}")

    print(f"   -> [MAINTENANCE FAIL] Khong sua duoc sau 2 lan -> Rollback ve ban goc!")
    agent.restore_skill(sname, backup_skill_path)
    with open(tpath, "w", encoding="utf-8") as f:
        f.write(orig_test_src)
    return False


async def propose_next_research(
    client: httpx.AsyncClient,
    meta: dict,
    mem: MemoryStore,
    curriculum: list[str],
) -> dict | None:
    """Pick 1 missing high-value zero-key API capability and craft a real test question."""
    past = mem.get_explorations(limit=30)
    done_topics = [p["topic"] for p in past if p.get("status") == "DONE"]
    bl = mem.get_blacklist(limit=15)
    existing_skills = [f"- `{k}`: {v.get('description', '')}" for k, v in sorted(meta.items())]

    prompt = f"""Ban la AI Research Director cua he thong AutoEvolve AI.
Nhiem vu: Chon DUNG 1 nang luc API JSON con THIEU trong kho skill de tao tool chuyen biet moi.

KHO SKILL HIEN CO:
{chr(10).join(existing_skills)}

DANH SACH LINH VUC UU TIEN (Zero-Key JSON API):
{chr(10).join(f'- {c}' for c in curriculum)}

CAC CHU DE DA HOAN THANH DONE (KHONG chon lap lai):
{json.dumps(done_topics, ensure_ascii=False)}

DANH SACH DEN (KHONG dung):
{json.dumps([b['target'] for b in bl], ensure_ascii=False)}

Yeu cau:
- Chon 1 chu de CHUA CO tool chuyen biet trong kho skill va chua nam trong danh sach DONE.
- Dat 1 cau hoi thuc te dung JSON API mo (KHONG yeu cau cao trang HTML dong).

Tra ve DUY NHAT 1 JSON:
{{
  "topic": "ten ngan gon chu de (vd: github_public_api)",
  "expected_skill": "ten skill ky vong (vd: skill_github_api_client)",
  "api_endpoint_hint": "URL JSON API chinh thuc (vd: https://api.github.com/search/repositories)",
  "question": "cau hoi tieng Viet thuc te de giao cho live_evolve_agent chay va nghiem thu",
  "rationale": "ly do ngan gon tai sao tool nay huu ich"
}}"""
    raw = await agent.ask_llm(client, agent.BRAIN_THINK, "Ban la AI Research Director. Chi tra ve 1 JSON.", prompt, 0.4)
    try:
        return agent.extract_json(raw)
    except Exception as e:
        print(f"[EXPLORE] Khong parse duoc de xuat nghien cuu: {e}")
        return None


async def generate_core_improvement_proposal(client: httpx.AsyncClient, mem: MemoryStore):
    runs = []
    if os.path.exists(agent.RUNS_LOG_PATH):
        with open(agent.RUNS_LOG_PATH, encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if line:
                    try:
                        runs.append(json.loads(line))
                    except Exception:
                        pass

    recent_runs = runs[-15:]
    failed_runs = [r for r in recent_runs if not r.get("success") or r.get("forged_failed") or r.get("healed")]
    if not failed_runs:
        print("\n[SELF-IMPROVE] Cac phien gan day deu chay sach, chua can de xuat sua file loi.")
        return

    stats = mem.get_stats()
    prompt = f"""Phan tich lich su chay gan day cua AutoEvolve AI va de xuat cai tien cho file loi `live_evolve_agent.py` (KHONG sua truc tiep, chi viet bao cao de xuat cho lap trinh vien duyet).

Cac phien co loi / Self-Heal / Forge fail gan day:
{json.dumps(failed_runs, ensure_ascii=False, indent=2)}

5 bai hoc gan nhat trong memory.db:
{json.dumps(stats['recent_lessons'], ensure_ascii=False, indent=2)}

Viet bao cao Markdown ngan gon (tieng Viet) gom:
1. Diem nghen lap lai nhieu nhat
2. Nguyen nhan goc re
3. De xuat sua cu the trong `live_evolve_agent.py`"""

    report = await agent.ask_llm(client, agent.BRAIN_THINK, "Ban la Principal Systems Architect.", prompt, 0.2)
    os.makedirs(PATCHES_DIR, exist_ok=True)
    ts = time.strftime("%Y%m%d_%H%M%S")
    out_path = f"{PATCHES_DIR}/proposal_{ts}.md"
    with open(out_path, "w", encoding="utf-8") as f:
        f.write(f"# De xuat nang cap he thong ({agent.now_vn_str()})\n\n{report}\n")
    print(f"\n[SELF-IMPROVE] Da luu bao cao de xuat nang cap tai: {out_path}")


async def run_autonomous_loop(
    max_cycles: int = 2,
    budget_minutes: float = 15.0,
    maintenance_only: bool = False,
):
    t_start = time.perf_counter()
    budget_sec = budget_minutes * 60.0
    meta = agent.sync_registry_meta()
    mem = MemoryStore()
    reindex_empty_vectors(meta, mem)

    print("============================================================")
    print(f"AUTO-RESEARCHER KHOI DONG | max_cycles={max_cycles} | budget={budget_minutes} phut")
    print("============================================================")

    async with httpx.AsyncClient() as client:
        # GIAI DOAN 1: Uu tien bao tri cac skill dang FAIL trong kho
        failing = [k for k, v in sorted(meta.items()) if v.get("health") in ("FAIL", "SYNTAX_ERROR")]
        if failing:
            print(f"\n[PHASE 1] Phat hien {len(failing)} skill dang FAIL: {failing}")
            for sname in failing:
                if time.perf_counter() - t_start >= budget_sec:
                    print("[BUDGET] Het ngan sach thoi gian trong giai doan bao tri.")
                    break
                await maintain_failed_skill(client, sname, meta, mem)
            meta = agent.sync_registry_meta()
            mem.sync_skill_vectors(meta)
        else:
            print("\n[PHASE 1] Toan bo kho skills/ dang xanh, bo qua bao tri.")

        if maintenance_only:
            agent.run_health_check()
            return

        # GIAI DOAN 2: Tu nghien cuu co dinh huong & tao tool API chuyen biet moi
        print(f"\n[PHASE 2] Bat dau vong tu nghien cuu ({max_cycles} chu ky)...")
        for cycle in range(1, max_cycles + 1):
            elapsed = time.perf_counter() - t_start
            if elapsed >= budget_sec:
                print(f"\n[BUDGET] Da dung {elapsed/60:.1f}/{budget_minutes} phut -> Dung an toan.")
                break

            print(f"\n---------------- CHU KY NGHIEN CUU #{cycle}/{max_cycles} ----------------")
            meta = agent.sync_registry_meta()
            idea = await propose_next_research(client, meta, mem, DEFAULT_CURRICULUM)
            if not idea or not idea.get("question"):
                print("   -> Khong tao duoc y tuong hop le, bo qua chu ky nay.")
                continue

            topic = idea.get("topic", f"topic_{cycle}")
            expected_skill = idea.get("expected_skill", "")
            endpoint_hint = idea.get("api_endpoint_hint", "")
            q = idea["question"]
            print(f"   -> Chu de : {topic} (ky vong: {expected_skill})")
            print(f"   -> API    : {endpoint_hint}")
            print(f"   -> Cau hoi: {q}\n")

            forge_hint = (
                f"Hay thiet ke va tao moi skill `{expected_skill}` trong `missing_tools` "
                f"su dung truc tiep JSON API `{endpoint_hint}` (KHONG dung skill_web_reader de cao HTML)."
                if expected_skill and expected_skill not in meta
                else ""
            )

            try:
                res = await agent.solve_question(q, force_forge_hint=forge_hint)
                ok = bool(res and res.get("success"))
                forged = ", ".join(res.get("forged", [])) if isinstance(res, dict) else ""
                mem.record_exploration(
                    topic=topic,
                    status="DONE" if ok else "FAILED",
                    skill_created=forged,
                    notes=f"Q: {q[:120]} | forged={forged} | elapsed={res.get('elapsed_sec', 0)}s",
                )
            except Exception as e:
                print(f"   -> [ERROR] Chu ky #{cycle} gap loi: {e}")
                mem.record_exploration(topic=topic, status="ERROR", notes=str(e)[:200])

        # GIAI DOAN 3: Tong hop de xuat nang cap file loi
        if time.perf_counter() - t_start < budget_sec + 120:
            await generate_core_improvement_proposal(client, mem)

    print("\n============================================================")
    print(f"HOAN THANH AUTO-RESEARCHER trong {(time.perf_counter() - t_start)/60:.1f} phut.")
    print("============================================================")
    agent.run_health_check()
    agent.show_memory_stats()


def main():
    parser = argparse.ArgumentParser(description="AutoEvolve AI - Autonomous Maintenance & Research Loop")
    parser.add_argument("--max-cycles", type=int, default=2, help="So chu ky tu nghien cuu toi da")
    parser.add_argument("--budget-minutes", type=float, default=15.0, help="Gioi han thoi gian chay (phut)")
    parser.add_argument("--maintenance-only", action="store_true", help="Chi bao tri sua cac skill FAIL, khong tao tool moi")
    args = parser.parse_args()
    asyncio.run(
        run_autonomous_loop(
            max_cycles=args.max_cycles,
            budget_minutes=args.budget_minutes,
            maintenance_only=args.maintenance_only,
        )
    )


if __name__ == "__main__":
    main()
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/memory_store.py
"""Long-term SQLite + vector memory store for AutoEvolve AI."""
import os
import re
import math
import json
import sqlite3
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
DB_PATH = f"{BASE_DIR}/memory.db"
OLLAMA_BASE = "http://localhost:11434"
EMBED_MODEL = "nomic-embed-text"


def _now_vn() -> str:
    return datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")


def _tokenize(text: str) -> list[str]:
    return re.findall(r"[a-zA-Z0-9_]+|[\u00C0-\u1EF9]+", (text or "").lower())


def _cosine(v1: list[float], v2: list[float]) -> float:
    if not v1 or not v2 or len(v1) != len(v2):
        return 0.0
    dot = sum(a * b for a, b in zip(v1, v2))
    n1 = math.sqrt(sum(a * a for a in v1))
    n2 = math.sqrt(sum(b * b for b in v2))
    return dot / (n1 * n2) if n1 > 0 and n2 > 0 else 0.0


def _token_similarity(q: str, doc: str) -> float:
    t1 = set(_tokenize(q))
    t2 = set(_tokenize(doc))
    if not t1 or not t2:
        return 0.0
    inter = len(t1 & t2)
    return inter / math.sqrt(len(t1) * len(t2))


class MemoryStore:
    def __init__(self, db_path: str = DB_PATH, embed_model: str = EMBED_MODEL):
        self.db_path = db_path
        self.embed_model = embed_model
        self._embed_available: bool | None = None
        os.makedirs(os.path.dirname(db_path), exist_ok=True)
        self._init_db()

    def _conn(self) -> sqlite3.Connection:
        conn = sqlite3.connect(self.db_path)
        conn.row_factory = sqlite3.Row
        return conn

    def _init_db(self):
        with self._conn() as conn:
            conn.executescript(
                """
                CREATE TABLE IF NOT EXISTS episodes (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    timestamp TEXT NOT NULL,
                    question TEXT NOT NULL,
                    skills_used TEXT NOT NULL,
                    outcome TEXT NOT NULL,
                    error_pattern TEXT DEFAULT '',
                    lesson TEXT NOT NULL,
                    embedding TEXT DEFAULT ''
                );

                CREATE TABLE IF NOT EXISTS blacklist (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    target TEXT UNIQUE NOT NULL,
                    reason TEXT NOT NULL,
                    fail_count INTEGER DEFAULT 1,
                    updated_at TEXT NOT NULL
                );

                CREATE TABLE IF NOT EXISTS explorations (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    topic TEXT UNIQUE NOT NULL,
                    status TEXT NOT NULL,
                    skill_created TEXT DEFAULT '',
                    notes TEXT DEFAULT '',
                    updated_at TEXT NOT NULL
                );

                CREATE TABLE IF NOT EXISTS skill_vectors (
                    skill_name TEXT PRIMARY KEY,
                    description TEXT NOT NULL,
                    embedding TEXT DEFAULT '',
                    updated_at TEXT NOT NULL
                );
                """
            )

    def _embed(self, text: str) -> list[float]:
        if self._embed_available is False or not text.strip():
            return []
        snippet = text.strip()[:1500]
        try:
            r = httpx.post(
                f"{OLLAMA_BASE}/api/embed",
                json={"model": self.embed_model, "input": snippet},
                timeout=15.0,
            )
            if r.status_code == 200:
                data = r.json()
                embs = data.get("embeddings")
                if embs and isinstance(embs[0], list):
                    self._embed_available = True
                    return embs[0]
            r2 = httpx.post(
                f"{OLLAMA_BASE}/api/embeddings",
                json={"model": self.embed_model, "prompt": snippet},
                timeout=15.0,
            )
            if r2.status_code == 200 and "embedding" in r2.json():
                self._embed_available = True
                return r2.json()["embedding"]
            self._embed_available = False
        except Exception:
            self._embed_available = False
        return []

    def add_episode(
        self,
        question: str,
        skills_used: list[str],
        outcome: str,
        error_pattern: str,
        lesson: str,
    ):
        text_for_vec = f"{question} | {' '.join(skills_used)} | {error_pattern} | {lesson}"
        vec = self._embed(text_for_vec)
        with self._conn() as conn:
            conn.execute(
                """
                INSERT INTO episodes (timestamp, question, skills_used, outcome, error_pattern, lesson, embedding)
                VALUES (?, ?, ?, ?, ?, ?, ?)
                """,
                (
                    _now_vn(),
                    question,
                    json.dumps(skills_used, ensure_ascii=False),
                    outcome,
                    (error_pattern or "")[:400],
                    (lesson or "").strip()[:800],
                    json.dumps(vec) if vec else "",
                ),
            )

    def search_lessons(self, query: str, top_k: int = 3) -> list[dict]:
        q_vec = self._embed(query)
        with self._conn() as conn:
            rows = conn.execute(
                "SELECT id, timestamp, question, skills_used, outcome, error_pattern, lesson, embedding FROM episodes ORDER BY id DESC LIMIT 200"
            ).fetchall()

        scored = []
        for row in rows:
            if not row["lesson"]:
                continue
            doc_text = f"{row['question']} {row['skills_used']} {row['error_pattern']} {row['lesson']}"
            r_vec = json.loads(row["embedding"]) if row["embedding"] else []
            if q_vec and r_vec and len(q_vec) == len(r_vec):
                score = _cosine(q_vec, r_vec)
            else:
                score = _token_similarity(query, doc_text)
            if score >= 0.15:
                # Uu tien nhe cac bai hoc da kiem chung thanh cong (PASS)
                rank_score = score + (0.05 if row["outcome"] == "PASS" else 0.0)
                scored.append(
                    {
                        "score": round(score, 3),
                        "rank_score": rank_score,
                        "outcome": row["outcome"],
                        "lesson": row["lesson"],
                        "error_pattern": row["error_pattern"],
                        "skills_used": json.loads(row["skills_used"]),
                    }
                )
        scored.sort(key=lambda x: x["rank_score"], reverse=True)
        return scored[:top_k]

    def add_blacklist(self, target: str, reason: str):
        target = (target or "").strip().lower()[:200]
        if not target:
            return
        with self._conn() as conn:
            conn.execute(
                """
                INSERT INTO blacklist (target, reason, fail_count, updated_at)
                VALUES (?, ?, 1, ?)
                ON CONFLICT(target) DO UPDATE SET
                    reason = excluded.reason,
                    fail_count = blacklist.fail_count + 1,
                    updated_at = excluded.updated_at
                """,
                (target, (reason or "")[:300], _now_vn()),
            )

    def get_blacklist(self, limit: int = 15) -> list[dict]:
        with self._conn() as conn:
            rows = conn.execute(
                "SELECT target, reason, fail_count, updated_at FROM blacklist ORDER BY fail_count DESC, updated_at DESC LIMIT ?",
                (limit,),
            ).fetchall()
        return [dict(r) for r in rows]

    def record_exploration(self, topic: str, status: str, skill_created: str = "", notes: str = ""):
        topic = (topic or "").strip()[:200]
        if not topic:
            return
        with self._conn() as conn:
            conn.execute(
                """
                INSERT INTO explorations (topic, status, skill_created, notes, updated_at)
                VALUES (?, ?, ?, ?, ?)
                ON CONFLICT(topic) DO UPDATE SET
                    status = excluded.status,
                    skill_created = CASE WHEN excluded.skill_created != '' THEN excluded.skill_created ELSE explorations.skill_created END,
                    notes = excluded.notes,
                    updated_at = excluded.updated_at
                """,
                (topic, status, skill_created, (notes or "")[:400], _now_vn()),
            )

    def get_explorations(self, limit: int = 50) -> list[dict]:
        with self._conn() as conn:
            rows = conn.execute(
                "SELECT topic, status, skill_created, notes, updated_at FROM explorations ORDER BY updated_at DESC LIMIT ?",
                (limit,),
            ).fetchall()
        return [dict(r) for r in rows]

    def sync_skill_vectors(self, meta: dict):
        with self._conn() as conn:
            existing = {
                r["skill_name"]: r["description"]
                for r in conn.execute("SELECT skill_name, description FROM skill_vectors").fetchall()
            }
            for mod, info in meta.items():
                desc = f"{mod}: {info.get('description', '')} | {info.get('signatures', '')}"
                if existing.get(mod) != desc:
                    vec = self._embed(desc)
                    conn.execute(
                        """
                        INSERT INTO skill_vectors (skill_name, description, embedding, updated_at)
                        VALUES (?, ?, ?, ?)
                        ON CONFLICT(skill_name) DO UPDATE SET
                            description = excluded.description,
                            embedding = excluded.embedding,
                            updated_at = excluded.updated_at
                        """,
                        (mod, desc, json.dumps(vec) if vec else "", _now_vn()),
                    )
            for dead in set(existing.keys()) - set(meta.keys()):
                conn.execute("DELETE FROM skill_vectors WHERE skill_name = ?", (dead,))

    def search_skills(self, query: str, meta: dict, top_k: int = 12) -> list[str]:
        if len(meta) <= top_k:
            return sorted(meta.keys())
        q_vec = self._embed(query)
        with self._conn() as conn:
            rows = conn.execute("SELECT skill_name, description, embedding FROM skill_vectors").fetchall()
        scored = []
        for r in rows:
            if r["skill_name"] not in meta:
                continue
            r_vec = json.loads(r["embedding"]) if r["embedding"] else []
            if q_vec and r_vec and len(q_vec) == len(r_vec):
                score = _cosine(q_vec, r_vec)
            else:
                score = _token_similarity(query, r["description"])
            scored.append((score, r["skill_name"]))
        scored.sort(key=lambda x: x[0], reverse=True)
        return [name for _, name in scored[:top_k]]

    def format_context_for_prompt(self, query: str) -> str:
        lessons = self.search_lessons(query, top_k=3)
        bl = self.get_blacklist(limit=10)
        parts = []
        if lessons:
            parts.append("BAI HOC TU CAC PHIEN TRUOC (Bo nho dai han):")
            for idx, item in enumerate(lessons, 1):
                tag = (
                    "VERIFIED"
                    if item["outcome"] == "PASS"
                    else "UNVERIFIED - CANH BAO LOI CU, KHONG TIN TUYET DOI GIA THUYET CHUA KIEM CHUNG"
                )
                parts.append(f"  {idx}. [{tag}] {item['lesson']}")
        if bl:
            parts.append("NGUON/API TRONG DANH SACH DEN (KHONG DUNG):")
            for item in bl:
                parts.append(f"  - {item['target']} ({item['reason']})")
        return "\n".join(parts)

    def get_stats(self) -> dict:
        with self._conn() as conn:
            ep = conn.execute("SELECT COUNT(*) FROM episodes").fetchone()[0]
            bl = conn.execute("SELECT COUNT(*) FROM blacklist").fetchone()[0]
            ex = conn.execute("SELECT COUNT(*) FROM explorations").fetchone()[0]
            sv = conn.execute("SELECT COUNT(*) FROM skill_vectors").fetchone()[0]
            recent = [
                dict(r)
                for r in conn.execute(
                    "SELECT timestamp, outcome, lesson FROM episodes ORDER BY id DESC LIMIT 5"
                ).fetchall()
            ]
        return {
            "db_path": self.db_path,
            "embed_model": self.embed_model,
            "episodes": ep,
            "blacklist": bl,
            "explorations": ex,
            "skill_vectors": sv,
            "recent_lessons": recent,
        }
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py
import os
import re
import ast
import sys
import json
import time
import shutil
import asyncio
import argparse
import subprocess
from datetime import datetime, timezone, timedelta
import httpx

BASE_DIR = "/content/drive/MyDrive/AutoEvolve_AI"
SKILLS_DIR = f"{BASE_DIR}/skills"
for _p in (BASE_DIR, SKILLS_DIR):
    if _p not in sys.path:
        sys.path.insert(0, _p)

from memory_store import MemoryStore

SANDBOX_DIR = f"{BASE_DIR}/sandbox"
BACKUP_DIR = f"{BASE_DIR}/skill_backups"
META_PATH = f"{BASE_DIR}/registry_meta.json"
RUNS_LOG_PATH = f"{BASE_DIR}/runs.jsonl"

OLLAMA_URL = "http://localhost:11434/api/chat"
BRAIN_THINK = "qwq:32b"
BRAIN_CODE = "qwen2.5-coder:32b"
FENCE = "`" * 3
MAX_BACKUPS_PER_SKILL = 5
CORE_WEB_SKILLS = {"skill_web_search", "skill_web_reader"}


def now_vn_str() -> str:
    return datetime.now(timezone(timedelta(hours=7))).strftime("%Y-%m-%d %H:%M:%S (UTC+7)")


def strip_think(text: str) -> str:
    return re.sub(r"<think>.*?</think>", "", text or "", flags=re.DOTALL).strip()


def extract_code(text: str, module_name: str | None = None, is_test: bool = False) -> str:
    blocks = re.findall(FENCE + r"(?:python)?\s*(.*?)" + FENCE, text, re.DOTALL)
    if not blocks and FENCE in text:
        part = text.split(FENCE, 1)[1]
        part = re.sub(r"^python\s*", "", part, flags=re.IGNORECASE)
        blocks = [part]
    if not blocks:
        blocks = [text]

    valid = []
    for b in blocks:
        code = b.strip()
        if not code:
            continue
        if module_name and not is_test:
            if re.search(rf"\b(from|import)\s+{re.escape(module_name)}\b", code):
                continue
            if "def test_" in code and "class " not in code:
                continue
        try:
            ast.parse(code)
            valid.append(code)
        except SyntaxError:
            continue

    if valid:
        return max(valid, key=len)
    return blocks[0].strip()


def extract_json(text: str) -> dict:
    candidates = re.findall(FENCE + r"(?:json)?\s*(\{.*?\})\s*" + FENCE, text, re.DOTALL)
    candidates.append(text)
    decoder = json.JSONDecoder(strict=False)
    for chunk in candidates:
        pos = 0
        while True:
            idx = chunk.find("{", pos)
            if idx == -1:
                break
            try:
                obj, _ = decoder.raw_decode(chunk, idx)
                if isinstance(obj, dict):
                    return obj
            except json.JSONDecodeError:
                pass
            pos = idx + 1
    raise ValueError(f"Khong tim thay JSON hop le trong: {text[:200]}")


async def ask_llm(client: httpx.AsyncClient, model: str, sys_p: str, usr_p: str, temp: float = 0.2) -> str:
    payload = {
        "model": model,
        "stream": False,
        "keep_alive": "60m",
        "options": {"temperature": temp, "num_ctx": 12288},
        "messages": [
            {"role": "system", "content": sys_p},
            {"role": "user", "content": usr_p},
        ],
    }
    r = await client.post(OLLAMA_URL, json=payload, timeout=240.0)
    r.raise_for_status()
    return strip_think(r.json()["message"]["content"])


def inspect_skill_file(filepath: str) -> tuple[str, str]:
    try:
        src = open(filepath, encoding="utf-8").read()
        tree = ast.parse(src)
        doc = (ast.get_docstring(tree) or "").splitlines()
        desc = doc[0].strip()[:120] if doc else ""
        sigs = []
        for n in tree.body:
            if isinstance(n, ast.ClassDef):
                if not desc:
                    cdoc = (ast.get_docstring(n) or "").splitlines()
                    if cdoc:
                        desc = cdoc[0].strip()[:120]
                ms = []
                for m in n.body:
                    if isinstance(m, (ast.FunctionDef, ast.AsyncFunctionDef)) and (
                        not m.name.startswith("_") or m.name == "__init__"
                    ):
                        args_s = ", ".join(a.arg for a in m.args.args if a.arg != "self")
                        ret_s = f" -> {ast.unparse(m.returns)}" if m.returns else ""
                        mdoc = (ast.get_docstring(m) or "").splitlines()
                        doc_s = f" /* {mdoc[0].strip()[:80]} */" if mdoc else ""
                        ms.append(f"{m.name}({args_s}){ret_s}{doc_s}")
                sigs.append(f"class {n.name}: [{', '.join(ms)}]")
            elif isinstance(n, (ast.FunctionDef, ast.AsyncFunctionDef)) and not n.name.startswith("_"):
                args_s = ", ".join(a.arg for a in n.args.args)
                ret_s = f" -> {ast.unparse(n.returns)}" if n.returns else ""
                sigs.append(f"def {n.name}({args_s}){ret_s}")
        return desc or "Python skill module", " | ".join(sigs)
    except Exception as e:
        return f"Parse error: {e}", ""


def load_meta() -> dict:
    if os.path.exists(META_PATH):
        try:
            with open(META_PATH, "r", encoding="utf-8") as f:
                return json.load(f)
        except Exception:
            pass
    return {}


def save_meta(meta: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(META_PATH, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)


def sync_registry_meta() -> dict:
    os.makedirs(SKILLS_DIR, exist_ok=True)
    meta = load_meta()
    existing = set()
    for fn in sorted(os.listdir(SKILLS_DIR)):
        if not fn.endswith(".py") or fn.startswith("test_"):
            continue
        mod = fn[:-3]
        existing.add(mod)
        desc, sigs = inspect_skill_file(f"{SKILLS_DIR}/{fn}")
        entry = meta.get(mod, {})
        entry["signatures"] = sigs
        if not entry.get("description") or entry["description"] == "Python skill module":
            entry["description"] = desc
        entry.setdefault("success_count", 0)
        entry.setdefault("fail_count", 0)
        entry.setdefault("health", "UNKNOWN")
        entry.setdefault("updated_at", now_vn_str())
        meta[mod] = entry

    for dead in list(meta.keys()):
        if dead not in existing:
            del meta[dead]

    save_meta(meta)
    return meta


def format_registry_for_architect(meta: dict, subset: list[str] | None = None) -> str:
    if not meta:
        return "(Kho dang trong)"
    keys = sorted(subset) if subset else sorted(meta.keys())
    lines = []
    for mod in keys:
        if mod not in meta:
            continue
        info = meta[mod]
        stats = f"ok={info.get('success_count', 0)}, fail={info.get('fail_count', 0)}, health={info.get('health', 'UNKNOWN')}"
        lines.append(f"- `{mod}` ({stats}): {info.get('description', '')} | {info.get('signatures', '')}")
    return "\n".join(lines)


def format_registry_for_coder(meta: dict, selected: list[str] | None = None) -> str:
    if not meta:
        return "(Kho dang trong)"
    wanted = set()
    if selected:
        for s in selected:
            name = s.replace(".py", "").strip()
            if name in meta:
                wanted.add(name)
    for core in CORE_WEB_SKILLS:
        if core in meta:
            wanted.add(core)
    if not wanted:
        wanted = set(meta.keys())
    lines = [f"- `{m}`: {meta[m].get('signatures', '')}" for m in sorted(wanted)]
    return "\n".join(lines)


def update_skill_stats(meta: dict, skill_names: list[str], success: bool):
    ts = now_vn_str()
    for s in skill_names:
        name = s.replace(".py", "").strip()
        if name in meta:
            key = "success_count" if success else "fail_count"
            meta[name][key] = meta[name].get(key, 0) + 1
            meta[name]["last_used"] = ts
    save_meta(meta)


def append_run_log(record: dict):
    os.makedirs(BASE_DIR, exist_ok=True)
    with open(RUNS_LOG_PATH, "a", encoding="utf-8") as f:
        f.write(json.dumps(record, ensure_ascii=False) + "\n")


def cleanup_backups(max_per_skill: int = MAX_BACKUPS_PER_SKILL):
    if not os.path.exists(BACKUP_DIR):
        return
    groups: dict[str, list[str]] = {}
    for fn in os.listdir(BACKUP_DIR):
        if not fn.endswith(".py.bak"):
            continue
        m = re.match(r"^(skill_.+)_\d{8}_\d{6}\.py\.bak$", fn)
        key = m.group(1) if m else fn
        groups.setdefault(key, []).append(os.path.join(BACKUP_DIR, fn))
    for files in groups.values():
        files.sort(key=lambda p: os.path.getmtime(p), reverse=True)
        for old_path in files[max_per_skill:]:
            try:
                os.remove(old_path)
            except OSError:
                pass


def cleanup_sandbox():
    if not os.path.exists(SANDBOX_DIR):
        return
    for name in os.listdir(SANDBOX_DIR):
        if name.startswith(("forge_", "heal_verify", "health_")):
            full = os.path.join(SANDBOX_DIR, name)
            try:
                if os.path.isdir(full):
                    shutil.rmtree(full, ignore_errors=True)
                else:
                    os.remove(full)
            except OSError:
                pass


def run_cmd(cmd: list[str], cwd: str, timeout: int = 30) -> tuple[bool, str, str]:
    env = os.environ.copy()
    env["PYTHONPATH"] = f"{cwd}:{SKILLS_DIR}:{BASE_DIR}:" + env.get("PYTHONPATH", "")
    try:
        p = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
        out = (p.stdout + "\n" + p.stderr).strip()
        lines = out.splitlines()
        errs = [l for l in lines if "Error" in l or "FAILED" in l or "AssertionError" in l]
        summary = errs[-1][:160] if errs else (out.replace("\n", " | ")[:160] if out else "Output rong")
        return p.returncode == 0, out, summary
    except subprocess.TimeoutExpired:
        return False, f"Timeout ({timeout}s)", f"Timeout ({timeout}s)"


def backup_skill(sname: str) -> str | None:
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return None
    os.makedirs(BACKUP_DIR, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S")
    bpath = f"{BACKUP_DIR}/{sname}_{ts}.py.bak"
    shutil.copy2(spath, bpath)
    cleanup_backups()
    return bpath


def restore_skill(sname: str, bpath: str | None):
    spath = f"{SKILLS_DIR}/{sname}.py"
    if bpath and os.path.exists(bpath):
        shutil.copy2(bpath, spath)
        print(f"     -> [ROLLBACK] Da khoi phuc ban goc cua `{sname}.py` do ban va chua dat.")


def verify_skill_pytest(sname: str) -> tuple[bool, str]:
    tpath = f"{SKILLS_DIR}/test_{sname}.py"
    spath = f"{SKILLS_DIR}/{sname}.py"
    if not os.path.exists(spath):
        return False, "Khong tim thay file skill"
    if not os.path.exists(tpath):
        return True, "NO_TEST"
    vdir = f"{SANDBOX_DIR}/heal_verify_{sname}"
    os.makedirs(vdir, exist_ok=True)
    try:
        shutil.copy2(tpath, f"{vdir}/test_{sname}.py")
        shutil.copy2(spath, f"{vdir}/{sname}.py")
        ok, _, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], vdir, 20)
        return ok, ("PASS" if ok else summary)
    finally:
        shutil.rmtree(vdir, ignore_errors=True)


def lookup_api_docs_on_http_error(out: str, skills: list[str]) -> str:
    """When HTTP 400/404/422 or invalid param occurs, search real API docs via WebSearchClient."""
    if not re.search(r"\b(400|404|422)\b|Invalid\s+\w+\s+parameter|Bad Request", out, re.IGNORECASE):
        return ""
    try:
        from skill_web_search import WebSearchClient
    except Exception:
        return ""

    # Trich xuat ten dich vu tu skill hoac URL trong log
    target = ""
    url_m = re.search(r"https?://([a-zA-Z0-9.-]+/[a-zA-Z0-9._/-]{0,40})", out)
    if url_m:
        target = url_m.group(1).replace("/", " ")
    elif skills:
        clean_names = [
            re.sub(r"^skill_|_client$|_api$", "", s).replace("_", " ")
            for s in skills
            if s not in CORE_WEB_SKILLS
        ]
        target = " ".join(clean_names[:2])

    err_m = re.search(r"(Invalid[^\"\n}]{1,60}|400[^\"\n}]{0,60}|422[^\"\n}]{0,60}|404[^\"\n}]{0,60})", out, re.IGNORECASE)
    err_part = err_m.group(1).strip() if err_m else "API parameters"
    query = re.sub(r"\s+", " ", f"{target} API {err_part} documentation").strip()[:120]
    if not query:
        return ""

    print(f"   -> [DOC LOOKUP] Phat hien loi HTTP 4xx -> Tra tai lieu API: `{query}`")
    try:
        results = WebSearchClient(timeout=12.0).search(query, max_results=3, lang="en")
        lines = ["TAI LIEU API VUA TRA CUU TU WEB (dung de sua dung tham so/endpoint):"]
        for r in results[:3]:
            lines.append(f"  - {r.get('title', '')}: {r.get('snippet', '')[:260]} ({r.get('url', '')})")
        return "\n".join(lines)
    except Exception as e:
        print(f"   -> [DOC LOOKUP SKIP] Khong tra duoc docs: {e}")
        return ""


def run_health_check():
    print("=== HEALTH CHECK KHO SKILLS ===")
    meta = sync_registry_meta()
    mem = MemoryStore()
    mem.sync_skill_vectors(meta)
    if not meta:
        print("Kho skills/ dang trong.")
        return
    passed = failed = no_test = 0
    for mod in sorted(meta.keys()):
        spath = f"{SKILLS_DIR}/{mod}.py"
        try:
            ast.parse(open(spath, encoding="utf-8").read())
        except SyntaxError as e:
            meta[mod]["health"] = "SYNTAX_ERROR"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> SyntaxError: {e}")
            continue

        ok, msg = verify_skill_pytest(mod)
        if msg == "NO_TEST":
            meta[mod]["health"] = "NO_TEST"
            no_test += 1
            print(f"  [WARN] {mod:<35} -> Khong co file test")
        elif ok:
            meta[mod]["health"] = "PASS"
            passed += 1
            print(f"  [PASS] {mod:<35} -> pytest OK")
        else:
            meta[mod]["health"] = "FAIL"
            failed += 1
            print(f"  [FAIL] {mod:<35} -> {msg}")
        meta[mod]["last_health_check"] = now_vn_str()

    save_meta(meta)
    cleanup_backups()
    cleanup_sandbox()
    print(f"\nTong ket: {passed} PASS | {failed} FAIL | {no_test} NO_TEST (da cap nhat registry_meta.json, memory.db & don rac).")


def show_memory_stats():
    mem = MemoryStore()
    stats = mem.get_stats()
    print("=== TRANG THAI BO NHO DAI HAN (memory.db) ===")
    print(f"  File DB        : {stats['db_path']}")
    print(f"  Embed Model    : {stats['embed_model']}")
    print(f"  Episodes       : {stats['episodes']} bai hoc")
    print(f"  Blacklist      : {stats['blacklist']} nguon bi chan")
    print(f"  Explorations   : {stats['explorations']} chu de da nghien cuu")
    print(f"  Skill Vectors  : {stats['skill_vectors']} skill da lap chi muc")
    if stats["recent_lessons"]:
        print("\n5 bai hoc gan nhat:")
        for r in stats["recent_lessons"]:
            print(f"  - [{r['timestamp']}] ({r['outcome']}) {r['lesson']}")
    bl = mem.get_blacklist(10)
    if bl:
        print("\nDanh sach den:")
        for b in bl:
            print(f"  - {b['target']} (fail={b['fail_count']}): {b['reason']}")


async def review_code(client: httpx.AsyncClient, sname: str, goal: str, stub: str, code: str) -> tuple[bool, str]:
    rev_p = f"""Kiem duyet `{sname}.py` (Muc tieu: {goal}):
--- STUB ---
{FENCE}python
{stub}
{FENCE}
--- CODE ---
{FENCE}python
{code}
{FENCE}
Kiem tra:
1. Co goi API/nguon that khong can API key khong (tuyet doi khong hardcode du lieu gia)?
2. Co bam dung ten class/ham trong Stub khong (cho phep them tham so tuy chon co gia tri mac dinh)?
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
REASON: <ngan gon>"""
    rev = await ask_llm(client, BRAIN_THINK, "Ban la Code Reviewer nghiem khac.", rev_p, 0.1)
    ok = "STATUS: PASS" in rev.upper()
    m = re.search(r"REASON:\s*(.+)", rev)
    return ok, (m.group(1).strip() if m else rev.splitlines()[-1][:100])


async def verify_collected_data(client: httpx.AsyncClient, question: str, data: str) -> tuple[bool, str]:
    ver_p = f"""Cau hoi nguoi dung: {question}

Du lieu script vua thu thap duoc:
{data[:8000]}

Kiem tra:
- Du lieu tren da cung cap du thong tin thuc te de tra loi cau hoi chua?
- LUU Y: Neu cau hoi dung tu "hoac" / "or" (vi du "A hoac B"), chi can co du lieu tot cho A HOAC B la DAT (PASS), KHONG bat buoc phai co ca hai.
- Chi danh FAIL neu du lieu bi loi/Traceback/rong/N/A/example.com/404, hoac thieu mot ve bat buoc ("va").
Tra loi DUNG 2 dong:
STATUS: PASS (hoac FAIL)
MISSING: <neu FAIL, ghi ro phan nao bi thieu hoac rong; neu PASS ghi None>"""
    res = await ask_llm(client, BRAIN_THINK, "Ban la QA Data Verifier chinh xac.", ver_p, 0.1)
    ok = "STATUS: PASS" in res.upper()
    m = re.search(r"MISSING:\s*(.+)", res)
    return ok, (m.group(1).strip() if m else res.splitlines()[-1][:120])


async def distill_lesson(
    client: httpx.AsyncClient,
    question: str,
    skills: list[str],
    errors: list[str],
    healed: list[str],
    success: bool,
) -> tuple[str, list[tuple[str, str]]]:
    """Return (concise_lesson, [(blocked_target, reason), ...]). Prevents memory poisoning on FAIL runs."""
    rule_anti_poison = (
        "PHIEN NAY THANH CONG (PASS): Hay ghi ro cach goi/tham so/endpoint nao da giup chay thanh cong."
        if success
        else "PHIEN NAY THAT BAI (FAIL): CHI ghi ro cach lam nao da THAT BAI va ma loi cu the la gi. TUYET DOI KHONG tu doan mo gia tri tham so sua loi chua duoc kiem chung (tranh nhiem doc bo nho)."
    )
    prompt = f"""Tom tat 1 bai hoc ky thuat ngan gon (1-2 cau):
- Cau hoi: {question}
- Skills da dung: {skills}
- Ket qua: {'PASS' if success else 'FAIL'}
- Skills da Self-Heal: {healed}
- Cac loi gap phai: {errors[-3:]}

QUY TAC BAT BUOC:
{rule_anti_poison}

Tra ve DUY NHAT JSON:
{{
  "lesson": "1-2 cau bai hoc trung thuc, khong suy dien",
  "blacklist": [
    {{"target": "ten domain hoac endpoint bi 403/404/429/API key (chi them neu that su bi chan)", "reason": "ly do ngan gon"}}
  ]
}}"""
    try:
        raw = await ask_llm(client, BRAIN_THINK, "Ban la ky su rut kinh nghiem he thong. Chi tra ve JSON.", prompt, 0.1)
        data = extract_json(raw)
        lesson = str(data.get("lesson", "")).strip()
        bl = [
            (str(b.get("target", "")).strip(), str(b.get("reason", "")).strip())
            for b in (data.get("blacklist") or [])
            if isinstance(b, dict) and b.get("target")
        ]
        return lesson, bl
    except Exception:
        fallback = f"Dung {skills} cho tac vu tuong tu; loi tung gap: {errors[-1][:140] if errors else 'khong'}"
        return fallback, []


async def forge_tool(client: httpx.AsyncClient, spec: dict, meta: dict) -> bool:
    sname = spec["skill_name"]
    if not sname.startswith("skill_"):
        sname = "skill_" + sname.lstrip("_")
    stub = spec["interface_stub"]
    goal = spec["goal"]
    print(f"\n  [FORGE] Dang tu che tao tool moi: `{sname}` -> {goal}")

    test_p = f"""Viet `test_{sname}.py` dung `pytest` (import tu `{sname}`) theo Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- 3 test case ro rang (1 happy path + 1 loi HTTP status != 200 + 1 loi du lieu rong/mang).
- Code thuc thi dung `httpx.get(...)`, nen mock bang `unittest.mock.patch("httpx.get")` (KHONG mock `requests`).
- Khi test truong hop loi, LUON dung `with pytest.raises(Exception):`.
- Chi test interface public trong Stub, khong assert qua chat vao chuoi text cu the.
Chi tra ve duy nhat 1 block code Python cho `test_{sname}.py`."""

    code_p = f"""Hoan thien `{sname}.py` theo DUNG Stub:
{FENCE}python
{stub}
{FENCE}
QUY TAC BAT BUOC:
- LUON dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})` (KHONG dung `requests`).
- KHONG nuot loi mang im lang (neu HTTP status != 200 hoac du lieu loi, raise `RuntimeError` kem status_code va noi dung loi).
- Chi dung API MO, MIEN PHI, KHONG CAN API KEY (Open-Meteo, MediaWiki/Wikipedia, ArXiv, OpenAlex, Google News RSS, Frankfurter, Binance public, GitHub REST API `https://api.github.com/search/repositories`, PyPI `https://pypi.org/pypi/<pkg>/json`, Crates.io `https://crates.io/api/v1/crates/<pkg>`, Hacker News `https://hacker-news.firebaseio.com/v0`, Hugging Face `https://huggingface.co/api/models` [direction=-1 cho giam dan], World Bank `https://api.worldbank.org/v2`).
- Tuyet doi khong hardcode du lieu gia, khong viet code test trong file nay.
Chi tra ve duy nhat 1 block code Python cho `{sname}.py`."""

    res = await asyncio.gather(
        ask_llm(client, BRAIN_CODE, "Ban la QA Engineer. Chi tra ve 1 block code Python.", test_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.2),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.5),
        ask_llm(client, BRAIN_CODE, "Ban la Senior Python Dev. Chi tra ve 1 block code Python.", code_p, 0.8),
    )
    test_code = extract_code(res[0], module_name=sname, is_test=True)
    candidates = [extract_code(c, module_name=sname, is_test=False) for c in res[1:]]

    last_code0, last_log0 = candidates[0], ""

    def save_forged(code_str: str, test_str: str):
        with open(f"{SKILLS_DIR}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(code_str)
        with open(f"{SKILLS_DIR}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_str)
        _, sigs = inspect_skill_file(f"{SKILLS_DIR}/{sname}.py")
        meta[sname] = {
            "description": goal[:120],
            "signatures": sigs,
            "success_count": 0,
            "fail_count": 0,
            "health": "PASS",
            "updated_at": now_vn_str(),
        }
        save_meta(meta)

    for idx, code in enumerate(candidates):
        cdir = f"{SANDBOX_DIR}/forge_{idx}"
        os.makedirs(cdir, exist_ok=True)
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(test_code)

        curr = code
        for attempt in range(1, 3):
            with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
                f.write(curr)
            ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
            if idx == 0:
                last_code0, last_log0 = curr, log

            if ok:
                rev_ok, rev_reason = await review_code(client, sname, goal, stub, curr)
                if rev_ok:
                    save_forged(curr, test_code)
                    print(f"  [FORGE OK] Luu `{sname}.py` vao skills/ (Nhanh #{idx+1}, lan {attempt})")
                    return True
                else:
                    print(f"     -> Nhanh #{idx+1} pass pytest nhung REVIEW FAIL: {rev_reason}")
                    break
            else:
                print(f"     -> Nhanh #{idx+1} FAIL lan {attempt}: {summary}")
                if attempt < 2:
                    fix_p = f"""Sua file `{sname}.py` de pass pytest (van dung `httpx.get(..., follow_redirects=True)` va API that).
CHI tra ve code cua `{sname}.py`, KHONG tra ve file test.
Code:
{FENCE}python
{curr}
{FENCE}
Test:
{FENCE}python
{test_code}
{FENCE}
Log:
{log[-1500:]}"""
                    curr = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code cho file skill.", fix_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )

    print(f"     -> Ca 3 nhanh FAIL -> Trong tai ({BRAIN_THINK}) kiem tra TEST_BUG...")
    adj_p = f"""Ca 3 nhanh code cua `{sname}.py` deu fail pytest.
Neu file test viet sai (mock sai duong dan httpx, ep sai loai exception, sai cau truc response):
Tra ve `VERDICT: TEST_BUG` kem toan bo file `test_{sname}.py` da sua trong `{FENCE}python ... {FENCE}`.
Neu loi do code: tra ve `VERDICT: CODE_BUG`.

--- CODE ---
{FENCE}python
{last_code0}
{FENCE}
--- TEST ---
{FENCE}python
{test_code}
{FENCE}
--- LOG ---
{last_log0[-1500:]}"""
    adj = await ask_llm(client, BRAIN_THINK, "Ban la Staff QA Adjudicator.", adj_p, 0.2)
    if "VERDICT: TEST_BUG" in adj.upper() and FENCE in adj:
        fixed_test = extract_code(adj, module_name=sname, is_test=True)
        cdir = f"{SANDBOX_DIR}/forge_0"
        with open(f"{cdir}/test_{sname}.py", "w", encoding="utf-8") as f:
            f.write(fixed_test)
        with open(f"{cdir}/{sname}.py", "w", encoding="utf-8") as f:
            f.write(last_code0)
        ok, log, summary = run_cmd(["pytest", f"test_{sname}.py", "-q", "--tb=short"], cdir, 20)
        if ok:
            rev_ok, rev_reason = await review_code(client, sname, goal, stub, last_code0)
            if rev_ok:
                save_forged(last_code0, fixed_test)
                print(f"  [FORGE OK - ADJUDICATED] Trong tai sua Test thanh cong -> Luu `{sname}.py`!")
                return True
        else:
            print(f"     -> Sau khi Trong tai sua Test van FAIL: {summary}")

    print(f"  [FORGE FAIL] Khong tao duoc `{sname}`.")
    return False


async def solve_question(
    question: str,
    dry_run: bool = False,
    no_forge: bool = False,
    force_forge_hint: str = "",
) -> dict:
    t0 = time.perf_counter()
    os.makedirs(SANDBOX_DIR, exist_ok=True)
    cleanup_sandbox()
    cleanup_backups()
    vn_now = now_vn_str()
    meta = sync_registry_meta()
    mem = MemoryStore()
    mem.sync_skill_vectors(meta)

    memory_ctx = mem.format_context_for_prompt(question)
    forged_ok: list[str] = []
    failed_tools: list[str] = []
    healed_skills: list[str] = []
    session_errors: list[str] = []
    run_success = False
    attempts_used = 0

    async with httpx.AsyncClient() as client:
        # BUOC 1: Architect (qwq:32b) phan tich nhu cau + doc bo nho dai han
        if len(meta) > 25:
            top_skills = list(set(mem.search_skills(question, meta, top_k=15)) | (CORE_WEB_SKILLS & set(meta.keys())))
            reg_arch = format_registry_for_architect(meta, subset=top_skills)
        else:
            reg_arch = format_registry_for_architect(meta)

        mem_block = f"\n{memory_ctx}\n" if memory_ctx else ""
        hint_line = f"\n   - YEU CAU BAT BUOC TU RESEARCH DIRECTOR: {force_forge_hint}" if force_forge_hint else ""
        plan_p = f"""CAU HOI NGUOI DUNG: {question}
THOI GIAN HIEN TAI: {vn_now}
{mem_block}
KHO SKILL HIEN CO (tu registry_meta.json):
{reg_arch}

Nhiem vu:
1. Kiem tra kho skill hien co da du de lay du lieu thuc te tra loi cau hoi chua.
   - UU TIEN DUNG JSON API CHUYEN BIET (GitHub API `api.github.com`, PyPI `pypi.org/pypi/<pkg>/json`, Crates.io `crates.io/api/v1/crates/<pkg>`, Hacker News `hacker-news.firebaseio.com/v0`, Hugging Face `huggingface.co/api`, World Bank `api.worldbank.org/v2`).
   - KHONG dung `skill_web_reader` de cao HTML tu trang dong/SPA (nhu github.com/trending, crates.io/crates, pypi.org/project) khi dich vu do co JSON API mo -> hay tao tool API chuyen biet trong `missing_tools`.{hint_line}
2. Neu THIEU tool chuyen biet, thiet ke 1-2 tool moi trong `missing_tools`.
   QUY TAC BAT BUOC:
   - CAM dung dich vu doi API key (KHONG dung OpenWeatherMap, NewsAPI, Google Search API...) hoac nguon trong DANH SACH DEN.
3. Neu DU roi thi de `missing_tools`: [].

Tra ve DUY NHAT 1 JSON hop le, khong kem giai thich ben ngoai:
{{
  "missing_tools": [
    {{
      "skill_name": "skill_...",
      "goal": "mo ta ngan gon, ghi ro nguon API khong can key",
      "interface_stub": "class ...:\\n    def ...(self, ...):\\n        pass"
    }}
  ],
  "use_skills": ["ten_cac_skill_se_dung"],
  "execution_notes": "huong dan ngan gon cho Coder cach goi tool lay du lieu"
}}"""
        print(f"=== CAU HOI: {question} ===")
        if memory_ctx:
            print("[MEM] Da nap kinh nghiem & danh sach den tu memory.db")
        print(f"[1/4] Architect ({BRAIN_THINK}) dang phan tich nhu cau & kiem tra registry_meta.json...")
        raw_plan = await ask_llm(client, BRAIN_THINK, "Ban la Principal AI Architect. Chi tra ve 1 JSON hop le.", plan_p, 0.3)
        plan = extract_json(raw_plan)
        missing = plan.get("missing_tools", [])
        use_skills = plan.get("use_skills", [])
        print(f"   -> Dung skills: {use_skills} | Can tao moi {len(missing)} tool: {[m['skill_name'] for m in missing]}")

        if dry_run:
            print("\n[DRY-RUN] Ke hoach Architect:")
            print(json.dumps(plan, ensure_ascii=False, indent=2))
            return {"success": True, "dry_run": True, "plan": plan}

        # BUOC 2: Forge tu che tao tool con thieu
        if not no_forge:
            for spec in missing:
                ok_forge = await forge_tool(client, spec, meta)
                if ok_forge:
                    sname = spec["skill_name"]
                    if not sname.startswith("skill_"):
                        sname = "skill_" + sname.lstrip("_")
                    forged_ok.append(sname)
                else:
                    failed_tools.append(spec["skill_name"])
        elif missing:
            print(f"   -> [--no-forge] Bo qua tao moi: {[m['skill_name'] for m in missing]}")
            failed_tools = [m["skill_name"] for m in missing]

        # BUOC 3: Coder chay thuc chien + Data Quality Gate + Doc Lookup + Self-Healing + Rollback
        meta = sync_registry_meta()
        mem.sync_skill_vectors(meta)
        selected_for_coder = list(dict.fromkeys(use_skills + forged_ok))
        reg_coder = format_registry_for_coder(meta, selected_for_coder)
        runner_path = f"{SANDBOX_DIR}/live_runner.py"
        print(f"\n[2/4] Coder ({BRAIN_CODE}) dang viet & chay script thu thap du lieu that...")

        failed_note = ""
        if failed_tools:
            failed_note = f"\nLUU Y QUAN TRONG: Cac tool sau CHUA DUOC TAO nen KHONG import: {failed_tools}. Hay dung `skill_web_search` / `skill_web_reader` hoac goi truc tiep API mo bang `httpx.get(..., follow_redirects=True)`."

        run_p = f"""Viet script Python thu thap du lieu thuc te cho cau hoi: "{question}"
{mem_block}
Cac skill duoc phep import (chu y ky kieu tra ve `-> dict`, `-> list[dict]` va docstring key):
{reg_coder}
{failed_note}

Ghi chu tu Architect: {plan.get('execution_notes', '')}
Yeu cau BAT BUOC:
- Lay day du du lieu cho TAT CA cac ve cua cau hoi.
- TUYET DOI KHONG dung URL gia nhu `example.com`; neu `skill_web_search` tra ve link `news.google.com/rss/articles/` (readable=False) thi CHI dung title/snippet, KHONG truyen link do vao `WebReaderClient.read_url()`.
- KHONG nuot loi im lang thanh 'N/A': in thang ket qua tho hoac traceback loi ra stdout/stderr.
- Cat gon moi doan tom tat toi da ~500 ky tu de in day du TAT CA cac muc va link nguon trong ~6000 ky tu.
- Chi tra ve 1 block code Python."""
        runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Ban la Python Dev. Chi tra ve 1 block code Python.", run_p, 0.2))

        collected_data = ""
        best_data = ""
        api_doc_hint = ""
        doc_lookup_done = False
        heal_counts: dict[str, int] = {}
        max_attempts = 5
        for attempt in range(1, max_attempts + 1):
            attempts_used = attempt
            with open(runner_path, "w", encoding="utf-8") as f:
                f.write(runner_code)
            ok, out, summary = run_cmd(["python3", runner_path], SANDBOX_DIR, 45)

            if ok and len(out.strip()) > len(best_data) and "Traceback (most recent call last)" not in out:
                best_data = out[:8000]

            fail_reason = summary
            if ok and len(out.strip()) > 80:
                gate_ok, missing_info = await verify_collected_data(client, question, out)
                if gate_ok:
                    collected_data = out[:8000]
                    run_success = True
                    print(f"   -> [OK lan {attempt}] Thu thap DU du lieu ({len(collected_data)} ky tu)!")
                    print(f"   -> [PREVIEW] {collected_data[:220].replace(chr(10), ' | ')}...")
                    break
                else:
                    fail_reason = f"DATA_INCOMPLETE: {missing_info}"
                    session_errors.append(fail_reason)
                    print(f"   -> [THIEU DU LIEU lan {attempt}] {missing_info}")
                    print(f"      [RAW OUT] {out[:200].replace(chr(10), ' | ')}")
            else:
                session_errors.append(summary)
                print(f"   -> [FAIL lan {attempt}] {summary}")

            if attempt == max_attempts:
                collected_data = best_data if best_data else out[-4000:]
                break

            # Tu dong tra tai lieu API khi gap HTTP 400 / 404 / 422
            if not doc_lookup_done:
                found_docs = lookup_api_docs_on_http_error(out + "\n" + fail_reason, selected_for_coder)
                if found_docs:
                    api_doc_hint = found_docs
                    doc_lookup_done = True

            doc_block = f"\n{api_doc_hint}\n" if api_doc_hint else ""
            diag_p = f"""Script thu thap du lieu bi loi hoac thieu du lieu khi chay thuc te.
Van de: {fail_reason}
{mem_block}{doc_block}
Runner code:
{FENCE}python
{runner_code}
{FENCE}
Output / Log:
{out[-2000:]}

Hay chan doan:
- Neu loi nam trong 1 file `skill_*.py` trong `skills/`: tra ve JSON {{"fix_target": "skill", "skill_name": "skill_...", "reason": "..."}}
- Neu loi do `live_runner.py` (goi sai ham, truyen sai gia tri tham so API, doc sai kieu dict/str, dung URL gia): tra ve JSON {{"fix_target": "runner", "reason": "..."}}"""
            try:
                diag = extract_json(await ask_llm(client, BRAIN_THINK, "Ban la Debugger. Chi tra ve JSON.", diag_p, 0.1))
            except Exception:
                diag = {"fix_target": "runner", "reason": fail_reason}

            if diag.get("fix_target") == "skill" and diag.get("skill_name"):
                sname = diag["skill_name"].replace(".py", "")
                spath = f"{SKILLS_DIR}/{sname}.py"
                if os.path.exists(spath) and heal_counts.get(sname, 0) < 2:
                    heal_counts[sname] = heal_counts.get(sname, 0) + 1
                    print(f"   -> [SELF-HEAL #{heal_counts[sname]}] Va `{sname}.py` ({diag.get('reason')})...")
                    bpath = backup_skill(sname)
                    old_src = open(spath, encoding="utf-8").read()
                    patch_p = f"""Sua file `{sname}.py` de khac phuc van de chay thuc te: {diag.get('reason')}
{mem_block}{doc_block}
Code hien tai cua `{sname}.py`:
{FENCE}python
{old_src}
{FENCE}

Cach `live_runner.py` dang goi va doc ket qua tu tool nay:
{FENCE}python
{runner_code}
{FENCE}

Output / Log thuc te:
{out[-1500:]}

QUY TAC SUA BAT BUOC:
1. Dung `httpx.get(..., timeout=15.0, follow_redirects=True, headers={{"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"}})`.
2. TUYET DOI KHONG nuot loi im lang tra ve rong/None; neu API loi phai raise RuntimeError kem status_code va response text.
3. Giu nguyen ten class/ham public va kieu tra ve de pass pytest.
Chi tra ve 1 block code Python cho `{sname}.py`."""
                    patched = extract_code(
                        await ask_llm(client, BRAIN_CODE, "Sua loi tool Python. Chi tra ve 1 block code.", patch_p, 0.2),
                        module_name=sname,
                        is_test=False,
                    )
                    try:
                        ast.parse(patched)
                        with open(spath, "w", encoding="utf-8") as f:
                            f.write(patched)
                        pt_ok, pt_msg = verify_skill_pytest(sname)
                        if not pt_ok:
                            print(f"     -> Ban va `{sname}.py` khong qua pytest ({pt_msg}).")
                            restore_skill(sname, bpath)
                            diag["reason"] = f"Skill `{sname}` giu nguyen interface (ban va bi rollback). Hay sua `live_runner.py` de dung dung kieu tra ve va tham so cua skill."
                        else:
                            healed_skills.append(sname)
                            meta = sync_registry_meta()
                            meta[sname]["health"] = "PASS" if pt_msg == "PASS" else "NO_TEST"
                            save_meta(meta)
                            continue
                    except SyntaxError:
                        print(f"     -> Ban va `{sname}.py` loi cu phap.")
                        restore_skill(sname, bpath)
                        diag["reason"] = f"Ban va `{sname}` loi cu phap va da rollback. Hay sua `live_runner.py`."

            fix_run_p = f"""Sua `live_runner.py` de lay day du du lieu that va in ra stdout (KHONG dung example.com, KHONG truyen link `news.google.com/rss/articles/` vao WebReaderClient, doc dung kieu dict/list tra ve cua skill).
Van de: {diag.get('reason')}
{mem_block}{doc_block}
Cac skill duoc phep import:
{reg_coder}
Code cu:
{FENCE}python
{runner_code}
{FENCE}
Output/Log:
{out[-1500:]}"""
            runner_code = extract_code(await ask_llm(client, BRAIN_CODE, "Sua loi Python. Chi tra ve 1 block code.", fix_run_p, 0.2))

        update_skill_stats(meta, selected_for_coder, run_success)

        # Luu bai hoc vao bo nho dai han
        if session_errors or healed_skills or not run_success:
            lesson, bl_items = await distill_lesson(
                client, question, selected_for_coder, session_errors, healed_skills, run_success
            )
            mem.add_episode(
                question=question,
                skills_used=selected_for_coder,
                outcome="PASS" if run_success else "FAIL",
                error_pattern=" | ".join(session_errors[-2:]),
                lesson=lesson,
            )
            for target, reason in bl_items:
                mem.add_blacklist(target, reason)
            print(f"   -> [MEM SAVE] Da luu bai hoc vao memory.db: {lesson[:140]}")
        else:
            mem.add_episode(
                question=question,
                skills_used=selected_for_coder,
                outcome="PASS",
                error_pattern="",
                lesson=f"Thanh cong ngay luot 1 bang {selected_for_coder} cho dang cau hoi tuong tu.",
            )

        # BUOC 4: Analyst (qwq:32b) tong hop bao cao
        print(f"\n[3/4] Analyst ({BRAIN_THINK}) dang tong hop cau tra loi tu du lieu that...")
        ans_p = f"""THOI GIAN HIEN TAI CUA HE THONG: {vn_now} (nam 2026 la thoi diem hien tai that, KHONG phai tuong lai hay gia lap).
Cau hoi: {question}

Du lieu thu thap duoc tu cac tool:
{collected_data}

Yeu cau BAT BUOC:
- Tra loi chinh xac, ro rang bang tieng Viet chuan (TUYET DOI khong chen chu Han/tieng Trung).
- Chi dua tren du lieu thu thap o tren; phan nao khong co du lieu thi noi ro la chua thu thap duoc.
- KHONG bao gio tu bia link URL; chi trich dan cac link co mat trong du lieu thu thap."""
        final_ans = await ask_llm(client, BRAIN_THINK, "Ban la tro ly tra cuu realtime trung thuc.", ans_p, 0.2)
        elapsed = round(time.perf_counter() - t0, 2)
        result_record = {
            "timestamp": vn_now,
            "question": question,
            "use_skills": selected_for_coder,
            "forged": forged_ok,
            "forged_failed": failed_tools,
            "healed": healed_skills,
            "attempts": attempts_used,
            "success": run_success,
            "elapsed_sec": elapsed,
        }
        append_run_log(result_record)
        cleanup_sandbox()
        print(f"[4/4] Hoan thanh toan bo quy trinh trong {elapsed}s\n")
        print("================ TRA LOI ================")
        print(final_ans)
        return result_record


def main():
    parser = argparse.ArgumentParser(description="AutoEvolve AI - Demand-Driven Self-Evolving Agent")
    parser.add_argument("question", nargs="*", help="Cau hoi can tra cuu")
    parser.add_argument("--health-check", action="store_true", help="Chay pytest kiem tra toan bo kho skills/ va don rac")
    parser.add_argument("--memory-stats", action="store_true", help="Xem trang thai bo nho dai han memory.db")
    parser.add_argument("--dry-run", action="store_true", help="Chi len ke hoach Architect, khong tao tool hay chay code")
    parser.add_argument("--no-forge", action="store_true", help="Chi dung skill san co, khong tao tool moi")
    args = parser.parse_args()

    if args.health_check:
        run_health_check()
        return

    if args.memory_stats:
        show_memory_stats()
        return

    q = " ".join(args.question).strip()
    if not q:
        parser.print_help()
        sys.exit(1)
    asyncio.run(solve_question(q, dry_run=args.dry_run, no_forge=args.no_forge))


if __name__ == "__main__":
    main()
````

````python
# Cell 2: Tạo skill_browser_reader.py và unit test test_skill_browser_reader.py

%%writefile /content/drive/MyDrive/AutoEvolve_AI/skills/skill_browser_reader.py
"""Headless Chromium browser client via Playwright for reading JS-heavy/SPA webpages."""
import re
from playwright.sync_api import sync_playwright


class BrowserReaderClient:
    """Render dynamic JavaScript webpages and extract clean text or CSS elements."""

    def __init__(self, timeout_ms: int = 30000):
        self.timeout_ms = timeout_ms
        self.user_agent = (
            "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
            "(KHTML, like Gecko) Chrome/128.0.0.0 Safari/537.36"
        )
        self.launch_args = [
            "--no-sandbox",
            "--disable-setuid-sandbox",
            "--disable-dev-shm-usage",
            "--disable-gpu",
        ]

    def read_dynamic_page(
        self,
        url: str,
        wait_selector: str | None = None,
        wait_ms: int = 2000,
        max_chars: int = 6000,
    ) -> dict:
        """Return dict keys: url, final_url, title, text, links."""
        if not url or not url.startswith(("http://", "https://")):
            raise ValueError(f"URL khong hop le: {url}")

        try:
            with sync_playwright() as p:
                browser = p.chromium.launch(headless=True, args=self.launch_args)
                try:
                    context = browser.new_context(
                        user_agent=self.user_agent,
                        viewport={"width": 1280, "height": 800},
                    )
                    page = context.new_page()
                    page.set_default_timeout(self.timeout_ms)
                    resp = page.goto(url, wait_until="domcontentloaded", timeout=self.timeout_ms)
                    if resp is not None and resp.status >= 400:
                        raise RuntimeError(f"HTTP {resp.status} khi mo {url}")

                    if wait_selector:
                        page.wait_for_selector(wait_selector, timeout=min(self.timeout_ms, 15000))
                    if wait_ms > 0:
                        page.wait_for_timeout(wait_ms)

                    title = (page.title() or "").strip()
                    final_url = page.url

                    data = page.evaluate(
                        """() => {
                            const clone = document.body ? document.body.cloneNode(true) : null;
                            if (!clone) return { text: "", links: [] };
                            clone.querySelectorAll("script, style, noscript, svg, iframe").forEach(el => el.remove());
                            const text = clone.innerText || "";
                            const links = [];
                            const seen = new Set();
                            for (const a of document.querySelectorAll("a[href]")) {
                                const href = a.href || "";
                                const label = (a.innerText || "").trim().replace(/\\s+/g, " ");
                                if (href.startsWith("http") && label && !seen.has(href)) {
                                    seen.add(href);
                                    links.push({ text: label.slice(0, 120), href });
                                    if (links.length >= 25) break;
                                }
                            }
                            return { text, links };
                        }"""
                    )
                finally:
                    browser.close()
        except Exception as e:
            raise RuntimeError(f"Playwright loi khi doc {url}: {e}") from e

        raw_text = data.get("text", "") if isinstance(data, dict) else ""
        clean_text = re.sub(r"[ \t]+", " ", raw_text)
        clean_text = re.sub(r"\n{3,}", "\n\n", clean_text).strip()
        if not clean_text:
            raise RuntimeError(f"Khong trich xuat duoc noi dung van ban tu {url}")

        return {
            "url": url,
            "final_url": final_url,
            "title": title,
            "text": clean_text[:max_chars],
            "links": data.get("links", []) if isinstance(data, dict) else [],
        }

    def extract_elements(
        self,
        url: str,
        css_selector: str,
        wait_ms: int = 2000,
        limit: int = 20,
    ) -> list[dict]:
        """Return list of dicts with keys: text, href."""
        if not url or not url.startswith(("http://", "https://")):
            raise ValueError(f"URL khong hop le: {url}")
        if not css_selector or not css_selector.strip():
            raise ValueError("css_selector khong duoc rong")

        try:
            with sync_playwright() as p:
                browser = p.chromium.launch(headless=True, args=self.launch_args)
                try:
                    context = browser.new_context(user_agent=self.user_agent)
                    page = context.new_page()
                    page.set_default_timeout(self.timeout_ms)
                    resp = page.goto(url, wait_until="domcontentloaded", timeout=self.timeout_ms)
                    if resp is not None and resp.status >= 400:
                        raise RuntimeError(f"HTTP {resp.status} khi mo {url}")
                    page.wait_for_selector(css_selector, timeout=min(self.timeout_ms, 15000))
                    if wait_ms > 0:
                        page.wait_for_timeout(wait_ms)

                    items = page.evaluate(
                        """([sel, maxItems]) => {
                            const out = [];
                            for (const el of document.querySelectorAll(sel)) {
                                const text = (el.innerText || "").trim().replace(/\\s+/g, " ");
                                const a = el.tagName === "A" ? el : el.querySelector("a[href]");
                                const href = a ? a.href : "";
                                if (text || href) {
                                    out.push({ text: text.slice(0, 300), href });
                                    if (out.length >= maxItems) break;
                                }
                            }
                            return out;
                        }""",
                        [css_selector, limit],
                    )
                finally:
                    browser.close()
        except Exception as e:
            raise RuntimeError(f"Playwright loi khi trich xuat `{css_selector}` tu {url}: {e}") from e

        if not items:
            raise RuntimeError(f"Khong tim thay phan tu nao khop selector `{css_selector}` tai {url}")
        return items
````

````python
# Cell 3: Tạo skill_document_analyzer.py và unit test

%%writefile /content/drive/MyDrive/AutoEvolve_AI/skills/skill_document_analyzer.py
"""Document analyzer for PDF, Excel (.xlsx/.xls), and CSV files from local paths or URLs."""
import io
import os
import re
import json
import httpx
import pandas as pd
from pypdf import PdfReader


class DocumentAnalyzerClient:
    """Read and summarize PDF, Excel, and CSV documents from local storage or HTTP URLs."""

    def __init__(self, timeout: float = 25.0):
        self.timeout = timeout
        self.headers = {
            "User-Agent": (
                "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
                "(KHTML, like Gecko) Chrome/128.0.0.0 Safari/537.36"
            )
        }

    def _load_bytes(self, source: str) -> bytes:
        if not source or not source.strip():
            raise ValueError("Source path/URL khong duoc rong")
        if source.startswith(("http://", "https://")):
            r = httpx.get(
                source,
                headers=self.headers,
                timeout=self.timeout,
                follow_redirects=True,
            )
            if r.status_code != 200:
                raise RuntimeError(f"HTTP {r.status_code} khi tai tai lieu {source}")
            return r.content
        if not os.path.exists(source):
            raise FileNotFoundError(f"Khong tim thay file: {source}")
        with open(source, "rb") as f:
            return f.read()

    def read_pdf(self, source: str, max_pages: int = 10, max_chars: int = 6000) -> dict:
        """Return dict keys: source, total_pages, pages_read, text."""
        raw_bytes = self._load_bytes(source)
        reader = PdfReader(io.BytesIO(raw_bytes))
        total_pages = len(reader.pages)
        if total_pages == 0:
            raise RuntimeError(f"File PDF rong: {source}")

        pages_to_read = min(total_pages, max_pages)
        chunks = []
        for idx in range(pages_to_read):
            page_text = (reader.pages[idx].extract_text() or "").strip()
            if page_text:
                chunks.append(f"[Page {idx + 1}]\n{page_text}")

        full_text = re.sub(r"\n{3,}", "\n\n", "\n\n".join(chunks)).strip()
        if not full_text:
            raise RuntimeError(f"PDF khong co lop text (co the la file scan anh): {source}")

        return {
            "source": source,
            "total_pages": total_pages,
            "pages_read": pages_to_read,
            "text": full_text[:max_chars],
        }

    def analyze_tabular(
        self,
        source: str,
        sheet_name: str | int = 0,
        sample_rows: int = 5,
    ) -> dict:
        """Return dict keys: source, rows, cols, columns, dtypes, null_counts, summary_stats, sample."""
        raw_bytes = self._load_bytes(source)
        lower_src = source.lower().split("?")[0]
        buf = io.BytesIO(raw_bytes)

        if lower_src.endswith((".xlsx", ".xls")):
            df = pd.read_excel(buf, sheet_name=sheet_name)
        else:
            try:
                df = pd.read_csv(buf)
            except UnicodeDecodeError:
                buf.seek(0)
                df = pd.read_csv(buf, encoding="latin-1")

        if df.empty and len(df.columns) == 0:
            raise RuntimeError(f"Bang du lieu rong: {source}")

        num_df = df.select_dtypes(include="number")
        summary_stats = {}
        if not num_df.empty:
            desc = num_df.describe().round(4)
            summary_stats = json.loads(desc.to_json())

        sample = json.loads(df.head(sample_rows).to_json(orient="records", date_format="iso"))
        return {
            "source": source,
            "rows": int(df.shape[0]),
            "cols": int(df.shape[1]),
            "columns": [str(c) for c in df.columns],
            "dtypes": {str(k): str(v) for k, v in df.dtypes.items()},
            "null_counts": {str(k): int(v) for k, v in df.isna().sum().items()},
            "summary_stats": summary_stats,
            "sample": sample,
        }
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/skills/test_skill_document_analyzer.py
from unittest.mock import MagicMock, patch
import pandas as pd
import pytest
from skill_document_analyzer import DocumentAnalyzerClient


def test_analyze_csv_and_excel(tmp_path):
    df = pd.DataFrame({"product": ["A", "B", "C"], "revenue": [100, 250, 150]})
    csv_path = tmp_path / "sales.csv"
    xlsx_path = tmp_path / "sales.xlsx"
    df.to_csv(csv_path, index=False)
    df.to_excel(xlsx_path, index=False)

    client = DocumentAnalyzerClient()
    res_csv = client.analyze_tabular(str(csv_path), sample_rows=2)
    assert res_csv["rows"] == 3
    assert res_csv["cols"] == 2
    assert "revenue" in res_csv["summary_stats"]
    assert len(res_csv["sample"]) == 2

    res_xlsx = client.analyze_tabular(str(xlsx_path), sample_rows=2)
    assert res_xlsx["rows"] == 3
    assert res_xlsx["columns"] == ["product", "revenue"]


def test_read_pdf_from_url():
    fake_page = MagicMock()
    fake_page.extract_text.return_value = "Self-Evolving AI Agents PDF Content"
    fake_reader = MagicMock(pages=[fake_page])
    mock_resp = MagicMock(status_code=200, content=b"%PDF-1.4 fake")

    with patch("httpx.get", return_value=mock_resp), patch(
        "skill_document_analyzer.PdfReader", return_value=fake_reader
    ):
        client = DocumentAnalyzerClient()
        res = client.read_pdf("https://arxiv.org/pdf/2508.07407.pdf", max_pages=3)
        assert res["total_pages"] == 1
        assert "Self-Evolving AI Agents" in res["text"]


def test_missing_file_raises_error():
    client = DocumentAnalyzerClient()
    with pytest.raises(Exception):
        client.analyze_tabular("/nonexistent/path/data.csv")
````

````python
%%writefile /content/drive/MyDrive/AutoEvolve_AI/skills/test_skill_browser_reader.py
from unittest.mock import MagicMock, patch
import pytest
from skill_browser_reader import BrowserReaderClient


def _mock_playwright(status=200, title="Dynamic App", eval_result=None):
    mock_page = MagicMock()
    mock_page.goto.return_value = MagicMock(status=status)
    mock_page.title.return_value = title
    mock_page.url = "https://example.org/app"
    mock_page.evaluate.return_value = (
        eval_result
        if eval_result is not None
        else {"text": "Rendered JS content", "links": [{"text": "Docs", "href": "https://example.org/docs"}]}
    )

    mock_context = MagicMock()
    mock_context.new_page.return_value = mock_page

    mock_browser = MagicMock()
    mock_browser.new_context.return_value = mock_context

    mock_p = MagicMock()
    mock_p.chromium.launch.return_value = mock_browser

    mock_cm = MagicMock()
    mock_cm.__enter__.return_value = mock_p
    mock_cm.__exit__.return_value = None
    return mock_cm


def test_read_dynamic_page_happy_path():
    with patch("skill_browser_reader.sync_playwright", return_value=_mock_playwright()):
        client = BrowserReaderClient()
        res = client.read_dynamic_page("https://example.org/app", wait_ms=0)
        assert res["title"] == "Dynamic App"
        assert "Rendered JS content" in res["text"]
        assert len(res["links"]) == 1


def test_read_dynamic_page_http_error():
    with patch("skill_browser_reader.sync_playwright", return_value=_mock_playwright(status=403)):
        client = BrowserReaderClient()
        with pytest.raises(Exception):
            client.read_dynamic_page("https://example.org/blocked", wait_ms=0)


def test_read_dynamic_page_invalid_url():
    client = BrowserReaderClient()
    with pytest.raises(Exception):
        client.read_dynamic_page("not-a-url")
````

````python
%%bash
python3 - <<'EOF'
import os
import re
import subprocess

path = "/content/drive/MyDrive/AutoEvolve_AI/skills/skill_rss_feed_reader.py"
assert os.path.exists(path), f"Khong tim thay {path}"

src = open(path, encoding="utf-8").read()

helper = '''
from datetime import datetime, timezone
from email.utils import parsedate_to_datetime

def _parse_rss_datetime(val):
    if not val:
        return None
    if isinstance(val, datetime):
        return val.astimezone(timezone.utc) if val.tzinfo else val.replace(tzinfo=timezone.utc)
    s = str(val).strip()
    try:
        dt = parsedate_to_datetime(s)
        return dt.astimezone(timezone.utc) if dt.tzinfo else dt.replace(tzinfo=timezone.utc)
    except Exception:
        pass
    for fmt in (
        "%a, %d %b %Y %H:%M:%S %z",
        "%a, %d %b %Y %H:%M:%S %Z",
        "%a, %d %b %Y %H:%M %z",
        "%Y-%m-%dT%H:%M:%S%z",
        "%Y-%m-%dT%H:%M:%SZ",
        "%Y-%m-%d %H:%M:%S",
    ):
        try:
            dt = datetime.strptime(s, fmt)
            return dt.astimezone(timezone.utc) if dt.tzinfo else dt.replace(tzinfo=timezone.utc)
        except Exception:
            continue
    return s
'''

# Thay cac loi goi datetime.strptime cu neu dung %Z cho RSS pubDate
patched = re.sub(
    r"datetime\.strptime\(([^,]+),\s*['\"]%a, %d %b %Y %H:%M:%S %Z['\"]\)",
    r"_parse_rss_datetime(\1)",
    src,
)

if "_parse_rss_datetime" not in patched:
    # Chen ham helper ngay sau cac dong import
    lines = patched.splitlines(keepends=True)
    insert_idx = 0
    for i, ln in enumerate(lines):
        if ln.startswith(("import ", "from ")):
            insert_idx = i + 1
    lines.insert(insert_idx, "\n" + helper.lstrip("\n") + "\n")
    patched = "".join(lines)

if patched != src:
    open(path, "w", encoding="utf-8").write(patched)
    print("Da va _parse_rss_datetime trong skill_rss_feed_reader.py")
else:
    print("Da chen _parse_rss_datetime (hoac khong co mau %Z cu)")

res = subprocess.run(
    ["pytest", "-q", "/content/drive/MyDrive/AutoEvolve_AI/skills/test_skill_rss_feed_reader.py"],
    capture_output=True,
    text=True,
)
print(res.stdout.strip())
if res.returncode != 0:
    print(res.stderr.strip())
EOF
````

````python
%%bash
python3 - <<'EOF'
import subprocess

skill_p = "/content/drive/MyDrive/AutoEvolve_AI/skills/skill_rss_feed_reader.py"
test_p = "/content/drive/MyDrive/AutoEvolve_AI/skills/test_skill_rss_feed_reader.py"

code = '''import httpx
from xml.etree import ElementTree as ET
from datetime import datetime, timezone
from email.utils import parsedate_to_datetime


def _parse_rss_datetime(val):
    if not val:
        return None
    if isinstance(val, datetime):
        dt = val if val.tzinfo else val.replace(tzinfo=timezone.utc)
        return dt.astimezone(timezone.utc).strftime("%Y-%m-%d %H:%M:%S %z")
    s = str(val).strip()
    try:
        dt = parsedate_to_datetime(s)
        dt = dt if dt.tzinfo else dt.replace(tzinfo=timezone.utc)
        return dt.astimezone(timezone.utc).strftime("%Y-%m-%d %H:%M:%S %z")
    except Exception:
        pass
    for fmt in (
        "%a, %d %b %Y %H:%M:%S %z",
        "%a, %d %b %Y %H:%M:%S %Z",
        "%a, %d %b %Y %H:%M:%S",
        "%Y-%m-%dT%H:%M:%S%z",
        "%Y-%m-%dT%H:%M:%SZ",
        "%Y-%m-%d %H:%M:%S",
    ):
        try:
            dt = datetime.strptime(s, fmt)
            dt = dt if dt.tzinfo else dt.replace(tzinfo=timezone.utc)
            return dt.astimezone(timezone.utc).strftime("%Y-%m-%d %H:%M:%S %z")
        except Exception:
            continue
    return s


class RSSFeedReaderClient:
    def __init__(self, base_url):
        self.base_url = base_url

    def get_latest_articles(self, limit=5):
        response = httpx.get(
            self.base_url,
            timeout=15.0,
            follow_redirects=True,
            headers={"User-Agent": "AutoEvolveBot/1.0 (https://example.org)"},
        )
        if response.status_code != 200:
            raise RuntimeError(f"Failed to fetch RSS feed: HTTP {response.status_code}, {response.text}")

        try:
            root = ET.fromstring(response.content)
            articles = []
            for item in root.findall(".//item")[:limit]:
                title_el = item.find("title")
                link_el = item.find("link")
                author_el = item.find("author")
                pub_el = item.find("pubDate")

                title = title_el.text if title_el is not None else None
                link = link_el.text if link_el is not None else None
                author = author_el.text if author_el is not None else None
                published_time = pub_el.text if pub_el is not None else None
                if published_time:
                    published_time = _parse_rss_datetime(published_time)

                articles.append({
                    "title": title,
                    "link": link,
                    "author": author,
                    "published_time": published_time,
                })
            return articles
        except ET.ParseError as e:
            raise RuntimeError(f"Failed to parse RSS feed: {str(e)}")
        except Exception as e:
            raise RuntimeError(f"An unexpected error occurred: {str(e)}")
'''

with open(skill_p, "w", encoding="utf-8") as f:
    f.write(code)

print("Da cap nhat xong skill_rss_feed_reader.py")
r = subprocess.run(["pytest", "-v", test_p], capture_output=True, text=True)
print(r.stdout)
if r.stderr:
    print(r.stderr)
EOF

python3 -u /content/drive/MyDrive/AutoEvolve_AI/live_evolve_agent.py --health-check
````

````python
%%bash
cat <<'EOF' > /content/drive/MyDrive/AutoEvolve_AI/skill_contract.py
from __future__ import annotations

import functools
import inspect
import json
import re
from datetime import datetime, timezone
from typing import Any, Callable


TRANSIENT_PATTERNS = (
    r"\b403\b",
    r"\b408\b",
    r"\b425\b",
    r"\b429\b",
    r"\b500\b",
    r"\b502\b",
    r"\b503\b",
    r"\b504\b",
    r"timeout",
    r"timed out",
    r"readtimeout",
    r"connecttimeout",
    r"connection reset",
    r"temporarily unavailable",
    r"too many requests",
    r"rate limit",
    r"service unavailable",
    r"try again later",
)


def classify_error_kind(err_text: str, exc: BaseException | None = None) -> str:
    if not err_text and exc is None:
        return "none"
    combined = f"{type(exc).__name__ if exc else ''} {err_text}".lower()
    if isinstance(exc, (TimeoutError, ConnectionError)) or any(
        re.search(pat, combined) for pat in TRANSIENT_PATTERNS
    ):
        return "transient"
    if isinstance(exc, (ValueError, TypeError, KeyError, json.JSONDecodeError)):
        return "validation"
    return "permanent"


def make_envelope(
    ok: bool,
    data: Any = None,
    error: str | None = None,
    error_kind: str | None = None,
    status: str | None = None,
    meta: dict[str, Any] | None = None,
) -> dict[str, Any]:
    resolved_status = status or ("ok" if ok else "error")
    resolved_kind = "none" if ok else (error_kind or classify_error_kind(error or ""))
    return {
        "ok": bool(ok),
        "status": resolved_status,
        "data": data,
        "error": None if ok else (error or "Unknown error"),
        "error_kind": resolved_kind,
        "timestamp_utc": datetime.now(timezone.utc).isoformat(),
        "meta": meta or {},
    }


def normalize_skill_output(raw: Any, default_meta: dict[str, Any] | None = None) -> dict[str, Any]:
    """Chuẩn hóa mọi kiểu trả về (dict, str, list, tuple, scalar) về chuẩn envelope thống nhất."""
    if isinstance(raw, dict) and {"ok", "status", "data", "error"}.issubset(raw.keys()):
        out = dict(raw)
        out.setdefault("error_kind", "none" if out.get("ok") else classify_error_kind(str(out.get("error") or "")))
        out.setdefault("timestamp_utc", datetime.now(timezone.utc).isoformat())
        out.setdefault("meta", {})
        if default_meta:
            out["meta"] = {**default_meta, **out["meta"]}
        return out

    if isinstance(raw, str):
        stripped = raw.strip()
        # Nếu chuỗi là JSON hợp lệ có trường ok/error
        if stripped.startswith("{") and stripped.endswith("}"):
            try:
                parsed = json.loads(stripped)
                if isinstance(parsed, dict) and "ok" in parsed:
                    return normalize_skill_output(parsed, default_meta=default_meta)
            except Exception:
                pass

        lower = stripped.lower()
        looks_like_error = (
            not stripped
            or lower.startswith(("error:", "failed:", "traceback", "exception:"))
            or "http 403" in lower
            or "http 404" in lower
            or "http 429" in lower
            or "http 500" in lower
            or "http 502" in lower
            or "http 503" in lower
            or "not found" in lower
            or "n/a" == lower
        )
        if looks_like_error:
            msg = stripped or "Empty output from skill"
            return make_envelope(
                ok=False,
                data=None,
                error=msg,
                error_kind=classify_error_kind(msg),
                status="error",
                meta=default_meta,
            )
        return make_envelope(ok=True, data=raw, error=None, error_kind="none", status="ok", meta=default_meta)

    if raw is None:
        return make_envelope(
            ok=False,
            data=None,
            error="Skill returned None",
            error_kind="validation",
            status="empty",
            meta=default_meta,
        )

    if isinstance(raw, dict):
        if raw.get("error"):
            err_msg = str(raw["error"])
            return make_envelope(
                ok=False,
                data=raw.get("data"),
                error=err_msg,
                error_kind=classify_error_kind(err_msg),
                status="error",
                meta=default_meta,
            )
        return make_envelope(ok=True, data=raw, error=None, error_kind="none", status="ok", meta=default_meta)

    return make_envelope(ok=True, data=raw, error=None, error_kind="none", status="ok", meta=default_meta)


def wrap_callable_with_contract(
    fn: Callable[..., Any],
    skill_name: str = "",
    method_name: str = "",
) -> Callable[..., dict[str, Any]]:
    """Bọc 1 hàm/phương thức skill để luôn trả về dict chuẩn và không bao giờ ném exception ra ngoài."""
    sig = inspect.signature(fn)
    clean_params = [p for name, p in sig.parameters.items() if name != "self"]
    new_sig = sig.replace(parameters=clean_params)

    @functools.wraps(fn)
    def _guarded(*args: Any, **kwargs: Any) -> dict[str, Any]:
        meta = {"skill": skill_name, "method": method_name or fn.__name__}
        try:
            bound = new_sig.bind(*args, **kwargs)
            bound.apply_defaults()
            raw = fn(*bound.args, **bound.kwargs)
            return normalize_skill_output(raw, default_meta=meta)
        except Exception as exc:
            err_msg = f"{type(exc).__name__}: {exc}"
            return make_envelope(
                ok=False,
                data=None,
                error=err_msg,
                error_kind=classify_error_kind(err_msg, exc=exc),
                status="error",
                meta=meta,
            )

    _guarded.__signature__ = new_sig
    return _guarded
EOF
python3 -c "import sys; sys.path.insert(0, '/content/drive/MyDrive/AutoEvolve_AI'); import skill_contract; print('skill_contract OK')"
````

````python
%%bash
cat <<'EOF' > /content/drive/MyDrive/AutoEvolve_AI/memory_store.py
from __future__ import annotations

import json
import math
import os
import re
import sqlite3
from datetime import datetime, timezone
from typing import Any

import httpx


OLLAMA_BASE = os.environ.get("OLLAMA_BASE", "http://localhost:11434")
DEFAULT_DB_PATH = "/content/drive/MyDrive/AutoEvolve_AI/memory.db"
EMBED_MODEL = os.environ.get("OLLAMA_EMBED_MODEL", "nomic-embed-text")


def _now_vn() -> str:
    return datetime.now(timezone.utc).astimezone().strftime("%Y-%m-%d %H:%M:%S %z")


def _tokenize(text: str) -> list[str]:
    return re.findall(r"[a-z0-9_À-ỹ]+", (text or "").lower())


def _cosine(v1: list[float], v2: list[float]) -> float:
    if not v1 or not v2 or len(v1) != len(v2):
        return 0.0
    dot = sum(a * b for a, b in zip(v1, v2))
    n1 = math.sqrt(sum(a * a for a in v1))
    n2 = math.sqrt(sum(b * b for b in v2))
    if n1 == 0.0 or n2 == 0.0:
        return 0.0
    return dot / (n1 * n2)


def _token_similarity(q: str, doc: str) -> float:
    tq = set(_tokenize(q))
    td = set(_tokenize(doc))
    if not tq or not td:
        return 0.0
    return len(tq & td) / math.sqrt(len(tq) * len(td))


def _infer_error_kind(outcome: str, lesson: str, explicit_kind: str | None = None) -> str:
    if explicit_kind in {"none", "transient", "permanent"}:
        return explicit_kind
    if (outcome or "").upper() == "PASS":
        return "none"
    low = (lesson or "").lower()
    if any(
        k in low
        for k in (
            "403", "408", "425", "429", "500", "502", "503", "504",
            "timeout", "timed out", "readtimeout", "rate limit", "temporarily", "connection",
        )
    ):
        return "transient"
    return "permanent"


class MemoryStore:
    def __init__(self, db_path: str = DEFAULT_DB_PATH, embed_model: str = EMBED_MODEL):
        self.db_path = db_path
        self.embed_model = embed_model
        os.makedirs(os.path.dirname(os.path.abspath(self.db_path)), exist_ok=True)
        self.conn = sqlite3.connect(self.db_path, check_same_thread=False)
        self.conn.row_factory = sqlite3.Row
        self._init_schema()
        self._migrate_schema()

    def _init_schema(self) -> None:
        self.conn.executescript(
            """
            CREATE TABLE IF NOT EXISTS episodes (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                timestamp TEXT NOT NULL,
                question TEXT NOT NULL,
                skills_used TEXT NOT NULL,
                outcome TEXT NOT NULL,
                lesson TEXT NOT NULL,
                error_kind TEXT DEFAULT 'none',
                confidence REAL DEFAULT 0.6,
                verified_count INTEGER DEFAULT 1,
                embedding TEXT
            );

            CREATE TABLE IF NOT EXISTS blacklist (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                pattern TEXT UNIQUE NOT NULL,
                reason TEXT NOT NULL,
                error_kind TEXT DEFAULT 'permanent',
                fail_count INTEGER DEFAULT 1,
                updated_at TEXT NOT NULL
            );

            CREATE TABLE IF NOT EXISTS explorations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                topic TEXT UNIQUE NOT NULL,
                skill_name TEXT,
                status TEXT NOT NULL,
                notes TEXT,
                updated_at TEXT NOT NULL
            );

            CREATE TABLE IF NOT EXISTS skill_vectors (
                skill_name TEXT PRIMARY KEY,
                summary TEXT NOT NULL,
                embedding TEXT,
                updated_at TEXT NOT NULL
            );

            CREATE TABLE IF NOT EXISTS recipes (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                recipe_name TEXT UNIQUE NOT NULL,
                intent_summary TEXT NOT NULL,
                steps_json TEXT NOT NULL,
                success_count INTEGER DEFAULT 1,
                fail_count INTEGER DEFAULT 0,
                confidence REAL DEFAULT 0.7,
                updated_at TEXT NOT NULL
            );
            """
        )
        self.conn.commit()

    def _migrate_schema(self) -> None:
        cols = {r["name"] for r in self.conn.execute("PRAGMA table_info(episodes)").fetchall()}
        if "error_kind" not in cols:
            self.conn.execute("ALTER TABLE episodes ADD COLUMN error_kind TEXT DEFAULT 'none'")
        if "confidence" not in cols:
            self.conn.execute("ALTER TABLE episodes ADD COLUMN confidence REAL DEFAULT 0.6")
        if "verified_count" not in cols:
            self.conn.execute("ALTER TABLE episodes ADD COLUMN verified_count INTEGER DEFAULT 1")

        bl_cols = {r["name"] for r in self.conn.execute("PRAGMA table_info(blacklist)").fetchall()}
        if "error_kind" not in bl_cols:
            self.conn.execute("ALTER TABLE blacklist ADD COLUMN error_kind TEXT DEFAULT 'permanent'")
        self.conn.commit()

    def embed(self, text: str) -> list[float] | None:
        clean = (text or "").strip()
        if not clean:
            return None
        try:
            with httpx.Client(timeout=15.0) as client:
                r = client.post(
                    f"{OLLAMA_BASE}/api/embeddings",
                    json={"model": self.embed_model, "prompt": clean[:2000]},
                )
                if r.status_code == 200:
                    vec = r.json().get("embedding")
                    if isinstance(vec, list) and vec:
                        return [float(x) for x in vec]
        except Exception:
            pass
        return None

    def add_episode(
        self,
        question: str,
        skills_used: list[str],
        outcome: str,
        lesson: str,
        error_kind: str | None = None,
    ) -> int:
        norm_outcome = (outcome or "FAIL").upper()
        ekind = _infer_error_kind(norm_outcome, lesson, error_kind)
        ts = _now_vn()
        emb = self.embed(f"{question}\n{lesson}")
        emb_json = json.dumps(emb) if emb else None

        # Nếu có bài học PASS rất giống nhau, tăng độ tin cậy thay vì làm loãng bộ nhớ
        if norm_outcome == "PASS":
            existing = self.conn.execute(
                "SELECT id, confidence, verified_count FROM episodes WHERE outcome='PASS' AND lesson=? LIMIT 1",
                (lesson,),
            ).fetchone()
            if existing:
                new_cnt = int(existing["verified_count"] or 1) + 1
                new_conf = min(0.99, round(float(existing["confidence"] or 0.6) + 0.08 * (1.0 - float(existing["confidence"] or 0.6)), 4))
                self.conn.execute(
                    "UPDATE episodes SET timestamp=?, confidence=?, verified_count=?, embedding=COALESCE(?, embedding) WHERE id=?",
                    (ts, new_conf, new_cnt, emb_json, existing["id"]),
                )
                self.conn.commit()
                return int(existing["id"])

        init_conf = 0.72 if norm_outcome == "PASS" else (0.25 if ekind == "transient" else 0.55)
        cur = self.conn.execute(
            """
            INSERT INTO episodes (timestamp, question, skills_used, outcome, lesson, error_kind, confidence, verified_count, embedding)
            VALUES (?, ?, ?, ?, ?, ?, ?, 1, ?)
            """,
            (
                ts,
                question,
                json.dumps(skills_used, ensure_ascii=False),
                norm_outcome,
                lesson,
                ekind,
                init_conf,
                emb_json,
            ),
        )
        self.conn.commit()
        return int(cur.lastrowid)

    def decay_transient_lessons(self, decay_factor: float = 0.85) -> int:
        """Giảm dần độ tin cậy của bài học lỗi tạm thời để chúng tự nhạt dần theo thời gian."""
        rows = self.conn.execute(
            "SELECT id, confidence FROM episodes WHERE error_kind='transient' AND confidence > 0.05"
        ).fetchall()
        updated = 0
        for r in rows:
            nxt = max(0.05, round(float(r["confidence"]) * decay_factor, 4))
            self.conn.execute("UPDATE episodes SET confidence=? WHERE id=?", (nxt, r["id"]))
            updated += 1
        self.conn.commit()
        return updated

    def search_episodes(self, query: str, top_k: int = 4, min_confidence: float = 0.2) -> list[dict[str, Any]]:
        q_emb = self.embed(query)
        rows = self.conn.execute(
            "SELECT id, timestamp, question, skills_used, outcome, lesson, error_kind, confidence, verified_count, embedding FROM episodes"
        ).fetchall()
        scored: list[tuple[float, dict[str, Any]]] = []
        for r in rows:
            conf = float(r["confidence"] if r["confidence"] is not None else 0.5)
            if r["outcome"] == "FAIL" and r["error_kind"] == "transient" and conf < min_confidence:
                continue
            doc = f"{r['question']} {r['lesson']}"
            lex = _token_similarity(query, doc)
            sem = 0.0
            if q_emb and r["embedding"]:
                try:
                    sem = _cosine(q_emb, json.loads(r["embedding"]))
                except Exception:
                    sem = 0.0
            base_score = (0.65 * sem + 0.35 * lex) if q_emb else lex
            # Ưu tiên bài học PASS và độ tin cậy cao; hạ trọng số bài lỗi tạm thời
            trust_mult = (0.6 + 0.6 * conf) if r["outcome"] == "PASS" else (0.35 if r["error_kind"] == "transient" else 0.75)
            final_score = base_score * trust_mult
            scored.append(
                (
                    final_score,
                    {
                        "id": r["id"],
                        "timestamp": r["timestamp"],
                        "question": r["question"],
                        "skills_used": json.loads(r["skills_used"] or "[]"),
                        "outcome": r["outcome"],
                        "lesson": r["lesson"],
                        "error_kind": r["error_kind"],
                        "confidence": conf,
                        "verified_count": r["verified_count"],
                    },
                )
            )
        scored.sort(key=lambda x: x[0], reverse=True)
        return [item for _, item in scored[:top_k]]

    def add_blacklist(self, pattern: str, reason: str, error_kind: str | None = None) -> None:
        pat = (pattern or "").strip()
        if not pat:
            return
        ekind = _infer_error_kind("FAIL", reason, error_kind)
        ts = _now_vn()
        existing = self.conn.execute(
            "SELECT id, fail_count, error_kind FROM blacklist WHERE pattern=?",
            (pat,),
        ).fetchone()
        if existing:
            self.conn.execute(
                "UPDATE blacklist SET reason=?, error_kind=?, fail_count=?, updated_at=? WHERE id=?",
                (reason, ekind, int(existing["fail_count"] or 0) + 1, ts, existing["id"]),
            )
        else:
            self.conn.execute(
                "INSERT INTO blacklist (pattern, reason, error_kind, fail_count, updated_at) VALUES (?, ?, ?, 1, ?)",
                (pat, reason, ekind, ts),
            )
        self.conn.commit()

    def is_blacklisted(self, text: str, allow_transient_retry: bool = True) -> list[dict[str, Any]]:
        low = (text or "").lower()
        hits = []
        for r in self.conn.execute("SELECT pattern, reason, error_kind, fail_count FROM blacklist").fetchall():
            pat = (r["pattern"] or "").strip()
            if not pat or pat.lower() not in low:
                continue
            ekind = r["error_kind"] or "permanent"
            fc = int(r["fail_count"] or 1)
            # Lỗi tạm thời chỉ chặn khi fail lặp lại >= 3 lần và không bật chế độ thử lại
            if ekind == "transient" and allow_transient_retry and fc < 3:
                continue
            hits.append(dict(r))
        return hits

    def record_exploration(self, topic: str, skill_name: str | None, status: str, notes: str = "") -> None:
        ts = _now_vn()
        self.conn.execute(
            """
            INSERT INTO explorations (topic, skill_name, status, notes, updated_at)
            VALUES (?, ?, ?, ?, ?)
            ON CONFLICT(topic) DO UPDATE SET
                skill_name=COALESCE(excluded.skill_name, explorations.skill_name),
                status=excluded.status,
                notes=excluded.notes,
                updated_at=excluded.updated_at
            """,
            (topic, skill_name, status, notes, ts),
        )
        self.conn.commit()

    def upsert_skill_vector(self, skill_name: str, summary: str) -> None:
        ts = _now_vn()
        emb = self.embed(f"{skill_name}: {summary}")
        emb_json = json.dumps(emb) if emb else None
        self.conn.execute(
            """
            INSERT INTO skill_vectors (skill_name, summary, embedding, updated_at)
            VALUES (?, ?, ?, ?)
            ON CONFLICT(skill_name) DO UPDATE SET
                summary=excluded.summary,
                embedding=COALESCE(excluded.embedding, skill_vectors.embedding),
                updated_at=excluded.updated_at
            """,
            (skill_name, summary, emb_json, ts),
        )
        self.conn.commit()

    def save_recipe(self, recipe_name: str, intent_summary: str, steps: list[dict[str, Any]], ok: bool = True) -> None:
        """Lưu hoặc cập nhật công thức ghép chuỗi nhiều skill (Skill Composition)."""
        ts = _now_vn()
        steps_json = json.dumps(steps, ensure_ascii=False)
        existing = self.conn.execute(
            "SELECT id, success_count, fail_count, confidence FROM recipes WHERE recipe_name=?",
            (recipe_name,),
        ).fetchone()
        if existing:
            sc = int(existing["success_count"] or 0) + (1 if ok else 0)
            fc = int(existing["fail_count"] or 0) + (0 if ok else 1)
            conf = float(existing["confidence"] or 0.7)
            conf = min(0.99, round(conf + 0.07, 4)) if ok else max(0.1, round(conf - 0.15, 4))
            self.conn.execute(
                """
                UPDATE recipes
                SET intent_summary=?, steps_json=?, success_count=?, fail_count=?, confidence=?, updated_at=?
                WHERE id=?
                """,
                (intent_summary, steps_json, sc, fc, conf, ts, existing["id"]),
            )
        else:
            self.conn.execute(
                """
                INSERT INTO recipes (recipe_name, intent_summary, steps_json, success_count, fail_count, confidence, updated_at)
                VALUES (?, ?, ?, ?, ?, ?, ?)
                """,
                (
                    recipe_name,
                    intent_summary,
                    steps_json,
                    1 if ok else 0,
                    0 if ok else 1,
                    0.75 if ok else 0.3,
                    ts,
                ),
            )
        self.conn.commit()

    def find_recipes(self, query: str, top_k: int = 3) -> list[dict[str, Any]]:
        rows = self.conn.execute(
            "SELECT recipe_name, intent_summary, steps_json, success_count, fail_count, confidence FROM recipes WHERE confidence >= 0.35"
        ).fetchall()
        scored = []
        for r in rows:
            sim = _token_similarity(query, f"{r['recipe_name']} {r['intent_summary']} {r['steps_json']}")
            score = sim * float(r["confidence"] or 0.7)
            if score > 0.05:
                scored.append(
                    (
                        score,
                        {
                            "recipe_name": r["recipe_name"],
                            "intent_summary": r["intent_summary"],
                            "steps": json.loads(r["steps_json"] or "[]"),
                            "success_count": r["success_count"],
                            "confidence": r["confidence"],
                        },
                    )
                )
        scored.sort(key=lambda x: x[0], reverse=True)
        return [x[1] for x in scored[:top_k]]

    def stats(self) -> dict[str, Any]:
        ep_cnt = self.conn.execute("SELECT COUNT(*) FROM episodes").fetchone()[0]
        pass_cnt = self.conn.execute("SELECT COUNT(*) FROM episodes WHERE outcome='PASS'").fetchone()[0]
        trans_cnt = self.conn.execute("SELECT COUNT(*) FROM episodes WHERE error_kind='transient'").fetchone()[0]
        bl_cnt = self.conn.execute("SELECT COUNT(*) FROM blacklist").fetchone()[0]
        ex_cnt = self.conn.execute("SELECT COUNT(*) FROM explorations").fetchone()[0]
        sv_cnt = self.conn.execute("SELECT COUNT(*) FROM skill_vectors").fetchone()[0]
        rc_cnt = self.conn.execute("SELECT COUNT(*) FROM recipes").fetchone()[0]
        recent = [
            dict(r)
            for r in self.conn.execute(
                "SELECT timestamp, outcome, error_kind, confidence, verified_count, lesson FROM episodes ORDER BY id DESC LIMIT 5"
            ).fetchall()
        ]
        blacklist = [
            dict(r)
            for r in self.conn.execute(
                "SELECT pattern, reason, error_kind, fail_count FROM blacklist ORDER BY fail_count DESC"
            ).fetchall()
        ]
        recipes = [
            dict(r)
            for r in self.conn.execute(
                "SELECT recipe_name, intent_summary, success_count, confidence FROM recipes ORDER BY confidence DESC, success_count DESC LIMIT 5"
            ).fetchall()
        ]
        return {
            "db_path": self.db_path,
            "embed_model": self.embed_model,
            "episodes": ep_cnt,
            "pass_episodes": pass_cnt,
            "transient_episodes": trans_cnt,
            "blacklist": bl_cnt,
            "explorations": ex_cnt,
            "skill_vectors": sv_cnt,
            "recipes": rc_cnt,
            "recent_lessons": recent,
            "blacklist_items": blacklist,
            "top_recipes": recipes,
        }

    def close(self) -> None:
        try:
            self.conn.close()
        except Exception:
            pass
EOF
python3 -c "import sys; sys.path.insert(0, '/content/drive/MyDrive/AutoEvolve_AI'); import memory_store; print('memory_store OK')"
````

````python
%%bash
set -e
export BASE="/content/drive/MyDrive/AutoEvolve_AI"

# 1. Bo sung 2 method con thieu (get_stats, format_context_for_prompt) vao MemoryStore
PYTHONPATH="$BASE" python3 - <<'EOF'
import os

fp = os.path.join(os.environ["BASE"], "memory_store.py")
src = open(fp, encoding="utf-8").read()

patch = '''
    def get_stats(self) -> dict[str, Any]:
        return self.stats()

    def format_context_for_prompt(self, query: str = "", top_k: int = 4, **kwargs: Any) -> str:
        lines: list[str] = []
        eps = self.search_episodes(query, top_k=top_k) if query else self.get_recent_episodes(limit=top_k)
        if eps:
            lines.append("### Kinh nghiem lien quan (Memory Episodes):")
            for e in eps:
                lines.append(
                    f"- [{e.get('outcome')}|conf={e.get('confidence', 0.5)}] Q: {str(e.get('question', ''))[:100]} -> {str(e.get('lesson', ''))[:200]}"
                )
        bl = self.get_blacklist()
        if bl:
            lines.append("### Blacklist can tranh:")
            for b in bl[:8]:
                lines.append(
                    f"- {b.get('pattern')} ({b.get('error_kind')}, fails={b.get('fail_count')}): {str(b.get('reason', ''))[:120]}"
                )
        if query:
            recs = self.find_recipes(query, top_k=2)
            if recs:
                lines.append("### Recipe da kiem chung:")
                for r in recs:
                    lines.append(f"- {r.get('recipe_name')}: {r.get('intent_summary')}")
        return "\\n".join(lines)
'''

if "def format_context_for_prompt(" not in src:
    anchor = "    def close(self) -> None:"
    assert anchor in src, "Khong tim thay vi tri chen trong memory_store.py"
    src = src.replace(anchor, patch.lstrip("\n") + "\n" + anchor, 1)
    with open(fp, "w", encoding="utf-8") as f:
        f.write(src)
    print("Da bo sung get_stats va format_context_for_prompt vao memory_store.py")
EOF

# 2. Kiem tra 100% method mem.* + dem chinh xac so skill hien co
PYTHONPATH="$BASE" python3 - <<'EOF'
import glob, json, os, re
from memory_store import MemoryStore

base = os.environ["BASE"]
mem = MemoryStore()
called = set()
for fn in ("auto_researcher.py", "live_evolve_agent.py"):
    fp = f"{base}/{fn}"
    if os.path.exists(fp):
        called.update(re.findall(r"\bmem\.([a-zA-Z_][a-zA-Z0-9_]*)", open(fp, encoding="utf-8").read()))
missing = [m for m in sorted(called) if not hasattr(mem, m)]
print("Method con thieu:", missing)
assert not missing, f"Con thieu method: {missing}"
mem.close()

skill_files = sorted(
    os.path.basename(p)[:-3]
    for p in glob.glob(f"{base}/skills/*.py")
    if not os.path.basename(p).startswith("_")
)
meta_path = f"{base}/skills_meta.json"
meta_keys = sorted(json.load(open(meta_path, encoding="utf-8")).keys()) if os.path.exists(meta_path) else []
print(f"So file skill trong skills/: {len(skill_files)}")
print(f"So skill trong skills_meta.json: {len(meta_keys)}")
print("Danh sach skill:", skill_files)
EOF

````
